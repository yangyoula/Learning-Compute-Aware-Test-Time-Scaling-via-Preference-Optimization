```python
# =========================================================
# FAST A100 (≈ 2 hours) one-cell Colab script for POTTS
# =========================================================
# Runs:
# - Single / TTA-4 / TTA-8
# - Random policy
# - Entropy-threshold
# - Margin-threshold
# - POTTS
# - Feature ablation: entropy_only vs full
# - Preference ablation: acc_only vs acc_cost
# - Trade-off curve: 4 target compute points
#
# Skips for speed:
# - CIFAR-100-C
# - DeiT-S/16
# - large ablation grid
#
# Output CSVs:
# /content/potts_results/main_results.csv
# /content/potts_results/ablation_results.csv
# /content/potts_results/tradeoff_curve.csv
# /content/potts_results/paper_table_main.csv
# =========================================================

!pip -q install timm pandas scikit-learn tqdm

import os
import random
import warnings
from typing import Dict, List, Tuple, Optional

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset

import torchvision
import torchvision.transforms as T
from torchvision.datasets import CIFAR100

import timm

warnings.filterwarnings("ignore")

# =========================
# Config
# =========================
SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
NUM_WORKERS = 2
BATCH_SIZE = 256

TRAIN_POLICY_SUBSET = 12000
VAL_SUBSET = 1000
TEST_SUBSET = 2000

IMAGE_SIZE = 224
NUM_CLASSES = 100

BACKBONE_NAME = "resnet50"     # fast A100 version only uses resnet50
USE_PRETRAINED_HEAD = True     # trains a CIFAR-100 head on top of ImageNet pretrained backbone
HEAD_EPOCHS = 1
HEAD_LR = 1e-3

POLICY_HIDDEN = 64
POLICY_EPOCHS = 6
POLICY_LR = 1e-3

RESULT_DIR = "/content/potts_results"
os.makedirs(RESULT_DIR, exist_ok=True)

BUDGETS = [1, 4, 8]
COSTS = {1: 1.0, 4: 4.0, 8: 8.0}
BUDGET_TO_KEY = {1: "p1_probs", 4: "p4_probs", 8: "p8_probs"}

FEATURE_ABLATIONS = {
    "entropy_only": ["entropy"],
    "full": ["entropy", "margin", "disagreement", "embedding"],
}
PREFERENCE_MODES = ["acc_only", "acc_cost"]
CURVE_TARGETS = [1.05, 1.20, 2.00, 4.00]

# =========================
# Seed
# =========================
def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.benchmark = True

print("DEVICE =", DEVICE)

# =========================
# Metrics / utils
# =========================
def compute_ece(probs: np.ndarray, labels: np.ndarray, n_bins: int = 15) -> float:
    confidences = probs.max(axis=1)
    predictions = probs.argmax(axis=1)
    accuracies = (predictions == labels).astype(np.float32)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        mask = (confidences > bins[i]) & (confidences <= bins[i + 1])
        if mask.any():
            bin_acc = accuracies[mask].mean()
            bin_conf = confidences[mask].mean()
            ece += np.abs(bin_acc - bin_conf) * mask.mean()
    return float(ece)

def top1_accuracy(probs: np.ndarray, labels: np.ndarray) -> float:
    return float((probs.argmax(axis=1) == labels).mean() * 100.0)

def predictive_entropy(probs: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return -(probs * (probs + eps).log()).sum(dim=1)

def prediction_margin(probs: torch.Tensor) -> torch.Tensor:
    top2 = torch.topk(probs, k=2, dim=1).values
    return top2[:, 0] - top2[:, 1]

def kl_divergence_batch(p: torch.Tensor, q: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return (p * ((p + eps).log() - (q + eps).log())).sum(dim=1)

def softmax_np(logits: np.ndarray) -> np.ndarray:
    x = logits - logits.max(axis=1, keepdims=True)
    ex = np.exp(x)
    return ex / ex.sum(axis=1, keepdims=True)

def make_subset_indices(n: int, train_n: int, val_n: int, test_n: int, seed: int = 42):
    idx = np.arange(n)
    rng = np.random.default_rng(seed)
    rng.shuffle(idx)
    train_idx = idx[:train_n]
    val_idx = idx[train_n:train_n + val_n]
    test_idx = idx[train_n + val_n:train_n + val_n + test_n]
    return train_idx, val_idx, test_idx

# =========================
# Transforms / data
# =========================
mean = (0.5071, 0.4867, 0.4408)
std = (0.2675, 0.2565, 0.2761)

base_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.RandomHorizontalFlip(),
    T.RandomCrop(IMAGE_SIZE, padding=16),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_ds_full = CIFAR100(root="/content/data", train=True, download=True, transform=train_transform)
eval_train_ds_full = CIFAR100(root="/content/data", train=True, download=False, transform=base_transform)
test_ds_full = CIFAR100(root="/content/data", train=False, download=True, transform=base_transform)

train_idx, val_idx, _ = make_subset_indices(
    n=len(train_ds_full),
    train_n=TRAIN_POLICY_SUBSET,
    val_n=VAL_SUBSET,
    test_n=TEST_SUBSET,
    seed=SEED
)

head_train_ds = Subset(train_ds_full, train_idx)
policy_train_ds = Subset(eval_train_ds_full, train_idx)
val_ds = Subset(eval_train_ds_full, val_idx)
test_ds = Subset(test_ds_full, np.arange(TEST_SUBSET))

head_train_loader = DataLoader(head_train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
policy_train_loader = DataLoader(policy_train_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# =========================
# Model
# =========================
class BackboneWrapper(nn.Module):
    def __init__(self, backbone_name: str, num_classes: int = 100):
        super().__init__()
        if backbone_name == "resnet50":
            m = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
            feat_dim = m.fc.in_features
            m.fc = nn.Identity()
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
            self.backbone_name = backbone_name
        elif backbone_name == "deit_small_patch16_224":
            m = timm.create_model("deit_small_patch16_224", pretrained=True, num_classes=0)
            feat_dim = m.num_features
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
            self.backbone_name = backbone_name
        else:
            raise ValueError(f"Unsupported backbone: {backbone_name}")

    def forward_features(self, x: torch.Tensor) -> torch.Tensor:
        return self.backbone(x)

    def forward(self, x: torch.Tensor):
        feat = self.forward_features(x)
        logits = self.head(feat)
        return logits, feat

model = BackboneWrapper(BACKBONE_NAME, num_classes=NUM_CLASSES).to(DEVICE)

for p in model.backbone.parameters():
    p.requires_grad = False

if USE_PRETRAINED_HEAD:
    print(f"[Info] Training classifier head for {HEAD_EPOCHS} epoch(s)...")
    optimizer = torch.optim.Adam(model.head.parameters(), lr=HEAD_LR)
    model.train()
    for epoch in range(HEAD_EPOCHS):
        pbar = tqdm(head_train_loader, desc=f"Head train {epoch+1}/{HEAD_EPOCHS}")
        for images, labels in pbar:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            optimizer.zero_grad()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
                logits, _ = model(images)
                loss = F.cross_entropy(logits, labels)
            loss.backward()
            optimizer.step()
            pbar.set_postfix(loss=float(loss.detach().cpu()))
else:
    raise RuntimeError("Set USE_PRETRAINED_HEAD=True or add your checkpoint loading code.")

model.eval()

# =========================
# TTA helper
# =========================
@torch.no_grad()
def batch_tta_predict(model: nn.Module, images: torch.Tensor, k: int = 1):
    model.eval()
    probs_list = []
    feats_list = []

    with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
        for _ in range(k):
            x = images.clone()
            if k > 1:
                if random.random() < 0.5:
                    x = torch.flip(x, dims=[3])
                noise = torch.randn_like(x) * 0.01
                x = x + noise

            logits, feats = model(x)
            probs = F.softmax(logits, dim=1)
            probs_list.append(probs.float())
            feats_list.append(feats.float())

    probs_stack = torch.stack(probs_list, dim=0)   # [K,B,C]
    feats_stack = torch.stack(feats_list, dim=0)   # [K,B,D]
    avg_probs = probs_stack.mean(dim=0)
    avg_feats = feats_stack.mean(dim=0)

    if k == 1:
        disagreement = torch.zeros(images.size(0), device=images.device)
    else:
        kls = []
        for i in range(k):
            kls.append(kl_divergence_batch(probs_stack[i], avg_probs))
        disagreement = torch.stack(kls, dim=0).mean(dim=0)

    return avg_probs, avg_feats, disagreement

# =========================
# Collect cached candidate policy outputs
# =========================
@torch.no_grad()
def collect_policy_candidates(loader: DataLoader, name: str) -> Dict[str, np.ndarray]:
    model.eval()
    out = {
        "labels": [],
        "p1_probs": [], "p4_probs": [], "p8_probs": [],
        "p1_feats": [], "p4_feats": [], "p8_feats": [],
        "p1_dis": [], "p4_dis": [], "p8_dis": [],
    }

    for images, labels in tqdm(loader, desc=f"Collect [{name}]"):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        probs1, feat1, dis1 = batch_tta_predict(model, images, k=1)
        probs4, feat4, dis4 = batch_tta_predict(model, images, k=4)
        probs8, feat8, dis8 = batch_tta_predict(model, images, k=8)

        out["labels"].append(labels.cpu().numpy())
        out["p1_probs"].append(probs1.cpu().numpy())
        out["p4_probs"].append(probs4.cpu().numpy())
        out["p8_probs"].append(probs8.cpu().numpy())

        out["p1_feats"].append(feat1.cpu().numpy())
        out["p4_feats"].append(feat4.cpu().numpy())
        out["p8_feats"].append(feat8.cpu().numpy())

        out["p1_dis"].append(dis1.cpu().numpy())
        out["p4_dis"].append(dis4.cpu().numpy())
        out["p8_dis"].append(dis8.cpu().numpy())

    for k in list(out.keys()):
        out[k] = np.concatenate(out[k], axis=0)
    return out

print("[Info] Collecting cached candidate outputs...")
train_cache = collect_policy_candidates(policy_train_loader, "train")
val_cache = collect_policy_candidates(val_loader, "val")
test_cache = collect_policy_candidates(test_loader, "test")

# =========================
# Feature builder
# =========================
def build_feature_matrix(cache: Dict[str, np.ndarray], feature_set: List[str], source_budget: int = 1) -> np.ndarray:
    probs = cache[f"p{source_budget}_probs"]
    feats = cache[f"p{source_budget}_feats"]
    dis = cache["p4_dis"]

    probs_t = torch.from_numpy(probs)
    entropy = predictive_entropy(probs_t).numpy()[:, None]
    margin = prediction_margin(probs_t).numpy()[:, None]
    pmax = probs.max(axis=1, keepdims=True)
    disagreement = dis[:, None]

    xs = []
    for name in feature_set:
        if name == "entropy":
            xs.append(entropy)
        elif name == "margin":
            xs.append(margin)
            xs.append(pmax)
        elif name == "disagreement":
            xs.append(disagreement)
        elif name == "embedding":
            xs.append(feats)
        else:
            raise ValueError(f"Unknown feature: {name}")
    return np.concatenate(xs, axis=1).astype(np.float32)

# =========================
# Preferences
# =========================
def choose_preference(
    probs_a: np.ndarray,
    probs_b: np.ndarray,
    label: int,
    budget_a: int,
    budget_b: int,
    mode: str = "acc_cost",
):
    pred_a = int(np.argmax(probs_a))
    pred_b = int(np.argmax(probs_b))
    acc_a = int(pred_a == label)
    acc_b = int(pred_b == label)

    if mode == "acc_only":
        if acc_a > acc_b:
            return budget_a, budget_b
        if acc_b > acc_a:
            return budget_b, budget_a
        return None

    if mode == "acc_cost":
        if acc_a > acc_b:
            return budget_a, budget_b
        if acc_b > acc_a:
            return budget_b, budget_a
        if acc_a == 1 and acc_b == 1:
            if COSTS[budget_a] < COSTS[budget_b]:
                return budget_a, budget_b
            if COSTS[budget_b] < COSTS[budget_a]:
                return budget_b, budget_a
        return None

    raise ValueError(f"Unknown mode: {mode}")

def build_preference_dataset(cache: Dict[str, np.ndarray], X: np.ndarray, mode: str = "acc_cost"):
    labels = cache["labels"]
    xs, chosen, rejected = [], [], []

    for i in range(len(labels)):
        for a, b in [(1,4), (1,8), (4,8)]:
            pref = choose_preference(
                cache[BUDGET_TO_KEY[a]][i],
                cache[BUDGET_TO_KEY[b]][i],
                int(labels[i]),
                a, b,
                mode=mode,
            )
            if pref is None:
                continue
            c, r = pref
            xs.append(X[i])
            chosen.append([1,4,8].index(c))
            rejected.append([1,4,8].index(r))

    xs = np.asarray(xs, dtype=np.float32)
    chosen = np.asarray(chosen, dtype=np.int64)
    rejected = np.asarray(rejected, dtype=np.int64)
    return xs, chosen, rejected

# =========================
# Policy net
# =========================
class PolicyNet(nn.Module):
    def __init__(self, input_dim: int, hidden_dim: int = 64, num_policies: int = 3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_policies),
        )

    def forward(self, x: torch.Tensor):
        return self.net(x)

def pairwise_pref_loss(scores: torch.Tensor, chosen: torch.Tensor, rejected: torch.Tensor) -> torch.Tensor:
    chosen_score = scores.gather(1, chosen[:, None]).squeeze(1)
    rejected_score = scores.gather(1, rejected[:, None]).squeeze(1)
    return -torch.log(torch.sigmoid(chosen_score - rejected_score) + 1e-12).mean()

def train_policy(X_train, ch_train, rj_train, X_val=None, ch_val=None, rj_val=None):
    pol = PolicyNet(input_dim=X_train.shape[1], hidden_dim=POLICY_HIDDEN).to(DEVICE)
    opt = torch.optim.Adam(pol.parameters(), lr=POLICY_LR)

    Xtr = torch.from_numpy(X_train)
    Ctr = torch.from_numpy(ch_train)
    Rtr = torch.from_numpy(rj_train)

    best_state = None
    best_val = 1e9

    for epoch in range(POLICY_EPOCHS):
        pol.train()
        perm = torch.randperm(len(Xtr))
        losses = []

        for start in range(0, len(Xtr), 1024):
            idx = perm[start:start+1024]
            xb = Xtr[idx].to(DEVICE)
            cb = Ctr[idx].to(DEVICE)
            rb = Rtr[idx].to(DEVICE)

            opt.zero_grad()
            scores = pol(xb)
            loss = pairwise_pref_loss(scores, cb, rb)
            loss.backward()
            opt.step()
            losses.append(float(loss.detach().cpu()))

        val_loss = np.mean(losses)

        if X_val is not None and len(X_val) > 0:
            pol.eval()
            with torch.no_grad():
                xv = torch.from_numpy(X_val).to(DEVICE)
                cv = torch.from_numpy(ch_val).to(DEVICE)
                rv = torch.from_numpy(rj_val).to(DEVICE)
                scores = pol(xv)
                val_loss = float(pairwise_pref_loss(scores, cv, rv).detach().cpu())

        print(f"[Policy] epoch {epoch+1}/{POLICY_EPOCHS} train={np.mean(losses):.4f} val={val_loss:.4f}")

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in pol.state_dict().items()}

    if best_state is not None:
        pol.load_state_dict(best_state)

    pol.eval()
    return pol

# =========================
# Evaluation
# =========================
def evaluate_fixed(cache: Dict[str, np.ndarray], budget: int) -> Dict[str, float]:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    return {
        "method": f"TTA-{budget}" if budget > 1 else "Single",
        "avg_compute": float(COSTS[budget]),
        "acc": top1_accuracy(probs, labels),
        "ece": compute_ece(probs, labels),
    }

def apply_budget_selection(cache: Dict[str, np.ndarray], selected_budgets: np.ndarray, name: str) -> Dict[str, float]:
    labels = cache["labels"]
    final_probs = []
    costs = []

    for i, b in enumerate(selected_budgets.tolist()):
        final_probs.append(cache[BUDGET_TO_KEY[int(b)]][i])
        costs.append(COSTS[int(b)])

    final_probs = np.asarray(final_probs)
    return {
        "method": name,
        "avg_compute": float(np.mean(costs)),
        "acc": top1_accuracy(final_probs, labels),
        "ece": compute_ece(final_probs, labels),
    }

def tune_thresholds(values: np.ndarray, cache: Dict[str, np.ndarray], mode: str = "entropy", target_compute: Optional[float] = None):
    labels = cache["labels"]
    candidates = np.quantile(values, np.linspace(0.1, 0.9, 9))
    best = None
    best_score = -1e9

    for t1 in candidates:
        for t2 in candidates:
            if t2 < t1:
                continue

            if mode == "entropy":
                sel = np.where(values < t1, 1, np.where(values < t2, 4, 8))
            else:
                sel = np.where(values > t2, 1, np.where(values > t1, 4, 8))

            probs = np.stack([cache[BUDGET_TO_KEY[int(b)]][i] for i, b in enumerate(sel)], axis=0)
            acc = top1_accuracy(probs, labels)
            avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))

            score = acc
            if target_compute is not None:
                score -= 10.0 * abs(avg_compute - target_compute)

            if score > best_score:
                best_score = score
                best = (t1, t2, sel, acc, avg_compute)

    return best

def evaluate_random_policy(cache: Dict[str, np.ndarray], target_compute: float = 1.10, seed: int = 42):
    rng = np.random.default_rng(seed)
    best = None
    best_gap = 1e9
    probs_set = [
        np.array([1.0, 0.0, 0.0], dtype=np.float32),
        np.array([0.8, 0.18, 0.02], dtype=np.float32),
        np.array([0.7, 0.2, 0.1], dtype=np.float32),
        np.array([0.5, 0.3, 0.2], dtype=np.float32),
    ]
    for p in probs_set:
        sel_idx = rng.choice(3, size=len(cache["labels"]), p=p)
        sel = np.array([BUDGETS[i] for i in sel_idx])
        avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))
        gap = abs(avg_compute - target_compute)
        if gap < best_gap:
            best_gap = gap
            best = sel
    return apply_budget_selection(cache, best, f"Random(target={target_compute:.2f})")

@torch.no_grad()
def evaluate_policy_network(policy_model: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], name: str = "POTTS"):
    scores = policy_model(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    sel_idx = scores.argmax(axis=1)
    sel = np.array([BUDGETS[i] for i in sel_idx])
    return apply_budget_selection(cache, sel, name)

@torch.no_grad()
def evaluate_policy_with_temperature(policy_model: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], temperature: float, name: str):
    scores = policy_model(torch.from_numpy(X).to(DEVICE)).cpu().numpy() / temperature
    probs = softmax_np(scores)
    sel_idx = probs.argmax(axis=1)
    sel = np.array([BUDGETS[i] for i in sel_idx])
    return apply_budget_selection(cache, sel, name)

def search_policy_operating_points(policy_model: nn.Module, X_val: np.ndarray, cache_val: Dict[str, np.ndarray], targets: List[float]):
    rows = []
    temps = np.linspace(0.25, 4.0, 20)
    for target in targets:
        best = None
        best_gap = 1e9
        for t in temps:
            row = evaluate_policy_with_temperature(policy_model, X_val, cache_val, t, f"POTTS@T={t:.2f}")
            gap = abs(row["avg_compute"] - target)
            if gap < best_gap:
                best_gap = gap
                best = (t, row)
        rows.append({
            "target_compute": target,
            "temperature": best[0],
            **best[1]
        })
    return pd.DataFrame(rows)

# =========================
# Main experiments
# =========================
main_rows = []

# fixed baselines
main_rows.append(evaluate_fixed(test_cache, 1))
main_rows.append(evaluate_fixed(test_cache, 4))
main_rows.append(evaluate_fixed(test_cache, 8))

# random baseline
main_rows.append(evaluate_random_policy(test_cache, target_compute=1.10, seed=SEED))

# heuristic baselines
val_entropy = predictive_entropy(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_entropy = predictive_entropy(torch.from_numpy(test_cache["p1_probs"])).numpy()

val_margin = prediction_margin(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_margin = prediction_margin(torch.from_numpy(test_cache["p1_probs"])).numpy()

ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=1.10)
t1_ent, t2_ent, _, _, _ = ent_best
ent_sel_test = np.where(test_entropy < t1_ent, 1, np.where(test_entropy < t2_ent, 4, 8))
main_rows.append(apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold"))

mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=1.10)
t1_mar, t2_mar, _, _, _ = mar_best
mar_sel_test = np.where(test_margin > t2_mar, 1, np.where(test_margin > t1_mar, 4, 8))
main_rows.append(apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold"))

# ablations
ablation_rows = []
trained_policies = {}

for pref_mode in PREFERENCE_MODES:
    for feat_name, feat_list in FEATURE_ABLATIONS.items():
        print(f"\n[Info] Training policy: pref={pref_mode}, feat={feat_name}")

        X_train = build_feature_matrix(train_cache, feat_list, source_budget=1)
        X_val = build_feature_matrix(val_cache, feat_list, source_budget=1)
        X_test = build_feature_matrix(test_cache, feat_list, source_budget=1)

        Xtr, chtr, rjtr = build_preference_dataset(train_cache, X_train, mode=pref_mode)
        Xv, chv, rjv = build_preference_dataset(val_cache, X_val, mode=pref_mode)

        if len(Xtr) == 0:
            print("[Warn] No preference pairs. Skipping.")
            continue

        pol = train_policy(Xtr, chtr, rjtr, Xv, chv, rjv)
        trained_policies[(pref_mode, feat_name)] = pol

        row = evaluate_policy_network(pol, X_test, test_cache, name=f"POTTS[{pref_mode}|{feat_name}]")
        row["preference_mode"] = pref_mode
        row["feature_set"] = feat_name
        ablation_rows.append(row)

# main POTTS = acc_cost + full
potts_main = [r for r in ablation_rows if r["preference_mode"] == "acc_cost" and r["feature_set"] == "full"]
if len(potts_main) > 0:
    main_rows.append({
        "method": "POTTS",
        "avg_compute": potts_main[0]["avg_compute"],
        "acc": potts_main[0]["acc"],
        "ece": potts_main[0]["ece"],
    })

main_df = pd.DataFrame(main_rows).sort_values(["avg_compute", "acc"], ascending=[True, False]).reset_index(drop=True)
ablation_df = pd.DataFrame(ablation_rows)

print("\n=== MAIN RESULTS ===")
display(main_df)

print("\n=== ABLATION RESULTS ===")
display(ablation_df)

main_df.to_csv(os.path.join(RESULT_DIR, "main_results.csv"), index=False)
ablation_df.to_csv(os.path.join(RESULT_DIR, "ablation_results.csv"), index=False)

# =========================
# Trade-off curves
# =========================
curve_rows = []

# heuristic curves
for target in CURVE_TARGETS:
    ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=target)
    t1, t2, _, _, _ = ent_best
    ent_sel_test = np.where(test_entropy < t1, 1, np.where(test_entropy < t2, 4, 8))
    row = apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

    mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=target)
    t1m, t2m, _, _, _ = mar_best
    mar_sel_test = np.where(test_margin > t2m, 1, np.where(test_margin > t1m, 4, 8))
    row = apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

# POTTS curve
if ("acc_cost", "full") in trained_policies:
    pol = trained_policies[("acc_cost", "full")]
    X_val_full = build_feature_matrix(val_cache, FEATURE_ABLATIONS["full"], source_budget=1)
    X_test_full = build_feature_matrix(test_cache, FEATURE_ABLATIONS["full"], source_budget=1)
    curve_df_tmp = search_policy_operating_points(pol, X_val_full, val_cache, CURVE_TARGETS)

    for _, rr in curve_df_tmp.iterrows():
        row = evaluate_policy_with_temperature(pol, X_test_full, test_cache, rr["temperature"], "POTTS")
        row["target_compute"] = rr["target_compute"]
        row["temperature"] = rr["temperature"]
        curve_rows.append(row)

curve_df = pd.DataFrame(curve_rows)
curve_df.to_csv(os.path.join(RESULT_DIR, "tradeoff_curve.csv"), index=False)

print("\n=== TRADE-OFF CURVE ===")
display(curve_df)

# =========================
# Paper-friendly main table
# =========================
paper_rows = []
for _, r in main_df.iterrows():
    paper_rows.append({
        "Method": r["method"],
        "Compute": round(float(r["avg_compute"]), 3),
        "Acc (%)": round(float(r["acc"]), 3),
        "ECE": round(float(r["ece"]), 4),
    })
paper_df = pd.DataFrame(paper_rows)
paper_df.to_csv(os.path.join(RESULT_DIR, "paper_table_main.csv"), index=False)

print("\nSaved files:")
for fn in sorted(os.listdir(RESULT_DIR)):
    print(" -", os.path.join(RESULT_DIR, fn))
```

```python
# =========================================================
# ONE-CELL COLAB: POTTS fast rerun with fixed TTA
# =========================================================
# What this version does:
# - fixes broken TTA (removes Gaussian noise)
# - fixes random baseline so it does not collapse to Single
# - uses conservative deterministic TTA
# - runs a smaller fast setup first
#
# Outputs:
# /content/potts_results/main_results.csv
# /content/potts_results/ablation_results.csv
# /content/potts_results/tradeoff_curve.csv
# /content/potts_results/paper_table_main.csv
# =========================================================

!pip -q install timm pandas scikit-learn tqdm

import os
import random
import warnings
from typing import Dict, List, Tuple, Optional

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset

import torchvision
import torchvision.transforms as T
from torchvision.datasets import CIFAR100

import timm

warnings.filterwarnings("ignore")

# =========================
# Config
# =========================
SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
NUM_WORKERS = 2
BATCH_SIZE = 256

TRAIN_POLICY_SUBSET = 8000
VAL_SUBSET = 1000
TEST_SUBSET = 2000

IMAGE_SIZE = 224
NUM_CLASSES = 100

BACKBONE_NAME = "resnet50"
USE_PRETRAINED_HEAD = True
HEAD_EPOCHS = 1
HEAD_LR = 1e-3

POLICY_HIDDEN = 64
POLICY_EPOCHS = 4
POLICY_LR = 1e-3

RESULT_DIR = "/content/potts_results"
os.makedirs(RESULT_DIR, exist_ok=True)

BUDGETS = [1, 4, 8]
COSTS = {1: 1.0, 4: 4.0, 8: 8.0}
BUDGET_TO_KEY = {1: "p1_probs", 4: "p4_probs", 8: "p8_probs"}

FEATURE_ABLATIONS = {
    "entropy_only": ["entropy"],
    "full": ["entropy", "margin", "disagreement", "embedding"],
}

PREFERENCE_MODES = ["acc_only", "acc_cost"]
CURVE_TARGETS = [1.05, 1.20, 2.00, 4.00]

# =========================
# Seed
# =========================
def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.benchmark = True

print("DEVICE =", DEVICE)

# =========================
# Metrics / utils
# =========================
def compute_ece(probs: np.ndarray, labels: np.ndarray, n_bins: int = 15) -> float:
    confidences = probs.max(axis=1)
    predictions = probs.argmax(axis=1)
    accuracies = (predictions == labels).astype(np.float32)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        mask = (confidences > bins[i]) & (confidences <= bins[i + 1])
        if mask.any():
            bin_acc = accuracies[mask].mean()
            bin_conf = confidences[mask].mean()
            ece += np.abs(bin_acc - bin_conf) * mask.mean()
    return float(ece)

def top1_accuracy(probs: np.ndarray, labels: np.ndarray) -> float:
    return float((probs.argmax(axis=1) == labels).mean() * 100.0)

def predictive_entropy(probs: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return -(probs * (probs + eps).log()).sum(dim=1)

def prediction_margin(probs: torch.Tensor) -> torch.Tensor:
    top2 = torch.topk(probs, k=2, dim=1).values
    return top2[:, 0] - top2[:, 1]

def kl_divergence_batch(p: torch.Tensor, q: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return (p * ((p + eps).log() - (q + eps).log())).sum(dim=1)

def softmax_np(logits: np.ndarray) -> np.ndarray:
    x = logits - logits.max(axis=1, keepdims=True)
    ex = np.exp(x)
    return ex / ex.sum(axis=1, keepdims=True)

def make_subset_indices(n: int, train_n: int, val_n: int, test_n: int, seed: int = 42):
    idx = np.arange(n)
    rng = np.random.default_rng(seed)
    rng.shuffle(idx)
    train_idx = idx[:train_n]
    val_idx = idx[train_n:train_n + val_n]
    test_idx = idx[train_n + val_n:train_n + val_n + test_n]
    return train_idx, val_idx, test_idx

# =========================
# Data
# =========================
mean = (0.5071, 0.4867, 0.4408)
std = (0.2675, 0.2565, 0.2761)

base_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.RandomHorizontalFlip(),
    T.RandomCrop(IMAGE_SIZE, padding=16),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_ds_full = CIFAR100(root="/content/data", train=True, download=True, transform=train_transform)
eval_train_ds_full = CIFAR100(root="/content/data", train=True, download=False, transform=base_transform)
test_ds_full = CIFAR100(root="/content/data", train=False, download=True, transform=base_transform)

train_idx, val_idx, _ = make_subset_indices(
    n=len(train_ds_full),
    train_n=TRAIN_POLICY_SUBSET,
    val_n=VAL_SUBSET,
    test_n=TEST_SUBSET,
    seed=SEED
)

head_train_ds = Subset(train_ds_full, train_idx)
policy_train_ds = Subset(eval_train_ds_full, train_idx)
val_ds = Subset(eval_train_ds_full, val_idx)
test_ds = Subset(test_ds_full, np.arange(TEST_SUBSET))

head_train_loader = DataLoader(head_train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
policy_train_loader = DataLoader(policy_train_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# =========================
# Model
# =========================
class BackboneWrapper(nn.Module):
    def __init__(self, backbone_name: str, num_classes: int = 100):
        super().__init__()
        if backbone_name == "resnet50":
            m = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
            feat_dim = m.fc.in_features
            m.fc = nn.Identity()
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        elif backbone_name == "deit_small_patch16_224":
            m = timm.create_model("deit_small_patch16_224", pretrained=True, num_classes=0)
            feat_dim = m.num_features
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        else:
            raise ValueError(f"Unsupported backbone: {backbone_name}")

    def forward_features(self, x: torch.Tensor) -> torch.Tensor:
        return self.backbone(x)

    def forward(self, x: torch.Tensor):
        feat = self.forward_features(x)
        logits = self.head(feat)
        return logits, feat

model = BackboneWrapper(BACKBONE_NAME, num_classes=NUM_CLASSES).to(DEVICE)

for p in model.backbone.parameters():
    p.requires_grad = False

if USE_PRETRAINED_HEAD:
    print(f"[Info] Training classifier head for {HEAD_EPOCHS} epoch(s)...")
    optimizer = torch.optim.Adam(model.head.parameters(), lr=HEAD_LR)
    model.train()
    for epoch in range(HEAD_EPOCHS):
        pbar = tqdm(head_train_loader, desc=f"Head train {epoch+1}/{HEAD_EPOCHS}")
        for images, labels in pbar:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            optimizer.zero_grad()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
                logits, _ = model(images)
                loss = F.cross_entropy(logits, labels)
            loss.backward()
            optimizer.step()
            pbar.set_postfix(loss=float(loss.detach().cpu()))
else:
    raise RuntimeError("Set USE_PRETRAINED_HEAD=True or add your checkpoint loading code.")

model.eval()

# =========================
# FIXED TTA helper
# =========================
@torch.no_grad()
def batch_tta_predict(model: nn.Module, images: torch.Tensor, k: int = 1):
    model.eval()
    probs_list = []
    feats_list = []

    def apply_det_tta(x: torch.Tensor, aug_id: int):
        # Conservative deterministic TTA only. No Gaussian noise.
        if aug_id == 0:
            return x
        elif aug_id == 1:
            # horizontal flip
            return torch.flip(x, dims=[3])
        elif aug_id == 2:
            # mild left crop then resize back
            x2 = x[:, :, :, 4:]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 3:
            # mild right crop then resize back
            x2 = x[:, :, :, :-4]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 4:
            # mild top crop then resize back
            x2 = x[:, :, 4:, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 5:
            # mild bottom crop then resize back
            x2 = x[:, :, :-4, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 6:
            # center crop 90% then resize back
            H, W = x.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 7:
            # flip + center crop
            x2 = torch.flip(x, dims=[3])
            H, W = x2.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x2[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        else:
            return x

    with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
        for aug_id in range(k):
            x_aug = apply_det_tta(images, aug_id)
            logits, feats = model(x_aug)
            probs = F.softmax(logits, dim=1)
            probs_list.append(probs.float())
            feats_list.append(feats.float())

    probs_stack = torch.stack(probs_list, dim=0)
    feats_stack = torch.stack(feats_list, dim=0)
    avg_probs = probs_stack.mean(dim=0)
    avg_feats = feats_stack.mean(dim=0)

    if k == 1:
        disagreement = torch.zeros(images.size(0), device=images.device)
    else:
        kls = []
        for i in range(k):
            kls.append(kl_divergence_batch(probs_stack[i], avg_probs))
        disagreement = torch.stack(kls, dim=0).mean(dim=0)

    return avg_probs, avg_feats, disagreement

# =========================
# Collect caches
# =========================
@torch.no_grad()
def collect_policy_candidates(loader: DataLoader, name: str) -> Dict[str, np.ndarray]:
    model.eval()
    out = {
        "labels": [],
        "p1_probs": [], "p4_probs": [], "p8_probs": [],
        "p1_feats": [], "p4_feats": [], "p8_feats": [],
        "p1_dis": [], "p4_dis": [], "p8_dis": [],
    }

    for images, labels in tqdm(loader, desc=f"Collect [{name}]"):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        probs1, feat1, dis1 = batch_tta_predict(model, images, k=1)
        probs4, feat4, dis4 = batch_tta_predict(model, images, k=4)
        probs8, feat8, dis8 = batch_tta_predict(model, images, k=8)

        out["labels"].append(labels.cpu().numpy())
        out["p1_probs"].append(probs1.cpu().numpy())
        out["p4_probs"].append(probs4.cpu().numpy())
        out["p8_probs"].append(probs8.cpu().numpy())

        out["p1_feats"].append(feat1.cpu().numpy())
        out["p4_feats"].append(feat4.cpu().numpy())
        out["p8_feats"].append(feat8.cpu().numpy())

        out["p1_dis"].append(dis1.cpu().numpy())
        out["p4_dis"].append(dis4.cpu().numpy())
        out["p8_dis"].append(dis8.cpu().numpy())

    for k in list(out.keys()):
        out[k] = np.concatenate(out[k], axis=0)
    return out

print("[Info] Collecting cached candidate outputs...")
train_cache = collect_policy_candidates(policy_train_loader, "train")
val_cache = collect_policy_candidates(val_loader, "val")
test_cache = collect_policy_candidates(test_loader, "test")

# Quick sanity check first
single_acc = top1_accuracy(test_cache["p1_probs"], test_cache["labels"])
tta4_acc = top1_accuracy(test_cache["p4_probs"], test_cache["labels"])
tta8_acc = top1_accuracy(test_cache["p8_probs"], test_cache["labels"])
print("\n=== QUICK TTA SANITY CHECK ===")
print(f"Single: {single_acc:.2f}")
print(f"TTA-4 : {tta4_acc:.2f}")
print(f"TTA-8 : {tta8_acc:.2f}")

# =========================
# Feature builder
# =========================
def build_feature_matrix(cache: Dict[str, np.ndarray], feature_set: List[str], source_budget: int = 1) -> np.ndarray:
    probs = cache[f"p{source_budget}_probs"]
    feats = cache[f"p{source_budget}_feats"]
    dis = cache["p4_dis"]

    probs_t = torch.from_numpy(probs)
    entropy = predictive_entropy(probs_t).numpy()[:, None]
    margin = prediction_margin(probs_t).numpy()[:, None]
    pmax = probs.max(axis=1, keepdims=True)
    disagreement = dis[:, None]

    xs = []
    for name in feature_set:
        if name == "entropy":
            xs.append(entropy)
        elif name == "margin":
            xs.append(margin)
            xs.append(pmax)
        elif name == "disagreement":
            xs.append(disagreement)
        elif name == "embedding":
            xs.append(feats)
        else:
            raise ValueError(f"Unknown feature: {name}")
    return np.concatenate(xs, axis=1).astype(np.float32)

# =========================
# Preferences
# =========================
def choose_preference(
    probs_a: np.ndarray,
    probs_b: np.ndarray,
    label: int,
    budget_a: int,
    budget_b: int,
    mode: str = "acc_cost",
):
    pred_a = int(np.argmax(probs_a))
    pred_b = int(np.argmax(probs_b))
    acc_a = int(pred_a == label)
    acc_b = int(pred_b == label)

    if mode == "acc_only":
        if acc_a > acc_b:
            return budget_a, budget_b
        if acc_b > acc_a:
            return budget_b, budget_a
        return None

    if mode == "acc_cost":
        if acc_a > acc_b:
            return budget_a, budget_b
        if acc_b > acc_a:
            return budget_b, budget_a
        if acc_a == 1 and acc_b == 1:
            if COSTS[budget_a] < COSTS[budget_b]:
                return budget_a, budget_b
            if COSTS[budget_b] < COSTS[budget_a]:
                return budget_b, budget_a
        return None

    raise ValueError(f"Unknown mode: {mode}")

def build_preference_dataset(cache: Dict[str, np.ndarray], X: np.ndarray, mode: str = "acc_cost"):
    labels = cache["labels"]
    xs, chosen, rejected = [], [], []

    for i in range(len(labels)):
        for a, b in [(1, 4), (1, 8), (4, 8)]:
            pref = choose_preference(
                cache[BUDGET_TO_KEY[a]][i],
                cache[BUDGET_TO_KEY[b]][i],
                int(labels[i]),
                a, b,
                mode=mode,
            )
            if pref is None:
                continue
            c, r = pref
            xs.append(X[i])
            chosen.append([1, 4, 8].index(c))
            rejected.append([1, 4, 8].index(r))

    xs = np.asarray(xs, dtype=np.float32)
    chosen = np.asarray(chosen, dtype=np.int64)
    rejected = np.asarray(rejected, dtype=np.int64)
    return xs, chosen, rejected

# =========================
# Policy net
# =========================
class PolicyNet(nn.Module):
    def __init__(self, input_dim: int, hidden_dim: int = 64, num_policies: int = 3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_policies),
        )

    def forward(self, x: torch.Tensor):
        return self.net(x)

def pairwise_pref_loss(scores: torch.Tensor, chosen: torch.Tensor, rejected: torch.Tensor) -> torch.Tensor:
    chosen_score = scores.gather(1, chosen[:, None]).squeeze(1)
    rejected_score = scores.gather(1, rejected[:, None]).squeeze(1)
    return -torch.log(torch.sigmoid(chosen_score - rejected_score) + 1e-12).mean()

def train_policy(X_train, ch_train, rj_train, X_val=None, ch_val=None, rj_val=None):
    pol = PolicyNet(input_dim=X_train.shape[1], hidden_dim=POLICY_HIDDEN).to(DEVICE)
    opt = torch.optim.Adam(pol.parameters(), lr=POLICY_LR)

    Xtr = torch.from_numpy(X_train)
    Ctr = torch.from_numpy(ch_train)
    Rtr = torch.from_numpy(rj_train)

    best_state = None
    best_val = 1e9

    for epoch in range(POLICY_EPOCHS):
        pol.train()
        perm = torch.randperm(len(Xtr))
        losses = []

        for start in range(0, len(Xtr), 1024):
            idx = perm[start:start + 1024]
            xb = Xtr[idx].to(DEVICE)
            cb = Ctr[idx].to(DEVICE)
            rb = Rtr[idx].to(DEVICE)

            opt.zero_grad()
            scores = pol(xb)
            loss = pairwise_pref_loss(scores, cb, rb)
            loss.backward()
            opt.step()
            losses.append(float(loss.detach().cpu()))

        val_loss = np.mean(losses)
        if X_val is not None and len(X_val) > 0:
            pol.eval()
            with torch.no_grad():
                xv = torch.from_numpy(X_val).to(DEVICE)
                cv = torch.from_numpy(ch_val).to(DEVICE)
                rv = torch.from_numpy(rj_val).to(DEVICE)
                scores = pol(xv)
                val_loss = float(pairwise_pref_loss(scores, cv, rv).detach().cpu())

        print(f"[Policy] epoch {epoch+1}/{POLICY_EPOCHS} train={np.mean(losses):.4f} val={val_loss:.4f}")

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in pol.state_dict().items()}

    if best_state is not None:
        pol.load_state_dict(best_state)

    pol.eval()
    return pol

# =========================
# Evaluation
# =========================
def evaluate_fixed(cache: Dict[str, np.ndarray], budget: int) -> Dict[str, float]:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    return {
        "method": f"TTA-{budget}" if budget > 1 else "Single",
        "avg_compute": float(COSTS[budget]),
        "acc": top1_accuracy(probs, labels),
        "ece": compute_ece(probs, labels),
    }

def apply_budget_selection(cache: Dict[str, np.ndarray], selected_budgets: np.ndarray, name: str) -> Dict[str, float]:
    labels = cache["labels"]
    final_probs = []
    costs = []

    for i, b in enumerate(selected_budgets.tolist()):
        final_probs.append(cache[BUDGET_TO_KEY[int(b)]][i])
        costs.append(COSTS[int(b)])

    final_probs = np.asarray(final_probs)
    return {
        "method": name,
        "avg_compute": float(np.mean(costs)),
        "acc": top1_accuracy(final_probs, labels),
        "ece": compute_ece(final_probs, labels),
    }

def tune_thresholds(values: np.ndarray, cache: Dict[str, np.ndarray], mode: str = "entropy", target_compute: Optional[float] = None):
    labels = cache["labels"]
    candidates = np.quantile(values, np.linspace(0.1, 0.9, 9))
    best = None
    best_score = -1e9

    for t1 in candidates:
        for t2 in candidates:
            if t2 < t1:
                continue

            if mode == "entropy":
                sel = np.where(values < t1, 1, np.where(values < t2, 4, 8))
            else:
                sel = np.where(values > t2, 1, np.where(values > t1, 4, 8))

            probs = np.stack([cache[BUDGET_TO_KEY[int(b)]][i] for i, b in enumerate(sel)], axis=0)
            acc = top1_accuracy(probs, labels)
            avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))

            score = acc
            if target_compute is not None:
                score -= 10.0 * abs(avg_compute - target_compute)

            if score > best_score:
                best_score = score
                best = (t1, t2, sel, acc, avg_compute)

    return best

def evaluate_random_policy(cache: Dict[str, np.ndarray], target_compute: float = 1.10, seed: int = 42):
    rng = np.random.default_rng(seed)
    best = None
    best_gap = 1e9

    # Low-compute-biased random policies so this no longer collapses to Single.
    probs_set = [
        np.array([0.98, 0.02, 0.00], dtype=np.float32),
        np.array([0.96, 0.04, 0.00], dtype=np.float32),
        np.array([0.94, 0.05, 0.01], dtype=np.float32),
        np.array([0.92, 0.06, 0.02], dtype=np.float32),
        np.array([0.90, 0.08, 0.02], dtype=np.float32),
        np.array([0.88, 0.10, 0.02], dtype=np.float32),
    ]

    for p in probs_set:
        sel_idx = rng.choice(3, size=len(cache["labels"]), p=p)
        sel = np.array([BUDGETS[i] for i in sel_idx])
        avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))
        gap = abs(avg_compute - target_compute)
        if gap < best_gap:
            best_gap = gap
            best = sel

    return apply_budget_selection(cache, best, f"Random(target={target_compute:.2f})")

@torch.no_grad()
def evaluate_policy_network(policy_model: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], name: str = "POTTS"):
    scores = policy_model(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    sel_idx = scores.argmax(axis=1)
    sel = np.array([BUDGETS[i] for i in sel_idx])
    return apply_budget_selection(cache, sel, name)

@torch.no_grad()
def evaluate_policy_with_temperature(policy_model: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], temperature: float, name: str):
    scores = policy_model(torch.from_numpy(X).to(DEVICE)).cpu().numpy() / temperature
    probs = softmax_np(scores)
    sel_idx = probs.argmax(axis=1)
    sel = np.array([BUDGETS[i] for i in sel_idx])
    return apply_budget_selection(cache, sel, name)

def search_policy_operating_points(policy_model: nn.Module, X_val: np.ndarray, cache_val: Dict[str, np.ndarray], targets: List[float]):
    rows = []
    temps = np.linspace(0.25, 4.0, 20)
    for target in targets:
        best = None
        best_gap = 1e9
        for t in temps:
            row = evaluate_policy_with_temperature(policy_model, X_val, cache_val, t, f"POTTS@T={t:.2f}")
            gap = abs(row["avg_compute"] - target)
            if gap < best_gap:
                best_gap = gap
                best = (t, row)
        rows.append({
            "target_compute": target,
            "temperature": best[0],
            **best[1]
        })
    return pd.DataFrame(rows)

# =========================
# Main experiments
# =========================
main_rows = []
main_rows.append(evaluate_fixed(test_cache, 1))
main_rows.append(evaluate_fixed(test_cache, 4))
main_rows.append(evaluate_fixed(test_cache, 8))
main_rows.append(evaluate_random_policy(test_cache, target_compute=1.10, seed=SEED))

val_entropy = predictive_entropy(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_entropy = predictive_entropy(torch.from_numpy(test_cache["p1_probs"])).numpy()

val_margin = prediction_margin(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_margin = prediction_margin(torch.from_numpy(test_cache["p1_probs"])).numpy()

ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=1.10)
t1_ent, t2_ent, _, _, _ = ent_best
ent_sel_test = np.where(test_entropy < t1_ent, 1, np.where(test_entropy < t2_ent, 4, 8))
main_rows.append(apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold"))

mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=1.10)
t1_mar, t2_mar, _, _, _ = mar_best
mar_sel_test = np.where(test_margin > t2_mar, 1, np.where(test_margin > t1_mar, 4, 8))
main_rows.append(apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold"))

ablation_rows = []
trained_policies = {}

for pref_mode in PREFERENCE_MODES:
    for feat_name, feat_list in FEATURE_ABLATIONS.items():
        print(f"\n[Info] Training policy: pref={pref_mode}, feat={feat_name}")
        X_train = build_feature_matrix(train_cache, feat_list, source_budget=1)
        X_val = build_feature_matrix(val_cache, feat_list, source_budget=1)
        X_test = build_feature_matrix(test_cache, feat_list, source_budget=1)

        Xtr, chtr, rjtr = build_preference_dataset(train_cache, X_train, mode=pref_mode)
        Xv, chv, rjv = build_preference_dataset(val_cache, X_val, mode=pref_mode)

        if len(Xtr) == 0:
            print("[Warn] No preference pairs. Skipping.")
            continue

        pol = train_policy(Xtr, chtr, rjtr, Xv, chv, rjv)
        trained_policies[(pref_mode, feat_name)] = pol

        row = evaluate_policy_network(pol, X_test, test_cache, name=f"POTTS[{pref_mode}|{feat_name}]")
        row["preference_mode"] = pref_mode
        row["feature_set"] = feat_name
        ablation_rows.append(row)

potts_main = [r for r in ablation_rows if r["preference_mode"] == "acc_cost" and r["feature_set"] == "full"]
if len(potts_main) > 0:
    main_rows.append({
        "method": "POTTS",
        "avg_compute": potts_main[0]["avg_compute"],
        "acc": potts_main[0]["acc"],
        "ece": potts_main[0]["ece"],
    })

main_df = pd.DataFrame(main_rows).sort_values(["avg_compute", "acc"], ascending=[True, False]).reset_index(drop=True)
ablation_df = pd.DataFrame(ablation_rows)

print("\n=== MAIN RESULTS ===")
display(main_df)

print("\n=== ABLATION RESULTS ===")
display(ablation_df)

main_df.to_csv(os.path.join(RESULT_DIR, "main_results.csv"), index=False)
ablation_df.to_csv(os.path.join(RESULT_DIR, "ablation_results.csv"), index=False)

# =========================
# Trade-off curve
# =========================
curve_rows = []

for target in CURVE_TARGETS:
    ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=target)
    t1, t2, _, _, _ = ent_best
    ent_sel_test = np.where(test_entropy < t1, 1, np.where(test_entropy < t2, 4, 8))
    row = apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

    mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=target)
    t1m, t2m, _, _, _ = mar_best
    mar_sel_test = np.where(test_margin > t2m, 1, np.where(test_margin > t1m, 4, 8))
    row = apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

if ("acc_cost", "full") in trained_policies:
    pol = trained_policies[("acc_cost", "full")]
    X_val_full = build_feature_matrix(val_cache, FEATURE_ABLATIONS["full"], source_budget=1)
    X_test_full = build_feature_matrix(test_cache, FEATURE_ABLATIONS["full"], source_budget=1)
    curve_df_tmp = search_policy_operating_points(pol, X_val_full, val_cache, CURVE_TARGETS)

    for _, rr in curve_df_tmp.iterrows():
        row = evaluate_policy_with_temperature(pol, X_test_full, test_cache, rr["temperature"], "POTTS")
        row["target_compute"] = rr["target_compute"]
        row["temperature"] = rr["temperature"]
        curve_rows.append(row)

curve_df = pd.DataFrame(curve_rows)
curve_df.to_csv(os.path.join(RESULT_DIR, "tradeoff_curve.csv"), index=False)

print("\n=== TRADE-OFF CURVE ===")
display(curve_df)

# =========================
# Paper-friendly table
# =========================
paper_rows = []
for _, r in main_df.iterrows():
    paper_rows.append({
        "Method": r["method"],
        "Compute": round(float(r["avg_compute"]), 3),
        "Acc (%)": round(float(r["acc"]), 3),
        "ECE": round(float(r["ece"]), 4),
    })

paper_df = pd.DataFrame(paper_rows)
paper_df.to_csv(os.path.join(RESULT_DIR, "paper_table_main.csv"), index=False)

print("\nSaved files:")
for fn in sorted(os.listdir(RESULT_DIR)):
    print(" -", os.path.join(RESULT_DIR, fn))
```

```python
# =========================================================
# ONE-CELL COLAB: POTTS rerun with FIXED TTA + SOFT UTILITY PREFERENCES
# =========================================================

!pip -q install timm pandas scikit-learn tqdm

import os
import random
import warnings
from typing import Dict, List, Tuple, Optional

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset

import torchvision
import torchvision.transforms as T
from torchvision.datasets import CIFAR100

import timm

warnings.filterwarnings("ignore")

# =========================
# Config
# =========================
SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
NUM_WORKERS = 2
BATCH_SIZE = 256

TRAIN_POLICY_SUBSET = 8000
VAL_SUBSET = 1000
TEST_SUBSET = 2000

IMAGE_SIZE = 224
NUM_CLASSES = 100

BACKBONE_NAME = "resnet50"
USE_PRETRAINED_HEAD = True
HEAD_EPOCHS = 1
HEAD_LR = 1e-3

POLICY_HIDDEN = 64
POLICY_EPOCHS = 4
POLICY_LR = 1e-3

RESULT_DIR = "/content/potts_results"
os.makedirs(RESULT_DIR, exist_ok=True)

BUDGETS = [1, 4, 8]
COSTS = {1: 1.0, 4: 4.0, 8: 8.0}
BUDGET_TO_KEY = {1: "p1_probs", 4: "p4_probs", 8: "p8_probs"}

FEATURE_ABLATIONS = {
    "entropy_only": ["entropy"],
    "full": ["entropy", "margin", "disagreement", "embedding"],
}

# changed
PREFERENCE_MODES = ["acc_only", "soft_0.10", "soft_0.20"]
CURVE_TARGETS = [1.05, 1.20, 2.00, 4.00]

# =========================
# Seed
# =========================
def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.benchmark = True

print("DEVICE =", DEVICE)

# =========================
# Metrics / utils
# =========================
def compute_ece(probs: np.ndarray, labels: np.ndarray, n_bins: int = 15) -> float:
    confidences = probs.max(axis=1)
    predictions = probs.argmax(axis=1)
    accuracies = (predictions == labels).astype(np.float32)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        mask = (confidences > bins[i]) & (confidences <= bins[i + 1])
        if mask.any():
            bin_acc = accuracies[mask].mean()
            bin_conf = confidences[mask].mean()
            ece += np.abs(bin_acc - bin_conf) * mask.mean()
    return float(ece)

def top1_accuracy(probs: np.ndarray, labels: np.ndarray) -> float:
    return float((probs.argmax(axis=1) == labels).mean() * 100.0)

def predictive_entropy(probs: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return -(probs * (probs + eps).log()).sum(dim=1)

def prediction_margin(probs: torch.Tensor) -> torch.Tensor:
    top2 = torch.topk(probs, k=2, dim=1).values
    return top2[:, 0] - top2[:, 1]

def kl_divergence_batch(p: torch.Tensor, q: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return (p * ((p + eps).log() - (q + eps).log())).sum(dim=1)

def softmax_np(logits: np.ndarray) -> np.ndarray:
    x = logits - logits.max(axis=1, keepdims=True)
    ex = np.exp(x)
    return ex / ex.sum(axis=1, keepdims=True)

def make_subset_indices(n: int, train_n: int, val_n: int, test_n: int, seed: int = 42):
    idx = np.arange(n)
    rng = np.random.default_rng(seed)
    rng.shuffle(idx)
    train_idx = idx[:train_n]
    val_idx = idx[train_n:train_n + val_n]
    test_idx = idx[train_n + val_n:train_n + val_n + test_n]
    return train_idx, val_idx, test_idx

def parse_soft_lambda(mode: str) -> Optional[float]:
    if mode.startswith("soft_"):
        return float(mode.split("_")[1])
    return None

def normalized_cost(budget: int) -> float:
    # K=1 -> 0, K=4 -> 3/7, K=8 -> 1
    return (budget - 1.0) / 7.0

# =========================
# Data
# =========================
mean = (0.5071, 0.4867, 0.4408)
std = (0.2675, 0.2565, 0.2761)

base_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.RandomHorizontalFlip(),
    T.RandomCrop(IMAGE_SIZE, padding=16),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_ds_full = CIFAR100(root="/content/data", train=True, download=True, transform=train_transform)
eval_train_ds_full = CIFAR100(root="/content/data", train=True, download=False, transform=base_transform)
test_ds_full = CIFAR100(root="/content/data", train=False, download=True, transform=base_transform)

train_idx, val_idx, _ = make_subset_indices(
    n=len(train_ds_full),
    train_n=TRAIN_POLICY_SUBSET,
    val_n=VAL_SUBSET,
    test_n=TEST_SUBSET,
    seed=SEED
)

head_train_ds = Subset(train_ds_full, train_idx)
policy_train_ds = Subset(eval_train_ds_full, train_idx)
val_ds = Subset(eval_train_ds_full, val_idx)
test_ds = Subset(test_ds_full, np.arange(TEST_SUBSET))

head_train_loader = DataLoader(head_train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
policy_train_loader = DataLoader(policy_train_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# =========================
# Model
# =========================
class BackboneWrapper(nn.Module):
    def __init__(self, backbone_name: str, num_classes: int = 100):
        super().__init__()
        if backbone_name == "resnet50":
            m = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
            feat_dim = m.fc.in_features
            m.fc = nn.Identity()
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        elif backbone_name == "deit_small_patch16_224":
            m = timm.create_model("deit_small_patch16_224", pretrained=True, num_classes=0)
            feat_dim = m.num_features
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        else:
            raise ValueError(f"Unsupported backbone: {backbone_name}")

    def forward_features(self, x: torch.Tensor) -> torch.Tensor:
        return self.backbone(x)

    def forward(self, x: torch.Tensor):
        feat = self.forward_features(x)
        logits = self.head(feat)
        return logits, feat

model = BackboneWrapper(BACKBONE_NAME, num_classes=NUM_CLASSES).to(DEVICE)

for p in model.backbone.parameters():
    p.requires_grad = False

if USE_PRETRAINED_HEAD:
    print(f"[Info] Training classifier head for {HEAD_EPOCHS} epoch(s)...")
    optimizer = torch.optim.Adam(model.head.parameters(), lr=HEAD_LR)
    model.train()
    for epoch in range(HEAD_EPOCHS):
        pbar = tqdm(head_train_loader, desc=f"Head train {epoch+1}/{HEAD_EPOCHS}")
        for images, labels in pbar:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            optimizer.zero_grad()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
                logits, _ = model(images)
                loss = F.cross_entropy(logits, labels)
            loss.backward()
            optimizer.step()
            pbar.set_postfix(loss=float(loss.detach().cpu()))
else:
    raise RuntimeError("Set USE_PRETRAINED_HEAD=True or add your checkpoint loading code.")

model.eval()

# =========================
# FIXED TTA helper
# =========================
@torch.no_grad()
def batch_tta_predict(model: nn.Module, images: torch.Tensor, k: int = 1):
    model.eval()
    probs_list = []
    feats_list = []

    def apply_det_tta(x: torch.Tensor, aug_id: int):
        if aug_id == 0:
            return x
        elif aug_id == 1:
            return torch.flip(x, dims=[3])
        elif aug_id == 2:
            x2 = x[:, :, :, 4:]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 3:
            x2 = x[:, :, :, :-4]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 4:
            x2 = x[:, :, 4:, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 5:
            x2 = x[:, :, :-4, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 6:
            H, W = x.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 7:
            x2 = torch.flip(x, dims=[3])
            H, W = x2.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x2[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        else:
            return x

    with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
        for aug_id in range(k):
            x_aug = apply_det_tta(images, aug_id)
            logits, feats = model(x_aug)
            probs = F.softmax(logits, dim=1)
            probs_list.append(probs.float())
            feats_list.append(feats.float())

    probs_stack = torch.stack(probs_list, dim=0)
    feats_stack = torch.stack(feats_list, dim=0)
    avg_probs = probs_stack.mean(dim=0)
    avg_feats = feats_stack.mean(dim=0)

    if k == 1:
        disagreement = torch.zeros(images.size(0), device=images.device)
    else:
        kls = []
        for i in range(k):
            kls.append(kl_divergence_batch(probs_stack[i], avg_probs))
        disagreement = torch.stack(kls, dim=0).mean(dim=0)

    return avg_probs, avg_feats, disagreement

# =========================
# Collect caches
# =========================
@torch.no_grad()
def collect_policy_candidates(loader: DataLoader, name: str) -> Dict[str, np.ndarray]:
    model.eval()
    out = {
        "labels": [],
        "p1_probs": [], "p4_probs": [], "p8_probs": [],
        "p1_feats": [], "p4_feats": [], "p8_feats": [],
        "p1_dis": [], "p4_dis": [], "p8_dis": [],
    }

    for images, labels in tqdm(loader, desc=f"Collect [{name}]"):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        probs1, feat1, dis1 = batch_tta_predict(model, images, k=1)
        probs4, feat4, dis4 = batch_tta_predict(model, images, k=4)
        probs8, feat8, dis8 = batch_tta_predict(model, images, k=8)

        out["labels"].append(labels.cpu().numpy())
        out["p1_probs"].append(probs1.cpu().numpy())
        out["p4_probs"].append(probs4.cpu().numpy())
        out["p8_probs"].append(probs8.cpu().numpy())

        out["p1_feats"].append(feat1.cpu().numpy())
        out["p4_feats"].append(feat4.cpu().numpy())
        out["p8_feats"].append(feat8.cpu().numpy())

        out["p1_dis"].append(dis1.cpu().numpy())
        out["p4_dis"].append(dis4.cpu().numpy())
        out["p8_dis"].append(dis8.cpu().numpy())

    for k in list(out.keys()):
        out[k] = np.concatenate(out[k], axis=0)
    return out

print("[Info] Collecting cached candidate outputs...")
train_cache = collect_policy_candidates(policy_train_loader, "train")
val_cache = collect_policy_candidates(val_loader, "val")
test_cache = collect_policy_candidates(test_loader, "test")

single_acc = top1_accuracy(test_cache["p1_probs"], test_cache["labels"])
tta4_acc = top1_accuracy(test_cache["p4_probs"], test_cache["labels"])
tta8_acc = top1_accuracy(test_cache["p8_probs"], test_cache["labels"])
print("\n=== QUICK TTA SANITY CHECK ===")
print(f"Single: {single_acc:.2f}")
print(f"TTA-4 : {tta4_acc:.2f}")
print(f"TTA-8 : {tta8_acc:.2f}")

# =========================
# Feature builder
# =========================
def build_feature_matrix(cache: Dict[str, np.ndarray], feature_set: List[str], source_budget: int = 1) -> np.ndarray:
    probs = cache[f"p{source_budget}_probs"]
    feats = cache[f"p{source_budget}_feats"]
    dis = cache["p4_dis"]

    probs_t = torch.from_numpy(probs)
    entropy = predictive_entropy(probs_t).numpy()[:, None]
    margin = prediction_margin(probs_t).numpy()[:, None]
    pmax = probs.max(axis=1, keepdims=True)
    disagreement = dis[:, None]

    xs = []
    for name in feature_set:
        if name == "entropy":
            xs.append(entropy)
        elif name == "margin":
            xs.append(margin)
            xs.append(pmax)
        elif name == "disagreement":
            xs.append(disagreement)
        elif name == "embedding":
            xs.append(feats)
        else:
            raise ValueError(f"Unknown feature: {name}")
    return np.concatenate(xs, axis=1).astype(np.float32)

# =========================
# Preferences
# =========================
def choose_preference(
    probs_a: np.ndarray,
    probs_b: np.ndarray,
    label: int,
    budget_a: int,
    budget_b: int,
    mode: str = "soft_0.10",
):
    pred_a = int(np.argmax(probs_a))
    pred_b = int(np.argmax(probs_b))
    acc_a = int(pred_a == label)
    acc_b = int(pred_b == label)

    if mode == "acc_only":
        if acc_a > acc_b:
            return budget_a, budget_b
        if acc_b > acc_a:
            return budget_b, budget_a
        return None

    lam = parse_soft_lambda(mode)
    if lam is not None:
        ua = float(acc_a) - lam * normalized_cost(budget_a)
        ub = float(acc_b) - lam * normalized_cost(budget_b)

        # prefer higher utility; if tie, prefer lower cost
        if ua > ub + 1e-12:
            return budget_a, budget_b
        elif ub > ua + 1e-12:
            return budget_b, budget_a
        else:
            if COSTS[budget_a] < COSTS[budget_b]:
                return budget_a, budget_b
            elif COSTS[budget_b] < COSTS[budget_a]:
                return budget_b, budget_a
            return None

    raise ValueError(f"Unknown mode: {mode}")

def build_preference_dataset(cache: Dict[str, np.ndarray], X: np.ndarray, mode: str = "soft_0.10"):
    labels = cache["labels"]
    xs, chosen, rejected = [], [], []

    for i in range(len(labels)):
        for a, b in [(1, 4), (1, 8), (4, 8)]:
            pref = choose_preference(
                cache[BUDGET_TO_KEY[a]][i],
                cache[BUDGET_TO_KEY[b]][i],
                int(labels[i]),
                a, b,
                mode=mode,
            )
            if pref is None:
                continue
            c, r = pref
            xs.append(X[i])
            chosen.append([1, 4, 8].index(c))
            rejected.append([1, 4, 8].index(r))

    xs = np.asarray(xs, dtype=np.float32)
    chosen = np.asarray(chosen, dtype=np.int64)
    rejected = np.asarray(rejected, dtype=np.int64)
    return xs, chosen, rejected

# =========================
# Policy net
# =========================
class PolicyNet(nn.Module):
    def __init__(self, input_dim: int, hidden_dim: int = 64, num_policies: int = 3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_policies),
        )

    def forward(self, x: torch.Tensor):
        return self.net(x)

def pairwise_pref_loss(scores: torch.Tensor, chosen: torch.Tensor, rejected: torch.Tensor) -> torch.Tensor:
    chosen_score = scores.gather(1, chosen[:, None]).squeeze(1)
    rejected_score = scores.gather(1, rejected[:, None]).squeeze(1)
    return -torch.log(torch.sigmoid(chosen_score - rejected_score) + 1e-12).mean()

def train_policy(X_train, ch_train, rj_train, X_val=None, ch_val=None, rj_val=None):
    pol = PolicyNet(input_dim=X_train.shape[1], hidden_dim=POLICY_HIDDEN).to(DEVICE)
    opt = torch.optim.Adam(pol.parameters(), lr=POLICY_LR)

    Xtr = torch.from_numpy(X_train)
    Ctr = torch.from_numpy(ch_train)
    Rtr = torch.from_numpy(rj_train)

    best_state = None
    best_val = 1e9

    for epoch in range(POLICY_EPOCHS):
        pol.train()
        perm = torch.randperm(len(Xtr))
        losses = []

        for start in range(0, len(Xtr), 1024):
            idx = perm[start:start + 1024]
            xb = Xtr[idx].to(DEVICE)
            cb = Ctr[idx].to(DEVICE)
            rb = Rtr[idx].to(DEVICE)

            opt.zero_grad()
            scores = pol(xb)
            loss = pairwise_pref_loss(scores, cb, rb)
            loss.backward()
            opt.step()
            losses.append(float(loss.detach().cpu()))

        val_loss = np.mean(losses)
        if X_val is not None and len(X_val) > 0:
            pol.eval()
            with torch.no_grad():
                xv = torch.from_numpy(X_val).to(DEVICE)
                cv = torch.from_numpy(ch_val).to(DEVICE)
                rv = torch.from_numpy(rj_val).to(DEVICE)
                scores = pol(xv)
                val_loss = float(pairwise_pref_loss(scores, cv, rv).detach().cpu())

        print(f"[Policy] epoch {epoch+1}/{POLICY_EPOCHS} train={np.mean(losses):.4f} val={val_loss:.4f}")

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in pol.state_dict().items()}

    if best_state is not None:
        pol.load_state_dict(best_state)

    pol.eval()
    return pol

# =========================
# Evaluation
# =========================
def evaluate_fixed(cache: Dict[str, np.ndarray], budget: int) -> Dict[str, float]:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    return {
        "method": f"TTA-{budget}" if budget > 1 else "Single",
        "avg_compute": float(COSTS[budget]),
        "acc": top1_accuracy(probs, labels),
        "ece": compute_ece(probs, labels),
    }

def apply_budget_selection(cache: Dict[str, np.ndarray], selected_budgets: np.ndarray, name: str) -> Dict[str, float]:
    labels = cache["labels"]
    final_probs = []
    costs = []

    for i, b in enumerate(selected_budgets.tolist()):
        final_probs.append(cache[BUDGET_TO_KEY[int(b)]][i])
        costs.append(COSTS[int(b)])

    final_probs = np.asarray(final_probs)
    return {
        "method": name,
        "avg_compute": float(np.mean(costs)),
        "acc": top1_accuracy(final_probs, labels),
        "ece": compute_ece(final_probs, labels),
    }

def tune_thresholds(values: np.ndarray, cache: Dict[str, np.ndarray], mode: str = "entropy", target_compute: Optional[float] = None):
    labels = cache["labels"]
    candidates = np.quantile(values, np.linspace(0.1, 0.9, 9))
    best = None
    best_score = -1e9

    for t1 in candidates:
        for t2 in candidates:
            if t2 < t1:
                continue

            if mode == "entropy":
                sel = np.where(values < t1, 1, np.where(values < t2, 4, 8))
            else:
                sel = np.where(values > t2, 1, np.where(values > t1, 4, 8))

            probs = np.stack([cache[BUDGET_TO_KEY[int(b)]][i] for i, b in enumerate(sel)], axis=0)
            acc = top1_accuracy(probs, labels)
            avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))

            score = acc
            if target_compute is not None:
                score -= 10.0 * abs(avg_compute - target_compute)

            if score > best_score:
                best_score = score
                best = (t1, t2, sel, acc, avg_compute)

    return best

def evaluate_random_policy(cache: Dict[str, np.ndarray], target_compute: float = 1.10, seed: int = 42):
    rng = np.random.default_rng(seed)
    best = None
    best_gap = 1e9

    probs_set = [
        np.array([0.98, 0.02, 0.00], dtype=np.float32),
        np.array([0.96, 0.04, 0.00], dtype=np.float32),
        np.array([0.94, 0.05, 0.01], dtype=np.float32),
        np.array([0.92, 0.06, 0.02], dtype=np.float32),
        np.array([0.90, 0.08, 0.02], dtype=np.float32),
        np.array([0.88, 0.10, 0.02], dtype=np.float32),
    ]

    for p in probs_set:
        sel_idx = rng.choice(3, size=len(cache["labels"]), p=p)
        sel = np.array([BUDGETS[i] for i in sel_idx])
        avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))
        gap = abs(avg_compute - target_compute)
        if gap < best_gap:
            best_gap = gap
            best = sel

    return apply_budget_selection(cache, best, f"Random(target={target_compute:.2f})")

@torch.no_grad()
def evaluate_policy_network(policy_model: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], name: str = "POTTS"):
    scores = policy_model(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    sel_idx = scores.argmax(axis=1)
    sel = np.array([BUDGETS[i] for i in sel_idx])
    return apply_budget_selection(cache, sel, name)

@torch.no_grad()
def evaluate_policy_with_temperature(policy_model: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], temperature: float, name: str):
    scores = policy_model(torch.from_numpy(X).to(DEVICE)).cpu().numpy() / temperature
    probs = softmax_np(scores)
    sel_idx = probs.argmax(axis=1)
    sel = np.array([BUDGETS[i] for i in sel_idx])
    return apply_budget_selection(cache, sel, name)

def search_policy_operating_points(policy_model: nn.Module, X_val: np.ndarray, cache_val: Dict[str, np.ndarray], targets: List[float]):
    rows = []
    temps = np.linspace(0.25, 4.0, 20)
    for target in targets:
        best = None
        best_gap = 1e9
        for t in temps:
            row = evaluate_policy_with_temperature(policy_model, X_val, cache_val, t, f"POTTS@T={t:.2f}")
            gap = abs(row["avg_compute"] - target)
            if gap < best_gap:
                best_gap = gap
                best = (t, row)
        rows.append({
            "target_compute": target,
            "temperature": best[0],
            **best[1]
        })
    return pd.DataFrame(rows)

# =========================
# Main experiments
# =========================
main_rows = []
main_rows.append(evaluate_fixed(test_cache, 1))
main_rows.append(evaluate_fixed(test_cache, 4))
main_rows.append(evaluate_fixed(test_cache, 8))
main_rows.append(evaluate_random_policy(test_cache, target_compute=1.10, seed=SEED))

val_entropy = predictive_entropy(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_entropy = predictive_entropy(torch.from_numpy(test_cache["p1_probs"])).numpy()

val_margin = prediction_margin(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_margin = prediction_margin(torch.from_numpy(test_cache["p1_probs"])).numpy()

ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=1.10)
t1_ent, t2_ent, _, _, _ = ent_best
ent_sel_test = np.where(test_entropy < t1_ent, 1, np.where(test_entropy < t2_ent, 4, 8))
main_rows.append(apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold"))

mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=1.10)
t1_mar, t2_mar, _, _, _ = mar_best
mar_sel_test = np.where(test_margin > t2_mar, 1, np.where(test_margin > t1_mar, 4, 8))
main_rows.append(apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold"))

ablation_rows = []
trained_policies = {}

for pref_mode in PREFERENCE_MODES:
    for feat_name, feat_list in FEATURE_ABLATIONS.items():
        print(f"\n[Info] Training policy: pref={pref_mode}, feat={feat_name}")
        X_train = build_feature_matrix(train_cache, feat_list, source_budget=1)
        X_val = build_feature_matrix(val_cache, feat_list, source_budget=1)
        X_test = build_feature_matrix(test_cache, feat_list, source_budget=1)

        Xtr, chtr, rjtr = build_preference_dataset(train_cache, X_train, mode=pref_mode)
        Xv, chv, rjv = build_preference_dataset(val_cache, X_val, mode=pref_mode)

        if len(Xtr) == 0:
            print("[Warn] No preference pairs. Skipping.")
            continue

        pol = train_policy(Xtr, chtr, rjtr, Xv, chv, rjv)
        trained_policies[(pref_mode, feat_name)] = pol

        row = evaluate_policy_network(pol, X_test, test_cache, name=f"POTTS[{pref_mode}|{feat_name}]")
        row["preference_mode"] = pref_mode
        row["feature_set"] = feat_name
        ablation_rows.append(row)

# changed: main POTTS = soft_0.10 + full
potts_main = [r for r in ablation_rows if r["preference_mode"] == "soft_0.10" and r["feature_set"] == "full"]
if len(potts_main) > 0:
    main_rows.append({
        "method": "POTTS",
        "avg_compute": potts_main[0]["avg_compute"],
        "acc": potts_main[0]["acc"],
        "ece": potts_main[0]["ece"],
    })

main_df = pd.DataFrame(main_rows).sort_values(["avg_compute", "acc"], ascending=[True, False]).reset_index(drop=True)
ablation_df = pd.DataFrame(ablation_rows)

print("\n=== MAIN RESULTS ===")
display(main_df)

print("\n=== ABLATION RESULTS ===")
display(ablation_df)

main_df.to_csv(os.path.join(RESULT_DIR, "main_results.csv"), index=False)
ablation_df.to_csv(os.path.join(RESULT_DIR, "ablation_results.csv"), index=False)

# =========================
# Trade-off curve
# =========================
curve_rows = []

for target in CURVE_TARGETS:
    ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=target)
    t1, t2, _, _, _ = ent_best
    ent_sel_test = np.where(test_entropy < t1, 1, np.where(test_entropy < t2, 4, 8))
    row = apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

    mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=target)
    t1m, t2m, _, _, _ = mar_best
    mar_sel_test = np.where(test_margin > t2m, 1, np.where(test_margin > t1m, 4, 8))
    row = apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

# changed: curve uses soft_0.10 + full
if ("soft_0.10", "full") in trained_policies:
    pol = trained_policies[("soft_0.10", "full")]
    X_val_full = build_feature_matrix(val_cache, FEATURE_ABLATIONS["full"], source_budget=1)
    X_test_full = build_feature_matrix(test_cache, FEATURE_ABLATIONS["full"], source_budget=1)
    curve_df_tmp = search_policy_operating_points(pol, X_val_full, val_cache, CURVE_TARGETS)

    for _, rr in curve_df_tmp.iterrows():
        row = evaluate_policy_with_temperature(pol, X_test_full, test_cache, rr["temperature"], "POTTS")
        row["target_compute"] = rr["target_compute"]
        row["temperature"] = rr["temperature"]
        curve_rows.append(row)

curve_df = pd.DataFrame(curve_rows)
curve_df.to_csv(os.path.join(RESULT_DIR, "tradeoff_curve.csv"), index=False)

print("\n=== TRADE-OFF CURVE ===")
display(curve_df)

# =========================
# Paper-friendly table
# =========================
paper_rows = []
for _, r in main_df.iterrows():
    paper_rows.append({
        "Method": r["method"],
        "Compute": round(float(r["avg_compute"]), 3),
        "Acc (%)": round(float(r["acc"]), 3),
        "ECE": round(float(r["ece"]), 4),
    })

paper_df = pd.DataFrame(paper_rows)
paper_df.to_csv(os.path.join(RESULT_DIR, "paper_table_main.csv"), index=False)

print("\nSaved files:")
for fn in sorted(os.listdir(RESULT_DIR)):
    print(" -", os.path.join(RESULT_DIR, fn))
```

```python
# =========================================================
# ONE-CELL COLAB: POTTS -> stable label-based budget classifier
# =========================================================
# Main idea:
# - Keep the fixed TTA and cached K={1,4,8} predictions
# - Replace pairwise preference training with direct 3-class classification
# - For each sample, assign a target budget by utility:
#       utility = correct - lam * normalized_cost
# - Train a budget classifier to predict {1,4,8} from features
#
# This is usually much more stable than pairwise preference optimization
# for a first paper-ready result.
#
# Outputs:
# /content/potts_results/main_results.csv
# /content/potts_results/ablation_results.csv
# /content/potts_results/tradeoff_curve.csv
# /content/potts_results/paper_table_main.csv
# =========================================================

!pip -q install timm pandas scikit-learn tqdm

import os
import random
import warnings
from typing import Dict, List, Tuple, Optional

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset

import torchvision
import torchvision.transforms as T
from torchvision.datasets import CIFAR100

import timm

warnings.filterwarnings("ignore")

# =========================
# Config
# =========================
SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
NUM_WORKERS = 2
BATCH_SIZE = 256

TRAIN_POLICY_SUBSET = 8000
VAL_SUBSET = 1000
TEST_SUBSET = 2000

IMAGE_SIZE = 224
NUM_CLASSES = 100

BACKBONE_NAME = "resnet50"
USE_PRETRAINED_HEAD = True
HEAD_EPOCHS = 1
HEAD_LR = 1e-3

POLICY_HIDDEN = 64
POLICY_EPOCHS = 6
POLICY_LR = 1e-3
POLICY_WEIGHT_DECAY = 1e-4

RESULT_DIR = "/content/potts_results"
os.makedirs(RESULT_DIR, exist_ok=True)

BUDGETS = [1, 4, 8]
COSTS = {1: 1.0, 4: 4.0, 8: 8.0}
BUDGET_TO_KEY = {1: "p1_probs", 4: "p4_probs", 8: "p8_probs"}
BUDGET_TO_CLASS = {1: 0, 4: 1, 8: 2}
CLASS_TO_BUDGET = {0: 1, 1: 4, 2: 8}

FEATURE_ABLATIONS = {
    "entropy_only": ["entropy"],
    "full": ["entropy", "margin", "disagreement", "embedding"],
}

# use utility lambdas instead of pairwise preference modes
UTILITY_MODES = ["acc_only", "util_0.02", "util_0.05", "util_0.08"]

CURVE_TARGETS = [1.05, 1.20, 2.00, 4.00]

# =========================
# Seed
# =========================
def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.benchmark = True

print("DEVICE =", DEVICE)

# =========================
# Metrics / utils
# =========================
def compute_ece(probs: np.ndarray, labels: np.ndarray, n_bins: int = 15) -> float:
    confidences = probs.max(axis=1)
    predictions = probs.argmax(axis=1)
    accuracies = (predictions == labels).astype(np.float32)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        mask = (confidences > bins[i]) & (confidences <= bins[i + 1])
        if mask.any():
            bin_acc = accuracies[mask].mean()
            bin_conf = confidences[mask].mean()
            ece += np.abs(bin_acc - bin_conf) * mask.mean()
    return float(ece)

def top1_accuracy(probs: np.ndarray, labels: np.ndarray) -> float:
    return float((probs.argmax(axis=1) == labels).mean() * 100.0)

def predictive_entropy(probs: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return -(probs * (probs + eps).log()).sum(dim=1)

def prediction_margin(probs: torch.Tensor) -> torch.Tensor:
    top2 = torch.topk(probs, k=2, dim=1).values
    return top2[:, 0] - top2[:, 1]

def kl_divergence_batch(p: torch.Tensor, q: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return (p * ((p + eps).log() - (q + eps).log())).sum(dim=1)

def softmax_np(logits: np.ndarray) -> np.ndarray:
    x = logits - logits.max(axis=1, keepdims=True)
    ex = np.exp(x)
    return ex / ex.sum(axis=1, keepdims=True)

def make_subset_indices(n: int, train_n: int, val_n: int, test_n: int, seed: int = 42):
    idx = np.arange(n)
    rng = np.random.default_rng(seed)
    rng.shuffle(idx)
    train_idx = idx[:train_n]
    val_idx = idx[train_n:train_n + val_n]
    test_idx = idx[train_n + val_n:train_n + val_n + test_n]
    return train_idx, val_idx, test_idx

def parse_util_lambda(mode: str) -> Optional[float]:
    if mode.startswith("util_"):
        return float(mode.split("_")[1])
    return None

def normalized_cost(budget: int) -> float:
    # K=1 -> 0, K=4 -> 3/7, K=8 -> 1
    return (budget - 1.0) / 7.0

# =========================
# Data
# =========================
mean = (0.5071, 0.4867, 0.4408)
std = (0.2675, 0.2565, 0.2761)

base_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.RandomHorizontalFlip(),
    T.RandomCrop(IMAGE_SIZE, padding=16),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_ds_full = CIFAR100(root="/content/data", train=True, download=True, transform=train_transform)
eval_train_ds_full = CIFAR100(root="/content/data", train=True, download=False, transform=base_transform)
test_ds_full = CIFAR100(root="/content/data", train=False, download=True, transform=base_transform)

train_idx, val_idx, _ = make_subset_indices(
    n=len(train_ds_full),
    train_n=TRAIN_POLICY_SUBSET,
    val_n=VAL_SUBSET,
    test_n=TEST_SUBSET,
    seed=SEED
)

head_train_ds = Subset(train_ds_full, train_idx)
policy_train_ds = Subset(eval_train_ds_full, train_idx)
val_ds = Subset(eval_train_ds_full, val_idx)
test_ds = Subset(test_ds_full, np.arange(TEST_SUBSET))

head_train_loader = DataLoader(head_train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
policy_train_loader = DataLoader(policy_train_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# =========================
# Model
# =========================
class BackboneWrapper(nn.Module):
    def __init__(self, backbone_name: str, num_classes: int = 100):
        super().__init__()
        if backbone_name == "resnet50":
            m = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
            feat_dim = m.fc.in_features
            m.fc = nn.Identity()
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        elif backbone_name == "deit_small_patch16_224":
            m = timm.create_model("deit_small_patch16_224", pretrained=True, num_classes=0)
            feat_dim = m.num_features
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        else:
            raise ValueError(f"Unsupported backbone: {backbone_name}")

    def forward_features(self, x: torch.Tensor) -> torch.Tensor:
        return self.backbone(x)

    def forward(self, x: torch.Tensor):
        feat = self.forward_features(x)
        logits = self.head(feat)
        return logits, feat

model = BackboneWrapper(BACKBONE_NAME, num_classes=NUM_CLASSES).to(DEVICE)

for p in model.backbone.parameters():
    p.requires_grad = False

if USE_PRETRAINED_HEAD:
    print(f"[Info] Training classifier head for {HEAD_EPOCHS} epoch(s)...")
    optimizer = torch.optim.Adam(model.head.parameters(), lr=HEAD_LR)
    model.train()
    for epoch in range(HEAD_EPOCHS):
        pbar = tqdm(head_train_loader, desc=f"Head train {epoch+1}/{HEAD_EPOCHS}")
        for images, labels in pbar:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            optimizer.zero_grad()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
                logits, _ = model(images)
                loss = F.cross_entropy(logits, labels)
            loss.backward()
            optimizer.step()
            pbar.set_postfix(loss=float(loss.detach().cpu()))
else:
    raise RuntimeError("Set USE_PRETRAINED_HEAD=True or add your checkpoint loading code.")

model.eval()

# =========================
# Fixed TTA helper
# =========================
@torch.no_grad()
def batch_tta_predict(model: nn.Module, images: torch.Tensor, k: int = 1):
    model.eval()
    probs_list = []
    feats_list = []

    def apply_det_tta(x: torch.Tensor, aug_id: int):
        if aug_id == 0:
            return x
        elif aug_id == 1:
            return torch.flip(x, dims=[3])
        elif aug_id == 2:
            x2 = x[:, :, :, 4:]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 3:
            x2 = x[:, :, :, :-4]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 4:
            x2 = x[:, :, 4:, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 5:
            x2 = x[:, :, :-4, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 6:
            H, W = x.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 7:
            x2 = torch.flip(x, dims=[3])
            H, W = x2.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x2[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        else:
            return x

    with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
        for aug_id in range(k):
            x_aug = apply_det_tta(images, aug_id)
            logits, feats = model(x_aug)
            probs = F.softmax(logits, dim=1)
            probs_list.append(probs.float())
            feats_list.append(feats.float())

    probs_stack = torch.stack(probs_list, dim=0)
    feats_stack = torch.stack(feats_list, dim=0)
    avg_probs = probs_stack.mean(dim=0)
    avg_feats = feats_stack.mean(dim=0)

    if k == 1:
        disagreement = torch.zeros(images.size(0), device=images.device)
    else:
        kls = []
        for i in range(k):
            kls.append(kl_divergence_batch(probs_stack[i], avg_probs))
        disagreement = torch.stack(kls, dim=0).mean(dim=0)

    return avg_probs, avg_feats, disagreement

# =========================
# Collect caches
# =========================
@torch.no_grad()
def collect_policy_candidates(loader: DataLoader, name: str) -> Dict[str, np.ndarray]:
    model.eval()
    out = {
        "labels": [],
        "p1_probs": [], "p4_probs": [], "p8_probs": [],
        "p1_feats": [], "p4_feats": [], "p8_feats": [],
        "p1_dis": [], "p4_dis": [], "p8_dis": [],
    }

    for images, labels in tqdm(loader, desc=f"Collect [{name}]"):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        probs1, feat1, dis1 = batch_tta_predict(model, images, k=1)
        probs4, feat4, dis4 = batch_tta_predict(model, images, k=4)
        probs8, feat8, dis8 = batch_tta_predict(model, images, k=8)

        out["labels"].append(labels.cpu().numpy())
        out["p1_probs"].append(probs1.cpu().numpy())
        out["p4_probs"].append(probs4.cpu().numpy())
        out["p8_probs"].append(probs8.cpu().numpy())

        out["p1_feats"].append(feat1.cpu().numpy())
        out["p4_feats"].append(feat4.cpu().numpy())
        out["p8_feats"].append(feat8.cpu().numpy())

        out["p1_dis"].append(dis1.cpu().numpy())
        out["p4_dis"].append(dis4.cpu().numpy())
        out["p8_dis"].append(dis8.cpu().numpy())

    for k in list(out.keys()):
        out[k] = np.concatenate(out[k], axis=0)
    return out

print("[Info] Collecting cached candidate outputs...")
train_cache = collect_policy_candidates(policy_train_loader, "train")
val_cache = collect_policy_candidates(val_loader, "val")
test_cache = collect_policy_candidates(test_loader, "test")

single_acc = top1_accuracy(test_cache["p1_probs"], test_cache["labels"])
tta4_acc = top1_accuracy(test_cache["p4_probs"], test_cache["labels"])
tta8_acc = top1_accuracy(test_cache["p8_probs"], test_cache["labels"])
print("\n=== QUICK TTA SANITY CHECK ===")
print(f"Single: {single_acc:.2f}")
print(f"TTA-4 : {tta4_acc:.2f}")
print(f"TTA-8 : {tta8_acc:.2f}")

# =========================
# Feature builder
# =========================
def build_feature_matrix(cache: Dict[str, np.ndarray], feature_set: List[str], source_budget: int = 1) -> np.ndarray:
    probs = cache[f"p{source_budget}_probs"]
    feats = cache[f"p{source_budget}_feats"]
    dis = cache["p4_dis"]

    probs_t = torch.from_numpy(probs)
    entropy = predictive_entropy(probs_t).numpy()[:, None]
    margin = prediction_margin(probs_t).numpy()[:, None]
    pmax = probs.max(axis=1, keepdims=True)
    disagreement = dis[:, None]

    xs = []
    for name in feature_set:
        if name == "entropy":
            xs.append(entropy)
        elif name == "margin":
            xs.append(margin)
            xs.append(pmax)
        elif name == "disagreement":
            xs.append(disagreement)
        elif name == "embedding":
            xs.append(feats)
        else:
            raise ValueError(f"Unknown feature: {name}")
    return np.concatenate(xs, axis=1).astype(np.float32)

# =========================
# Label construction by utility
# =========================
def sample_correct(cache: Dict[str, np.ndarray], budget: int) -> np.ndarray:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    preds = probs.argmax(axis=1)
    return (preds == labels).astype(np.float32)

def build_budget_targets(cache: Dict[str, np.ndarray], mode: str):
    """
    mode:
      - acc_only
      - util_0.02
      - util_0.05
      - util_0.08
    """
    n = len(cache["labels"])

    corr1 = sample_correct(cache, 1)
    corr4 = sample_correct(cache, 4)
    corr8 = sample_correct(cache, 8)

    util = np.zeros((n, 3), dtype=np.float32)

    if mode == "acc_only":
        util[:, 0] = corr1
        util[:, 1] = corr4
        util[:, 2] = corr8
    else:
        lam = parse_util_lambda(mode)
        if lam is None:
            raise ValueError(f"Unknown utility mode: {mode}")

        util[:, 0] = corr1 - lam * normalized_cost(1)
        util[:, 1] = corr4 - lam * normalized_cost(4)
        util[:, 2] = corr8 - lam * normalized_cost(8)

    # tie-break: prefer lower cost
    # small penalties make argmax choose lower-cost class under ties
    util[:, 0] += 1e-6
    util[:, 1] += 5e-7
    util[:, 2] += 0.0

    y = util.argmax(axis=1).astype(np.int64)
    return y, util

# =========================
# Budget classifier
# =========================
class BudgetClassifier(nn.Module):
    def __init__(self, input_dim: int, hidden_dim: int = 64, num_classes: int = 3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_classes),
        )

    def forward(self, x: torch.Tensor):
        return self.net(x)

def compute_class_weights(y: np.ndarray) -> torch.Tensor:
    counts = np.bincount(y, minlength=3).astype(np.float32)
    counts[counts == 0] = 1.0
    weights = counts.sum() / counts
    weights = weights / weights.mean()
    return torch.tensor(weights, dtype=torch.float32)

def train_budget_classifier(X_train, y_train, X_val=None, y_val=None):
    clf = BudgetClassifier(input_dim=X_train.shape[1], hidden_dim=POLICY_HIDDEN).to(DEVICE)
    class_weights = compute_class_weights(y_train).to(DEVICE)
    opt = torch.optim.Adam(clf.parameters(), lr=POLICY_LR, weight_decay=POLICY_WEIGHT_DECAY)

    Xtr = torch.from_numpy(X_train)
    Ytr = torch.from_numpy(y_train)

    best_state = None
    best_val = 1e9

    for epoch in range(POLICY_EPOCHS):
        clf.train()
        perm = torch.randperm(len(Xtr))
        losses = []

        for start in range(0, len(Xtr), 1024):
            idx = perm[start:start + 1024]
            xb = Xtr[idx].to(DEVICE)
            yb = Ytr[idx].to(DEVICE)

            opt.zero_grad()
            logits = clf(xb)
            loss = F.cross_entropy(logits, yb, weight=class_weights)
            loss.backward()
            opt.step()
            losses.append(float(loss.detach().cpu()))

        val_loss = np.mean(losses)

        if X_val is not None and y_val is not None and len(X_val) > 0:
            clf.eval()
            with torch.no_grad():
                xv = torch.from_numpy(X_val).to(DEVICE)
                yv = torch.from_numpy(y_val).to(DEVICE)
                logits = clf(xv)
                val_loss = float(F.cross_entropy(logits, yv).detach().cpu())

        print(f"[Classifier] epoch {epoch+1}/{POLICY_EPOCHS} train={np.mean(losses):.4f} val={val_loss:.4f}")

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in clf.state_dict().items()}

    if best_state is not None:
        clf.load_state_dict(best_state)

    clf.eval()
    return clf

# =========================
# Evaluation
# =========================
def evaluate_fixed(cache: Dict[str, np.ndarray], budget: int) -> Dict[str, float]:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    return {
        "method": f"TTA-{budget}" if budget > 1 else "Single",
        "avg_compute": float(COSTS[budget]),
        "acc": top1_accuracy(probs, labels),
        "ece": compute_ece(probs, labels),
    }

def apply_budget_selection(cache: Dict[str, np.ndarray], selected_budgets: np.ndarray, name: str) -> Dict[str, float]:
    labels = cache["labels"]
    final_probs = []
    costs = []

    for i, b in enumerate(selected_budgets.tolist()):
        final_probs.append(cache[BUDGET_TO_KEY[int(b)]][i])
        costs.append(COSTS[int(b)])

    final_probs = np.asarray(final_probs)
    return {
        "method": name,
        "avg_compute": float(np.mean(costs)),
        "acc": top1_accuracy(final_probs, labels),
        "ece": compute_ece(final_probs, labels),
    }

def tune_thresholds(values: np.ndarray, cache: Dict[str, np.ndarray], mode: str = "entropy", target_compute: Optional[float] = None):
    labels = cache["labels"]
    candidates = np.quantile(values, np.linspace(0.1, 0.9, 9))
    best = None
    best_score = -1e9

    for t1 in candidates:
        for t2 in candidates:
            if t2 < t1:
                continue

            if mode == "entropy":
                sel = np.where(values < t1, 1, np.where(values < t2, 4, 8))
            else:
                sel = np.where(values > t2, 1, np.where(values > t1, 4, 8))

            probs = np.stack([cache[BUDGET_TO_KEY[int(b)]][i] for i, b in enumerate(sel)], axis=0)
            acc = top1_accuracy(probs, labels)
            avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))

            score = acc
            if target_compute is not None:
                score -= 10.0 * abs(avg_compute - target_compute)

            if score > best_score:
                best_score = score
                best = (t1, t2, sel, acc, avg_compute)

    return best

def evaluate_random_policy(cache: Dict[str, np.ndarray], target_compute: float = 1.10, seed: int = 42):
    rng = np.random.default_rng(seed)
    best = None
    best_gap = 1e9

    probs_set = [
        np.array([0.98, 0.02, 0.00], dtype=np.float32),
        np.array([0.96, 0.04, 0.00], dtype=np.float32),
        np.array([0.94, 0.05, 0.01], dtype=np.float32),
        np.array([0.92, 0.06, 0.02], dtype=np.float32),
        np.array([0.90, 0.08, 0.02], dtype=np.float32),
        np.array([0.88, 0.10, 0.02], dtype=np.float32),
    ]

    for p in probs_set:
        sel_idx = rng.choice(3, size=len(cache["labels"]), p=p)
        sel = np.array([BUDGETS[i] for i in sel_idx])
        avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))
        gap = abs(avg_compute - target_compute)
        if gap < best_gap:
            best_gap = gap
            best = sel

    return apply_budget_selection(cache, best, f"Random(target={target_compute:.2f})")

@torch.no_grad()
def evaluate_classifier(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], name: str = "POTTS-CLS"):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, name)

@torch.no_grad()
def evaluate_classifier_with_bias(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], bias_vec: np.ndarray, name: str):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    logits = logits + bias_vec[None, :]
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, name)

def search_classifier_operating_points(clf: nn.Module, X_val: np.ndarray, cache_val: Dict[str, np.ndarray], targets: List[float]):
    rows = []

    # bias the logits of classes {1,4,8}; positive bias on bigger budgets increases compute
    grid = [-2.0, -1.0, -0.5, 0.0, 0.5, 1.0, 2.0]
    bias_candidates = []
    for b4 in grid:
        for b8 in grid:
            bias_candidates.append(np.array([0.0, b4, b8], dtype=np.float32))

    for target in targets:
        best = None
        best_gap = 1e9
        for bias in bias_candidates:
            row = evaluate_classifier_with_bias(clf, X_val, cache_val, bias, "POTTS-CLS")
            gap = abs(row["avg_compute"] - target)
            if gap < best_gap:
                best_gap = gap
                best = (bias, row)
        rows.append({
            "target_compute": target,
            "bias_1": float(best[0][0]),
            "bias_4": float(best[0][1]),
            "bias_8": float(best[0][2]),
            **best[1]
        })
    return pd.DataFrame(rows)

# =========================
# Main experiments
# =========================
main_rows = []
main_rows.append(evaluate_fixed(test_cache, 1))
main_rows.append(evaluate_fixed(test_cache, 4))
main_rows.append(evaluate_fixed(test_cache, 8))
main_rows.append(evaluate_random_policy(test_cache, target_compute=1.10, seed=SEED))

val_entropy = predictive_entropy(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_entropy = predictive_entropy(torch.from_numpy(test_cache["p1_probs"])).numpy()

val_margin = prediction_margin(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_margin = prediction_margin(torch.from_numpy(test_cache["p1_probs"])).numpy()

ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=1.10)
t1_ent, t2_ent, _, _, _ = ent_best
ent_sel_test = np.where(test_entropy < t1_ent, 1, np.where(test_entropy < t2_ent, 4, 8))
main_rows.append(apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold"))

mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=1.10)
t1_mar, t2_mar, _, _, _ = mar_best
mar_sel_test = np.where(test_margin > t2_mar, 1, np.where(test_margin > t1_mar, 4, 8))
main_rows.append(apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold"))

ablation_rows = []
trained_classifiers = {}

for util_mode in UTILITY_MODES:
    for feat_name, feat_list in FEATURE_ABLATIONS.items():
        print(f"\n[Info] Training classifier: utility={util_mode}, feat={feat_name}")

        X_train = build_feature_matrix(train_cache, feat_list, source_budget=1)
        X_val = build_feature_matrix(val_cache, feat_list, source_budget=1)
        X_test = build_feature_matrix(test_cache, feat_list, source_budget=1)

        y_train, util_train = build_budget_targets(train_cache, util_mode)
        y_val, util_val = build_budget_targets(val_cache, util_mode)

        print("Train label counts:", np.bincount(y_train, minlength=3))
        print("Val   label counts:", np.bincount(y_val, minlength=3))

        clf = train_budget_classifier(X_train, y_train, X_val, y_val)
        trained_classifiers[(util_mode, feat_name)] = clf

        row = evaluate_classifier(clf, X_test, test_cache, name=f"POTTS-CLS[{util_mode}|{feat_name}]")
        row["utility_mode"] = util_mode
        row["feature_set"] = feat_name
        ablation_rows.append(row)

# choose main result from util_0.05 + full
main_key = ("util_0.05", "full")
if main_key in trained_classifiers:
    X_test_full = build_feature_matrix(test_cache, FEATURE_ABLATIONS["full"], source_budget=1)
    row = evaluate_classifier(trained_classifiers[main_key], X_test_full, test_cache, name="POTTS-CLS")
    main_rows.append(row)

main_df = pd.DataFrame(main_rows).sort_values(["avg_compute", "acc"], ascending=[True, False]).reset_index(drop=True)
ablation_df = pd.DataFrame(ablation_rows)

print("\n=== MAIN RESULTS ===")
display(main_df)

print("\n=== ABLATION RESULTS ===")
display(ablation_df)

main_df.to_csv(os.path.join(RESULT_DIR, "main_results.csv"), index=False)
ablation_df.to_csv(os.path.join(RESULT_DIR, "ablation_results.csv"), index=False)

# =========================
# Trade-off curve
# =========================
curve_rows = []

for target in CURVE_TARGETS:
    ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=target)
    t1, t2, _, _, _ = ent_best
    ent_sel_test = np.where(test_entropy < t1, 1, np.where(test_entropy < t2, 4, 8))
    row = apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

    mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=target)
    t1m, t2m, _, _, _ = mar_best
    mar_sel_test = np.where(test_margin > t2m, 1, np.where(test_margin > t1m, 4, 8))
    row = apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

if main_key in trained_classifiers:
    clf = trained_classifiers[main_key]
    X_val_full = build_feature_matrix(val_cache, FEATURE_ABLATIONS["full"], source_budget=1)
    X_test_full = build_feature_matrix(test_cache, FEATURE_ABLATIONS["full"], source_budget=1)

    curve_df_tmp = search_classifier_operating_points(clf, X_val_full, val_cache, CURVE_TARGETS)

    for _, rr in curve_df_tmp.iterrows():
        bias = np.array([rr["bias_1"], rr["bias_4"], rr["bias_8"]], dtype=np.float32)
        row = evaluate_classifier_with_bias(clf, X_test_full, test_cache, bias, "POTTS-CLS")
        row["target_compute"] = rr["target_compute"]
        row["bias_1"] = rr["bias_1"]
        row["bias_4"] = rr["bias_4"]
        row["bias_8"] = rr["bias_8"]
        curve_rows.append(row)

curve_df = pd.DataFrame(curve_rows)
curve_df.to_csv(os.path.join(RESULT_DIR, "tradeoff_curve.csv"), index=False)

print("\n=== TRADE-OFF CURVE ===")
display(curve_df)

# =========================
# Paper-friendly table
# =========================
paper_rows = []
for _, r in main_df.iterrows():
    paper_rows.append({
        "Method": r["method"],
        "Compute": round(float(r["avg_compute"]), 3),
        "Acc (%)": round(float(r["acc"]), 3),
        "ECE": round(float(r["ece"]), 4),
    })

paper_df = pd.DataFrame(paper_rows)
paper_df.to_csv(os.path.join(RESULT_DIR, "paper_table_main.csv"), index=False)

print("\nSaved files:")
for fn in sorted(os.listdir(RESULT_DIR)):
    print(" -", os.path.join(RESULT_DIR, fn))
```

```python
# =========================================================
# ONE-CELL COLAB: label-based budget classifier + lambda sweep
# =========================================================
# What this version does:
# - fixed TTA (no noisy TTA)
# - direct 3-class budget classifier for budgets {1,4,8}
# - utility-label sweep over lambdas: 0.01 / 0.03 / 0.06 / 0.08
# - feature ablation: entropy_only vs full
# - heuristic baselines: entropy-threshold / margin-threshold / random
# - operating-point search for learned classifier with logit bias
#
# Outputs:
# /content/potts_results/main_results.csv
# /content/potts_results/ablation_results.csv
# /content/potts_results/tradeoff_curve.csv
# /content/potts_results/paper_table_main.csv
# =========================================================

!pip -q install timm pandas scikit-learn tqdm

import os
import random
import warnings
from typing import Dict, List, Optional

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset

import torchvision
import torchvision.transforms as T
from torchvision.datasets import CIFAR100

import timm

warnings.filterwarnings("ignore")

# =========================
# Config
# =========================
SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
NUM_WORKERS = 2
BATCH_SIZE = 256

TRAIN_POLICY_SUBSET = 8000
VAL_SUBSET = 1000
TEST_SUBSET = 2000

IMAGE_SIZE = 224
NUM_CLASSES = 100

BACKBONE_NAME = "resnet50"
USE_PRETRAINED_HEAD = True
HEAD_EPOCHS = 1
HEAD_LR = 1e-3

POLICY_HIDDEN = 64
POLICY_EPOCHS = 6
POLICY_LR = 1e-3
POLICY_WEIGHT_DECAY = 1e-4

RESULT_DIR = "/content/potts_results"
os.makedirs(RESULT_DIR, exist_ok=True)

BUDGETS = [1, 4, 8]
COSTS = {1: 1.0, 4: 4.0, 8: 8.0}
BUDGET_TO_KEY = {1: "p1_probs", 4: "p4_probs", 8: "p8_probs"}
CLASS_TO_BUDGET = {0: 1, 1: 4, 2: 8}

FEATURE_ABLATIONS = {
    "entropy_only": ["entropy"],
    "full": ["entropy", "margin", "disagreement", "embedding"],
}

UTILITY_MODES = ["acc_only", "util_0.01", "util_0.03", "util_0.06", "util_0.08"]
CURVE_TARGETS = [1.05, 1.20, 2.00, 4.00]

# default main key to report as POTTS-CLS
PREFERRED_MAIN_KEYS = [
    ("util_0.06", "full"),
    ("util_0.03", "full"),
    ("util_0.01", "full"),
    ("util_0.08", "full"),
]

# =========================
# Seed
# =========================
def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.benchmark = True

print("DEVICE =", DEVICE)

# =========================
# Metrics / utils
# =========================
def compute_ece(probs: np.ndarray, labels: np.ndarray, n_bins: int = 15) -> float:
    confidences = probs.max(axis=1)
    predictions = probs.argmax(axis=1)
    accuracies = (predictions == labels).astype(np.float32)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        mask = (confidences > bins[i]) & (confidences <= bins[i + 1])
        if mask.any():
            bin_acc = accuracies[mask].mean()
            bin_conf = confidences[mask].mean()
            ece += np.abs(bin_acc - bin_conf) * mask.mean()
    return float(ece)

def top1_accuracy(probs: np.ndarray, labels: np.ndarray) -> float:
    return float((probs.argmax(axis=1) == labels).mean() * 100.0)

def predictive_entropy(probs: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return -(probs * (probs + eps).log()).sum(dim=1)

def prediction_margin(probs: torch.Tensor) -> torch.Tensor:
    top2 = torch.topk(probs, k=2, dim=1).values
    return top2[:, 0] - top2[:, 1]

def kl_divergence_batch(p: torch.Tensor, q: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return (p * ((p + eps).log() - (q + eps).log())).sum(dim=1)

def make_subset_indices(n: int, train_n: int, val_n: int, test_n: int, seed: int = 42):
    idx = np.arange(n)
    rng = np.random.default_rng(seed)
    rng.shuffle(idx)
    train_idx = idx[:train_n]
    val_idx = idx[train_n:train_n + val_n]
    test_idx = idx[train_n + val_n:train_n + val_n + test_n]
    return train_idx, val_idx, test_idx

def parse_util_lambda(mode: str) -> Optional[float]:
    if mode.startswith("util_"):
        return float(mode.split("_")[1])
    return None

def normalized_cost(budget: int) -> float:
    return (budget - 1.0) / 7.0  # 1->0, 4->3/7, 8->1

# =========================
# Data
# =========================
mean = (0.5071, 0.4867, 0.4408)
std = (0.2675, 0.2565, 0.2761)

base_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.RandomHorizontalFlip(),
    T.RandomCrop(IMAGE_SIZE, padding=16),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_ds_full = CIFAR100(root="/content/data", train=True, download=True, transform=train_transform)
eval_train_ds_full = CIFAR100(root="/content/data", train=True, download=False, transform=base_transform)
test_ds_full = CIFAR100(root="/content/data", train=False, download=True, transform=base_transform)

train_idx, val_idx, _ = make_subset_indices(
    n=len(train_ds_full),
    train_n=TRAIN_POLICY_SUBSET,
    val_n=VAL_SUBSET,
    test_n=TEST_SUBSET,
    seed=SEED
)

head_train_ds = Subset(train_ds_full, train_idx)
policy_train_ds = Subset(eval_train_ds_full, train_idx)
val_ds = Subset(eval_train_ds_full, val_idx)
test_ds = Subset(test_ds_full, np.arange(TEST_SUBSET))

head_train_loader = DataLoader(head_train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
policy_train_loader = DataLoader(policy_train_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# =========================
# Backbone + head
# =========================
class BackboneWrapper(nn.Module):
    def __init__(self, backbone_name: str, num_classes: int = 100):
        super().__init__()
        if backbone_name == "resnet50":
            m = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
            feat_dim = m.fc.in_features
            m.fc = nn.Identity()
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        elif backbone_name == "deit_small_patch16_224":
            m = timm.create_model("deit_small_patch16_224", pretrained=True, num_classes=0)
            feat_dim = m.num_features
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        else:
            raise ValueError(f"Unsupported backbone: {backbone_name}")

    def forward_features(self, x: torch.Tensor) -> torch.Tensor:
        return self.backbone(x)

    def forward(self, x: torch.Tensor):
        feat = self.forward_features(x)
        logits = self.head(feat)
        return logits, feat

model = BackboneWrapper(BACKBONE_NAME, num_classes=NUM_CLASSES).to(DEVICE)

for p in model.backbone.parameters():
    p.requires_grad = False

if USE_PRETRAINED_HEAD:
    print(f"[Info] Training classifier head for {HEAD_EPOCHS} epoch(s)...")
    optimizer = torch.optim.Adam(model.head.parameters(), lr=HEAD_LR)
    model.train()
    for epoch in range(HEAD_EPOCHS):
        pbar = tqdm(head_train_loader, desc=f"Head train {epoch+1}/{HEAD_EPOCHS}")
        for images, labels in pbar:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            optimizer.zero_grad()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
                logits, _ = model(images)
                loss = F.cross_entropy(logits, labels)
            loss.backward()
            optimizer.step()
            pbar.set_postfix(loss=float(loss.detach().cpu()))
else:
    raise RuntimeError("Set USE_PRETRAINED_HEAD=True or load your own checkpoint.")

model.eval()

# =========================
# Fixed TTA helper
# =========================
@torch.no_grad()
def batch_tta_predict(model: nn.Module, images: torch.Tensor, k: int = 1):
    model.eval()
    probs_list = []
    feats_list = []

    def apply_det_tta(x: torch.Tensor, aug_id: int):
        if aug_id == 0:
            return x
        elif aug_id == 1:
            return torch.flip(x, dims=[3])
        elif aug_id == 2:
            x2 = x[:, :, :, 4:]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 3:
            x2 = x[:, :, :, :-4]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 4:
            x2 = x[:, :, 4:, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 5:
            x2 = x[:, :, :-4, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 6:
            H, W = x.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 7:
            x2 = torch.flip(x, dims=[3])
            H, W = x2.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x2[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        else:
            return x

    with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
        for aug_id in range(k):
            x_aug = apply_det_tta(images, aug_id)
            logits, feats = model(x_aug)
            probs = F.softmax(logits, dim=1)
            probs_list.append(probs.float())
            feats_list.append(feats.float())

    probs_stack = torch.stack(probs_list, dim=0)
    feats_stack = torch.stack(feats_list, dim=0)
    avg_probs = probs_stack.mean(dim=0)
    avg_feats = feats_stack.mean(dim=0)

    if k == 1:
        disagreement = torch.zeros(images.size(0), device=images.device)
    else:
        kls = []
        for i in range(k):
            kls.append(kl_divergence_batch(probs_stack[i], avg_probs))
        disagreement = torch.stack(kls, dim=0).mean(dim=0)

    return avg_probs, avg_feats, disagreement

# =========================
# Cache candidate outputs
# =========================
@torch.no_grad()
def collect_policy_candidates(loader: DataLoader, name: str) -> Dict[str, np.ndarray]:
    model.eval()
    out = {
        "labels": [],
        "p1_probs": [], "p4_probs": [], "p8_probs": [],
        "p1_feats": [], "p4_feats": [], "p8_feats": [],
        "p1_dis": [], "p4_dis": [], "p8_dis": [],
    }

    for images, labels in tqdm(loader, desc=f"Collect [{name}]"):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        probs1, feat1, dis1 = batch_tta_predict(model, images, k=1)
        probs4, feat4, dis4 = batch_tta_predict(model, images, k=4)
        probs8, feat8, dis8 = batch_tta_predict(model, images, k=8)

        out["labels"].append(labels.cpu().numpy())
        out["p1_probs"].append(probs1.cpu().numpy())
        out["p4_probs"].append(probs4.cpu().numpy())
        out["p8_probs"].append(probs8.cpu().numpy())
        out["p1_feats"].append(feat1.cpu().numpy())
        out["p4_feats"].append(feat4.cpu().numpy())
        out["p8_feats"].append(feat8.cpu().numpy())
        out["p1_dis"].append(dis1.cpu().numpy())
        out["p4_dis"].append(dis4.cpu().numpy())
        out["p8_dis"].append(dis8.cpu().numpy())

    for k in list(out.keys()):
        out[k] = np.concatenate(out[k], axis=0)
    return out

print("[Info] Collecting cached candidate outputs...")
train_cache = collect_policy_candidates(policy_train_loader, "train")
val_cache = collect_policy_candidates(val_loader, "val")
test_cache = collect_policy_candidates(test_loader, "test")

print("\n=== QUICK TTA SANITY CHECK ===")
print(f"Single: {top1_accuracy(test_cache['p1_probs'], test_cache['labels']):.2f}")
print(f"TTA-4 : {top1_accuracy(test_cache['p4_probs'], test_cache['labels']):.2f}")
print(f"TTA-8 : {top1_accuracy(test_cache['p8_probs'], test_cache['labels']):.2f}")

# =========================
# Feature builder
# =========================
def build_feature_matrix(cache: Dict[str, np.ndarray], feature_set: List[str], source_budget: int = 1) -> np.ndarray:
    probs = cache[f"p{source_budget}_probs"]
    feats = cache[f"p{source_budget}_feats"]
    dis = cache["p4_dis"]

    probs_t = torch.from_numpy(probs)
    entropy = predictive_entropy(probs_t).numpy()[:, None]
    margin = prediction_margin(probs_t).numpy()[:, None]
    pmax = probs.max(axis=1, keepdims=True)
    disagreement = dis[:, None]

    xs = []
    for name in feature_set:
        if name == "entropy":
            xs.append(entropy)
        elif name == "margin":
            xs.append(margin)
            xs.append(pmax)
        elif name == "disagreement":
            xs.append(disagreement)
        elif name == "embedding":
            xs.append(feats)
        else:
            raise ValueError(f"Unknown feature: {name}")
    return np.concatenate(xs, axis=1).astype(np.float32)

# =========================
# Utility labels
# =========================
def sample_correct(cache: Dict[str, np.ndarray], budget: int) -> np.ndarray:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    preds = probs.argmax(axis=1)
    return (preds == labels).astype(np.float32)

def build_budget_targets(cache: Dict[str, np.ndarray], mode: str):
    n = len(cache["labels"])
    corr1 = sample_correct(cache, 1)
    corr4 = sample_correct(cache, 4)
    corr8 = sample_correct(cache, 8)

    util = np.zeros((n, 3), dtype=np.float32)

    if mode == "acc_only":
        util[:, 0] = corr1
        util[:, 1] = corr4
        util[:, 2] = corr8
    else:
        lam = parse_util_lambda(mode)
        if lam is None:
            raise ValueError(f"Unknown utility mode: {mode}")
        util[:, 0] = corr1 - lam * normalized_cost(1)
        util[:, 1] = corr4 - lam * normalized_cost(4)
        util[:, 2] = corr8 - lam * normalized_cost(8)

    # tie-break toward lower cost
    util[:, 0] += 1e-6
    util[:, 1] += 5e-7
    util[:, 2] += 0.0

    y = util.argmax(axis=1).astype(np.int64)
    return y, util

# =========================
# Budget classifier
# =========================
class BudgetClassifier(nn.Module):
    def __init__(self, input_dim: int, hidden_dim: int = 64, num_classes: int = 3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_classes),
        )

    def forward(self, x: torch.Tensor):
        return self.net(x)

def compute_class_weights(y: np.ndarray) -> torch.Tensor:
    counts = np.bincount(y, minlength=3).astype(np.float32)
    counts[counts == 0] = 1.0
    weights = counts.sum() / counts
    weights = weights / weights.mean()
    return torch.tensor(weights, dtype=torch.float32)

def train_budget_classifier(X_train, y_train, X_val=None, y_val=None):
    clf = BudgetClassifier(input_dim=X_train.shape[1], hidden_dim=POLICY_HIDDEN).to(DEVICE)
    class_weights = compute_class_weights(y_train).to(DEVICE)
    opt = torch.optim.Adam(clf.parameters(), lr=POLICY_LR, weight_decay=POLICY_WEIGHT_DECAY)

    Xtr = torch.from_numpy(X_train)
    Ytr = torch.from_numpy(y_train)

    best_state = None
    best_val = 1e9

    for epoch in range(POLICY_EPOCHS):
        clf.train()
        perm = torch.randperm(len(Xtr))
        losses = []

        for start in range(0, len(Xtr), 1024):
            idx = perm[start:start + 1024]
            xb = Xtr[idx].to(DEVICE)
            yb = Ytr[idx].to(DEVICE)

            opt.zero_grad()
            logits = clf(xb)
            loss = F.cross_entropy(logits, yb, weight=class_weights)
            loss.backward()
            opt.step()
            losses.append(float(loss.detach().cpu()))

        val_loss = np.mean(losses)
        if X_val is not None and y_val is not None and len(X_val) > 0:
            clf.eval()
            with torch.no_grad():
                xv = torch.from_numpy(X_val).to(DEVICE)
                yv = torch.from_numpy(y_val).to(DEVICE)
                logits = clf(xv)
                val_loss = float(F.cross_entropy(logits, yv).detach().cpu())

        print(f"[Classifier] epoch {epoch+1}/{POLICY_EPOCHS} train={np.mean(losses):.4f} val={val_loss:.4f}")

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in clf.state_dict().items()}

    if best_state is not None:
        clf.load_state_dict(best_state)

    clf.eval()
    return clf

# =========================
# Evaluation helpers
# =========================
def evaluate_fixed(cache: Dict[str, np.ndarray], budget: int) -> Dict[str, float]:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    return {
        "method": f"TTA-{budget}" if budget > 1 else "Single",
        "avg_compute": float(COSTS[budget]),
        "acc": top1_accuracy(probs, labels),
        "ece": compute_ece(probs, labels),
    }

def apply_budget_selection(cache: Dict[str, np.ndarray], selected_budgets: np.ndarray, name: str) -> Dict[str, float]:
    labels = cache["labels"]
    final_probs = []
    costs = []

    for i, b in enumerate(selected_budgets.tolist()):
        final_probs.append(cache[BUDGET_TO_KEY[int(b)]][i])
        costs.append(COSTS[int(b)])

    final_probs = np.asarray(final_probs)
    return {
        "method": name,
        "avg_compute": float(np.mean(costs)),
        "acc": top1_accuracy(final_probs, labels),
        "ece": compute_ece(final_probs, labels),
    }

def tune_thresholds(values: np.ndarray, cache: Dict[str, np.ndarray], mode: str = "entropy", target_compute: Optional[float] = None):
    labels = cache["labels"]
    candidates = np.quantile(values, np.linspace(0.1, 0.9, 9))
    best = None
    best_score = -1e9

    for t1 in candidates:
        for t2 in candidates:
            if t2 < t1:
                continue

            if mode == "entropy":
                sel = np.where(values < t1, 1, np.where(values < t2, 4, 8))
            else:
                sel = np.where(values > t2, 1, np.where(values > t1, 4, 8))

            probs = np.stack([cache[BUDGET_TO_KEY[int(b)]][i] for i, b in enumerate(sel)], axis=0)
            acc = top1_accuracy(probs, labels)
            avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))

            score = acc
            if target_compute is not None:
                score -= 10.0 * abs(avg_compute - target_compute)

            if score > best_score:
                best_score = score
                best = (t1, t2, sel, acc, avg_compute)

    return best

def evaluate_random_policy(cache: Dict[str, np.ndarray], target_compute: float = 1.10, seed: int = 42):
    rng = np.random.default_rng(seed)
    best = None
    best_gap = 1e9

    probs_set = [
        np.array([0.98, 0.02, 0.00], dtype=np.float32),
        np.array([0.96, 0.04, 0.00], dtype=np.float32),
        np.array([0.94, 0.05, 0.01], dtype=np.float32),
        np.array([0.92, 0.06, 0.02], dtype=np.float32),
        np.array([0.90, 0.08, 0.02], dtype=np.float32),
        np.array([0.88, 0.10, 0.02], dtype=np.float32),
    ]

    for p in probs_set:
        sel_idx = rng.choice(3, size=len(cache["labels"]), p=p)
        sel = np.array([BUDGETS[i] for i in sel_idx])
        avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))
        gap = abs(avg_compute - target_compute)
        if gap < best_gap:
            best_gap = gap
            best = sel

    return apply_budget_selection(cache, best, f"Random(target={target_compute:.2f})")

@torch.no_grad()
def evaluate_classifier(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], name: str = "POTTS-CLS"):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, name)

@torch.no_grad()
def evaluate_classifier_with_bias(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], bias_vec: np.ndarray, name: str):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    logits = logits + bias_vec[None, :]
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, name)

def search_classifier_operating_points(clf: nn.Module, X_val: np.ndarray, cache_val: Dict[str, np.ndarray], targets: List[float]):
    rows = []
    grid = [-2.0, -1.0, -0.5, 0.0, 0.5, 1.0, 2.0]
    bias_candidates = []
    for b4 in grid:
        for b8 in grid:
            bias_candidates.append(np.array([0.0, b4, b8], dtype=np.float32))

    for target in targets:
        best = None
        best_gap = 1e9
        for bias in bias_candidates:
            row = evaluate_classifier_with_bias(clf, X_val, cache_val, bias, "POTTS-CLS")
            gap = abs(row["avg_compute"] - target)
            if gap < best_gap:
                best_gap = gap
                best = (bias, row)
        rows.append({
            "target_compute": target,
            "bias_1": float(best[0][0]),
            "bias_4": float(best[0][1]),
            "bias_8": float(best[0][2]),
            **best[1]
        })
    return pd.DataFrame(rows)

# =========================
# Main experiments
# =========================
main_rows = []
main_rows.append(evaluate_fixed(test_cache, 1))
main_rows.append(evaluate_fixed(test_cache, 4))
main_rows.append(evaluate_fixed(test_cache, 8))
main_rows.append(evaluate_random_policy(test_cache, target_compute=1.10, seed=SEED))

val_entropy = predictive_entropy(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_entropy = predictive_entropy(torch.from_numpy(test_cache["p1_probs"])).numpy()

val_margin = prediction_margin(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_margin = prediction_margin(torch.from_numpy(test_cache["p1_probs"])).numpy()

ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=1.10)
t1_ent, t2_ent, _, _, _ = ent_best
ent_sel_test = np.where(test_entropy < t1_ent, 1, np.where(test_entropy < t2_ent, 4, 8))
main_rows.append(apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold"))

mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=1.10)
t1_mar, t2_mar, _, _, _ = mar_best
mar_sel_test = np.where(test_margin > t2_mar, 1, np.where(test_margin > t1_mar, 4, 8))
main_rows.append(apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold"))

ablation_rows = []
trained_classifiers = {}

for util_mode in UTILITY_MODES:
    for feat_name, feat_list in FEATURE_ABLATIONS.items():
        print(f"\n[Info] Training classifier: utility={util_mode}, feat={feat_name}")

        X_train = build_feature_matrix(train_cache, feat_list, source_budget=1)
        X_val = build_feature_matrix(val_cache, feat_list, source_budget=1)
        X_test = build_feature_matrix(test_cache, feat_list, source_budget=1)

        y_train, _ = build_budget_targets(train_cache, util_mode)
        y_val, _ = build_budget_targets(val_cache, util_mode)

        print("Train label counts:", np.bincount(y_train, minlength=3))
        print("Val   label counts:", np.bincount(y_val, minlength=3))

        clf = train_budget_classifier(X_train, y_train, X_val, y_val)
        trained_classifiers[(util_mode, feat_name)] = clf

        row = evaluate_classifier(clf, X_test, test_cache, name=f"POTTS-CLS[{util_mode}|{feat_name}]")
        row["utility_mode"] = util_mode
        row["feature_set"] = feat_name
        ablation_rows.append(row)

ablation_df = pd.DataFrame(ablation_rows)

# choose a main key from preferred list; fallback to best full-feature accuracy
main_key = None
for k in PREFERRED_MAIN_KEYS:
    if k in trained_classifiers:
        main_key = k
        break

if main_key is None:
    full_rows = ablation_df[ablation_df["feature_set"] == "full"]
    if len(full_rows) > 0:
        best_row = full_rows.sort_values(["acc", "avg_compute"], ascending=[False, True]).iloc[0]
        main_key = (best_row["utility_mode"], best_row["feature_set"])

if main_key is not None:
    print(f"\n[Info] Using main key: {main_key}")
    X_test_main = build_feature_matrix(test_cache, FEATURE_ABLATIONS[main_key[1]], source_budget=1)
    row = evaluate_classifier(trained_classifiers[main_key], X_test_main, test_cache, name="POTTS-CLS")
    row["main_key"] = str(main_key)
    main_rows.append(row)

main_df = pd.DataFrame(main_rows).sort_values(["avg_compute", "acc"], ascending=[True, False]).reset_index(drop=True)

print("\n=== MAIN RESULTS ===")
display(main_df)

print("\n=== ABLATION RESULTS ===")
display(ablation_df)

main_df.to_csv(os.path.join(RESULT_DIR, "main_results.csv"), index=False)
ablation_df.to_csv(os.path.join(RESULT_DIR, "ablation_results.csv"), index=False)

# =========================
# Trade-off curve
# =========================
curve_rows = []

for target in CURVE_TARGETS:
    ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=target)
    t1, t2, _, _, _ = ent_best
    ent_sel_test = np.where(test_entropy < t1, 1, np.where(test_entropy < t2, 4, 8))
    row = apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

    mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=target)
    t1m, t2m, _, _, _ = mar_best
    mar_sel_test = np.where(test_margin > t2m, 1, np.where(test_margin > t1m, 4, 8))
    row = apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

if main_key is not None:
    clf = trained_classifiers[main_key]
    X_val_main = build_feature_matrix(val_cache, FEATURE_ABLATIONS[main_key[1]], source_budget=1)
    X_test_main = build_feature_matrix(test_cache, FEATURE_ABLATIONS[main_key[1]], source_budget=1)

    curve_df_tmp = search_classifier_operating_points(clf, X_val_main, val_cache, CURVE_TARGETS)

    for _, rr in curve_df_tmp.iterrows():
        bias = np.array([rr["bias_1"], rr["bias_4"], rr["bias_8"]], dtype=np.float32)
        row = evaluate_classifier_with_bias(clf, X_test_main, test_cache, bias, "POTTS-CLS")
        row["target_compute"] = rr["target_compute"]
        row["bias_1"] = rr["bias_1"]
        row["bias_4"] = rr["bias_4"]
        row["bias_8"] = rr["bias_8"]
        curve_rows.append(row)

curve_df = pd.DataFrame(curve_rows)
curve_df.to_csv(os.path.join(RESULT_DIR, "tradeoff_curve.csv"), index=False)

print("\n=== TRADE-OFF CURVE ===")
display(curve_df)

# =========================
# Paper-friendly main table
# =========================
paper_rows = []
for _, r in main_df.iterrows():
    paper_rows.append({
        "Method": r["method"],
        "Compute": round(float(r["avg_compute"]), 3),
        "Acc (%)": round(float(r["acc"]), 3),
        "ECE": round(float(r["ece"]), 4),
    })

paper_df = pd.DataFrame(paper_rows)
paper_df.to_csv(os.path.join(RESULT_DIR, "paper_table_main.csv"), index=False)

print("\nSaved files:")
for fn in sorted(os.listdir(RESULT_DIR)):
    print(" -", os.path.join(RESULT_DIR, fn))
```

```python
# =========================================================
# ONE-CELL COLAB: POTTS-CLS matched-compute lambda sweep
# =========================================================
# Purpose:
# - fixed TTA
# - direct budget classifier on {1,4,8}
# - sweep utility lambdas: 0.01 / 0.03 / 0.04 / 0.05 / 0.06 / 0.07 / 0.08
# - compare against entropy-threshold / margin-threshold / random
# - automatically search matched-compute operating points
# - report BEST learned point under matched compute against heuristic baselines
#
# Key output files:
# /content/potts_results/main_results.csv
# /content/potts_results/ablation_results.csv
# /content/potts_results/tradeoff_curve.csv
# /content/potts_results/matched_compute_summary.csv
# /content/potts_results/paper_table_main.csv
# =========================================================

!pip -q install timm pandas scikit-learn tqdm

import os
import random
import warnings
from typing import Dict, List, Optional, Tuple

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset

import torchvision
import torchvision.transforms as T
from torchvision.datasets import CIFAR100

import timm

warnings.filterwarnings("ignore")

# =========================
# Config
# =========================
SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
NUM_WORKERS = 2
BATCH_SIZE = 256

TRAIN_POLICY_SUBSET = 8000
VAL_SUBSET = 1000
TEST_SUBSET = 2000

IMAGE_SIZE = 224
NUM_CLASSES = 100

BACKBONE_NAME = "resnet50"
USE_PRETRAINED_HEAD = True
HEAD_EPOCHS = 1
HEAD_LR = 1e-3

POLICY_HIDDEN = 64
POLICY_EPOCHS = 6
POLICY_LR = 1e-3
POLICY_WEIGHT_DECAY = 1e-4

RESULT_DIR = "/content/potts_results"
os.makedirs(RESULT_DIR, exist_ok=True)

BUDGETS = [1, 4, 8]
COSTS = {1: 1.0, 4: 4.0, 8: 8.0}
BUDGET_TO_KEY = {1: "p1_probs", 4: "p4_probs", 8: "p8_probs"}
CLASS_TO_BUDGET = {0: 1, 1: 4, 2: 8}

FEATURE_ABLATIONS = {
    "entropy_only": ["entropy"],
    "full": ["entropy", "margin", "disagreement", "embedding"],
}

UTILITY_MODES = [
    "acc_only",
    "util_0.01",
    "util_0.03",
    "util_0.04",
    "util_0.05",
    "util_0.06",
    "util_0.07",
    "util_0.08",
]

CURVE_TARGETS = [1.05, 1.20, 1.50, 1.70, 2.00, 2.50, 3.00, 4.00]

# Matched-compute comparison anchors
MATCH_TARGETS = [1.10, 1.20, 1.50, 1.70, 2.00, 2.50, 3.00, 4.00]

# =========================
# Seed
# =========================
def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.benchmark = True

print("DEVICE =", DEVICE)

# =========================
# Metrics / utils
# =========================
def compute_ece(probs: np.ndarray, labels: np.ndarray, n_bins: int = 15) -> float:
    confidences = probs.max(axis=1)
    predictions = probs.argmax(axis=1)
    accuracies = (predictions == labels).astype(np.float32)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        mask = (confidences > bins[i]) & (confidences <= bins[i + 1])
        if mask.any():
            bin_acc = accuracies[mask].mean()
            bin_conf = confidences[mask].mean()
            ece += np.abs(bin_acc - bin_conf) * mask.mean()
    return float(ece)

def top1_accuracy(probs: np.ndarray, labels: np.ndarray) -> float:
    return float((probs.argmax(axis=1) == labels).mean() * 100.0)

def predictive_entropy(probs: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return -(probs * (probs + eps).log()).sum(dim=1)

def prediction_margin(probs: torch.Tensor) -> torch.Tensor:
    top2 = torch.topk(probs, k=2, dim=1).values
    return top2[:, 0] - top2[:, 1]

def kl_divergence_batch(p: torch.Tensor, q: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return (p * ((p + eps).log() - (q + eps).log())).sum(dim=1)

def make_subset_indices(n: int, train_n: int, val_n: int, test_n: int, seed: int = 42):
    idx = np.arange(n)
    rng = np.random.default_rng(seed)
    rng.shuffle(idx)
    train_idx = idx[:train_n]
    val_idx = idx[train_n:train_n + val_n]
    test_idx = idx[train_n + val_n:train_n + val_n + test_n]
    return train_idx, val_idx, test_idx

def parse_util_lambda(mode: str) -> Optional[float]:
    if mode.startswith("util_"):
        return float(mode.split("_")[1])
    return None

def normalized_cost(budget: int) -> float:
    return (budget - 1.0) / 7.0

# =========================
# Data
# =========================
mean = (0.5071, 0.4867, 0.4408)
std = (0.2675, 0.2565, 0.2761)

base_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.RandomHorizontalFlip(),
    T.RandomCrop(IMAGE_SIZE, padding=16),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_ds_full = CIFAR100(root="/content/data", train=True, download=True, transform=train_transform)
eval_train_ds_full = CIFAR100(root="/content/data", train=True, download=False, transform=base_transform)
test_ds_full = CIFAR100(root="/content/data", train=False, download=True, transform=base_transform)

train_idx, val_idx, _ = make_subset_indices(
    n=len(train_ds_full),
    train_n=TRAIN_POLICY_SUBSET,
    val_n=VAL_SUBSET,
    test_n=TEST_SUBSET,
    seed=SEED
)

head_train_ds = Subset(train_ds_full, train_idx)
policy_train_ds = Subset(eval_train_ds_full, train_idx)
val_ds = Subset(eval_train_ds_full, val_idx)
test_ds = Subset(test_ds_full, np.arange(TEST_SUBSET))

head_train_loader = DataLoader(head_train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
policy_train_loader = DataLoader(policy_train_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# =========================
# Backbone + head
# =========================
class BackboneWrapper(nn.Module):
    def __init__(self, backbone_name: str, num_classes: int = 100):
        super().__init__()
        if backbone_name == "resnet50":
            m = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
            feat_dim = m.fc.in_features
            m.fc = nn.Identity()
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        elif backbone_name == "deit_small_patch16_224":
            m = timm.create_model("deit_small_patch16_224", pretrained=True, num_classes=0)
            feat_dim = m.num_features
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        else:
            raise ValueError(f"Unsupported backbone: {backbone_name}")

    def forward_features(self, x: torch.Tensor) -> torch.Tensor:
        return self.backbone(x)

    def forward(self, x: torch.Tensor):
        feat = self.forward_features(x)
        logits = self.head(feat)
        return logits, feat

model = BackboneWrapper(BACKBONE_NAME, num_classes=NUM_CLASSES).to(DEVICE)

for p in model.backbone.parameters():
    p.requires_grad = False

if USE_PRETRAINED_HEAD:
    print(f"[Info] Training classifier head for {HEAD_EPOCHS} epoch(s)...")
    optimizer = torch.optim.Adam(model.head.parameters(), lr=HEAD_LR)
    model.train()
    for epoch in range(HEAD_EPOCHS):
        pbar = tqdm(head_train_loader, desc=f"Head train {epoch+1}/{HEAD_EPOCHS}")
        for images, labels in pbar:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            optimizer.zero_grad()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
                logits, _ = model(images)
                loss = F.cross_entropy(logits, labels)
            loss.backward()
            optimizer.step()
            pbar.set_postfix(loss=float(loss.detach().cpu()))
else:
    raise RuntimeError("Set USE_PRETRAINED_HEAD=True or load your own checkpoint.")

model.eval()

# =========================
# Fixed TTA helper
# =========================
@torch.no_grad()
def batch_tta_predict(model: nn.Module, images: torch.Tensor, k: int = 1):
    model.eval()
    probs_list = []
    feats_list = []

    def apply_det_tta(x: torch.Tensor, aug_id: int):
        if aug_id == 0:
            return x
        elif aug_id == 1:
            return torch.flip(x, dims=[3])
        elif aug_id == 2:
            x2 = x[:, :, :, 4:]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 3:
            x2 = x[:, :, :, :-4]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 4:
            x2 = x[:, :, 4:, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 5:
            x2 = x[:, :, :-4, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 6:
            H, W = x.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 7:
            x2 = torch.flip(x, dims=[3])
            H, W = x2.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x2[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        else:
            return x

    with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
        for aug_id in range(k):
            x_aug = apply_det_tta(images, aug_id)
            logits, feats = model(x_aug)
            probs = F.softmax(logits, dim=1)
            probs_list.append(probs.float())
            feats_list.append(feats.float())

    probs_stack = torch.stack(probs_list, dim=0)
    feats_stack = torch.stack(feats_list, dim=0)
    avg_probs = probs_stack.mean(dim=0)
    avg_feats = feats_stack.mean(dim=0)

    if k == 1:
        disagreement = torch.zeros(images.size(0), device=images.device)
    else:
        kls = []
        for i in range(k):
            kls.append(kl_divergence_batch(probs_stack[i], avg_probs))
        disagreement = torch.stack(kls, dim=0).mean(dim=0)

    return avg_probs, avg_feats, disagreement

# =========================
# Cache candidate outputs
# =========================
@torch.no_grad()
def collect_policy_candidates(loader: DataLoader, name: str) -> Dict[str, np.ndarray]:
    model.eval()
    out = {
        "labels": [],
        "p1_probs": [], "p4_probs": [], "p8_probs": [],
        "p1_feats": [], "p4_feats": [], "p8_feats": [],
        "p1_dis": [], "p4_dis": [], "p8_dis": [],
    }

    for images, labels in tqdm(loader, desc=f"Collect [{name}]"):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        probs1, feat1, dis1 = batch_tta_predict(model, images, k=1)
        probs4, feat4, dis4 = batch_tta_predict(model, images, k=4)
        probs8, feat8, dis8 = batch_tta_predict(model, images, k=8)

        out["labels"].append(labels.cpu().numpy())
        out["p1_probs"].append(probs1.cpu().numpy())
        out["p4_probs"].append(probs4.cpu().numpy())
        out["p8_probs"].append(probs8.cpu().numpy())
        out["p1_feats"].append(feat1.cpu().numpy())
        out["p4_feats"].append(feat4.cpu().numpy())
        out["p8_feats"].append(feat8.cpu().numpy())
        out["p1_dis"].append(dis1.cpu().numpy())
        out["p4_dis"].append(dis4.cpu().numpy())
        out["p8_dis"].append(dis8.cpu().numpy())

    for k in list(out.keys()):
        out[k] = np.concatenate(out[k], axis=0)
    return out

print("[Info] Collecting cached candidate outputs...")
train_cache = collect_policy_candidates(policy_train_loader, "train")
val_cache = collect_policy_candidates(val_loader, "val")
test_cache = collect_policy_candidates(test_loader, "test")

print("\n=== QUICK TTA SANITY CHECK ===")
print(f"Single: {top1_accuracy(test_cache['p1_probs'], test_cache['labels']):.2f}")
print(f"TTA-4 : {top1_accuracy(test_cache['p4_probs'], test_cache['labels']):.2f}")
print(f"TTA-8 : {top1_accuracy(test_cache['p8_probs'], test_cache['labels']):.2f}")

# =========================
# Feature builder
# =========================
def build_feature_matrix(cache: Dict[str, np.ndarray], feature_set: List[str], source_budget: int = 1) -> np.ndarray:
    probs = cache[f"p{source_budget}_probs"]
    feats = cache[f"p{source_budget}_feats"]
    dis = cache["p4_dis"]

    probs_t = torch.from_numpy(probs)
    entropy = predictive_entropy(probs_t).numpy()[:, None]
    margin = prediction_margin(probs_t).numpy()[:, None]
    pmax = probs.max(axis=1, keepdims=True)
    disagreement = dis[:, None]

    xs = []
    for name in feature_set:
        if name == "entropy":
            xs.append(entropy)
        elif name == "margin":
            xs.append(margin)
            xs.append(pmax)
        elif name == "disagreement":
            xs.append(disagreement)
        elif name == "embedding":
            xs.append(feats)
        else:
            raise ValueError(f"Unknown feature: {name}")
    return np.concatenate(xs, axis=1).astype(np.float32)

# =========================
# Utility labels
# =========================
def sample_correct(cache: Dict[str, np.ndarray], budget: int) -> np.ndarray:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    preds = probs.argmax(axis=1)
    return (preds == labels).astype(np.float32)

def build_budget_targets(cache: Dict[str, np.ndarray], mode: str):
    n = len(cache["labels"])
    corr1 = sample_correct(cache, 1)
    corr4 = sample_correct(cache, 4)
    corr8 = sample_correct(cache, 8)

    util = np.zeros((n, 3), dtype=np.float32)

    if mode == "acc_only":
        util[:, 0] = corr1
        util[:, 1] = corr4
        util[:, 2] = corr8
    else:
        lam = parse_util_lambda(mode)
        if lam is None:
            raise ValueError(f"Unknown utility mode: {mode}")
        util[:, 0] = corr1 - lam * normalized_cost(1)
        util[:, 1] = corr4 - lam * normalized_cost(4)
        util[:, 2] = corr8 - lam * normalized_cost(8)

    util[:, 0] += 1e-6
    util[:, 1] += 5e-7
    util[:, 2] += 0.0

    y = util.argmax(axis=1).astype(np.int64)
    return y, util

# =========================
# Budget classifier
# =========================
class BudgetClassifier(nn.Module):
    def __init__(self, input_dim: int, hidden_dim: int = 64, num_classes: int = 3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_classes),
        )

    def forward(self, x: torch.Tensor):
        return self.net(x)

def compute_class_weights(y: np.ndarray) -> torch.Tensor:
    counts = np.bincount(y, minlength=3).astype(np.float32)
    counts[counts == 0] = 1.0
    weights = counts.sum() / counts
    weights = weights / weights.mean()
    return torch.tensor(weights, dtype=torch.float32)

def train_budget_classifier(X_train, y_train, X_val=None, y_val=None):
    clf = BudgetClassifier(input_dim=X_train.shape[1], hidden_dim=POLICY_HIDDEN).to(DEVICE)
    class_weights = compute_class_weights(y_train).to(DEVICE)
    opt = torch.optim.Adam(clf.parameters(), lr=POLICY_LR, weight_decay=POLICY_WEIGHT_DECAY)

    Xtr = torch.from_numpy(X_train)
    Ytr = torch.from_numpy(y_train)

    best_state = None
    best_val = 1e9

    for epoch in range(POLICY_EPOCHS):
        clf.train()
        perm = torch.randperm(len(Xtr))
        losses = []

        for start in range(0, len(Xtr), 1024):
            idx = perm[start:start + 1024]
            xb = Xtr[idx].to(DEVICE)
            yb = Ytr[idx].to(DEVICE)

            opt.zero_grad()
            logits = clf(xb)
            loss = F.cross_entropy(logits, yb, weight=class_weights)
            loss.backward()
            opt.step()
            losses.append(float(loss.detach().cpu()))

        val_loss = np.mean(losses)
        if X_val is not None and y_val is not None and len(X_val) > 0:
            clf.eval()
            with torch.no_grad():
                xv = torch.from_numpy(X_val).to(DEVICE)
                yv = torch.from_numpy(y_val).to(DEVICE)
                logits = clf(xv)
                val_loss = float(F.cross_entropy(logits, yv).detach().cpu())

        print(f"[Classifier] epoch {epoch+1}/{POLICY_EPOCHS} train={np.mean(losses):.4f} val={val_loss:.4f}")

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in clf.state_dict().items()}

    if best_state is not None:
        clf.load_state_dict(best_state)

    clf.eval()
    return clf

# =========================
# Evaluation helpers
# =========================
def evaluate_fixed(cache: Dict[str, np.ndarray], budget: int) -> Dict[str, float]:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    return {
        "method": f"TTA-{budget}" if budget > 1 else "Single",
        "avg_compute": float(COSTS[budget]),
        "acc": top1_accuracy(probs, labels),
        "ece": compute_ece(probs, labels),
    }

def apply_budget_selection(cache: Dict[str, np.ndarray], selected_budgets: np.ndarray, name: str) -> Dict[str, float]:
    labels = cache["labels"]
    final_probs = []
    costs = []

    for i, b in enumerate(selected_budgets.tolist()):
        final_probs.append(cache[BUDGET_TO_KEY[int(b)]][i])
        costs.append(COSTS[int(b)])

    final_probs = np.asarray(final_probs)
    return {
        "method": name,
        "avg_compute": float(np.mean(costs)),
        "acc": top1_accuracy(final_probs, labels),
        "ece": compute_ece(final_probs, labels),
    }

def tune_thresholds(values: np.ndarray, cache: Dict[str, np.ndarray], mode: str = "entropy", target_compute: Optional[float] = None):
    labels = cache["labels"]
    candidates = np.quantile(values, np.linspace(0.1, 0.9, 9))
    best = None
    best_score = -1e9

    for t1 in candidates:
        for t2 in candidates:
            if t2 < t1:
                continue

            if mode == "entropy":
                sel = np.where(values < t1, 1, np.where(values < t2, 4, 8))
            else:
                sel = np.where(values > t2, 1, np.where(values > t1, 4, 8))

            probs = np.stack([cache[BUDGET_TO_KEY[int(b)]][i] for i, b in enumerate(sel)], axis=0)
            acc = top1_accuracy(probs, labels)
            avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))

            score = acc
            if target_compute is not None:
                score -= 10.0 * abs(avg_compute - target_compute)

            if score > best_score:
                best_score = score
                best = (t1, t2, sel, acc, avg_compute)

    return best

def evaluate_random_policy(cache: Dict[str, np.ndarray], target_compute: float = 1.10, seed: int = 42):
    rng = np.random.default_rng(seed)
    best = None
    best_gap = 1e9

    probs_set = [
        np.array([0.98, 0.02, 0.00], dtype=np.float32),
        np.array([0.96, 0.04, 0.00], dtype=np.float32),
        np.array([0.94, 0.05, 0.01], dtype=np.float32),
        np.array([0.92, 0.06, 0.02], dtype=np.float32),
        np.array([0.90, 0.08, 0.02], dtype=np.float32),
        np.array([0.88, 0.10, 0.02], dtype=np.float32),
    ]

    for p in probs_set:
        sel_idx = rng.choice(3, size=len(cache["labels"]), p=p)
        sel = np.array([BUDGETS[i] for i in sel_idx])
        avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))
        gap = abs(avg_compute - target_compute)
        if gap < best_gap:
            best_gap = gap
            best = sel

    return apply_budget_selection(cache, best, f"Random(target={target_compute:.2f})")

@torch.no_grad()
def evaluate_classifier(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], name: str = "POTTS-CLS"):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, name)

@torch.no_grad()
def evaluate_classifier_with_bias(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], bias_vec: np.ndarray, name: str):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    logits = logits + bias_vec[None, :]
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, name)

def search_classifier_operating_points(clf: nn.Module, X_val: np.ndarray, cache_val: Dict[str, np.ndarray], targets: List[float]):
    rows = []
    grid = [-2.0, -1.0, -0.5, 0.0, 0.5, 1.0, 2.0]
    bias_candidates = []
    for b4 in grid:
        for b8 in grid:
            bias_candidates.append(np.array([0.0, b4, b8], dtype=np.float32))

    for target in targets:
        best = None
        best_gap = 1e9
        for bias in bias_candidates:
            row = evaluate_classifier_with_bias(clf, X_val, cache_val, bias, "POTTS-CLS")
            gap = abs(row["avg_compute"] - target)
            if gap < best_gap:
                best_gap = gap
                best = (bias, row)
        rows.append({
            "target_compute": target,
            "bias_1": float(best[0][0]),
            "bias_4": float(best[0][1]),
            "bias_8": float(best[0][2]),
            **best[1]
        })
    return pd.DataFrame(rows)

# =========================
# Prepare features / baselines
# =========================
X_train_features = {name: build_feature_matrix(train_cache, feat_list, source_budget=1)
                    for name, feat_list in FEATURE_ABLATIONS.items()}
X_val_features = {name: build_feature_matrix(val_cache, feat_list, source_budget=1)
                  for name, feat_list in FEATURE_ABLATIONS.items()}
X_test_features = {name: build_feature_matrix(test_cache, feat_list, source_budget=1)
                   for name, feat_list in FEATURE_ABLATIONS.items()}

main_rows = []
main_rows.append(evaluate_fixed(test_cache, 1))
main_rows.append(evaluate_fixed(test_cache, 4))
main_rows.append(evaluate_fixed(test_cache, 8))
main_rows.append(evaluate_random_policy(test_cache, target_compute=1.10, seed=SEED))

val_entropy = predictive_entropy(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_entropy = predictive_entropy(torch.from_numpy(test_cache["p1_probs"])).numpy()

val_margin = prediction_margin(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_margin = prediction_margin(torch.from_numpy(test_cache["p1_probs"])).numpy()

ent_best_main = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=1.10)
t1_ent, t2_ent, _, _, _ = ent_best_main
ent_sel_test = np.where(test_entropy < t1_ent, 1, np.where(test_entropy < t2_ent, 4, 8))
main_rows.append(apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold"))

mar_best_main = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=1.10)
t1_mar, t2_mar, _, _, _ = mar_best_main
mar_sel_test = np.where(test_margin > t2_mar, 1, np.where(test_margin > t1_mar, 4, 8))
main_rows.append(apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold"))

# =========================
# Train sweep
# =========================
ablation_rows = []
trained_classifiers = {}

for util_mode in UTILITY_MODES:
    y_train, _ = build_budget_targets(train_cache, util_mode)
    y_val, _ = build_budget_targets(val_cache, util_mode)

    print(f"\n========== UTILITY MODE: {util_mode} ==========")
    print("Train label counts:", np.bincount(y_train, minlength=3))
    print("Val   label counts:", np.bincount(y_val, minlength=3))

    for feat_name in FEATURE_ABLATIONS.keys():
        print(f"[Info] Training classifier: utility={util_mode}, feat={feat_name}")
        clf = train_budget_classifier(
            X_train_features[feat_name], y_train,
            X_val_features[feat_name], y_val
        )
        trained_classifiers[(util_mode, feat_name)] = clf

        row = evaluate_classifier(clf, X_test_features[feat_name], test_cache, name=f"POTTS-CLS[{util_mode}|{feat_name}]")
        row["utility_mode"] = util_mode
        row["feature_set"] = feat_name
        ablation_rows.append(row)

ablation_df = pd.DataFrame(ablation_rows)

print("\n=== ABLATION RESULTS ===")
display(ablation_df.sort_values(["feature_set", "acc", "avg_compute"], ascending=[True, False, True]).reset_index(drop=True))

# =========================
# Trade-off curves for heuristics
# =========================
curve_rows = []

for target in CURVE_TARGETS:
    ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=target)
    t1, t2, _, _, _ = ent_best
    ent_sel_test = np.where(test_entropy < t1, 1, np.where(test_entropy < t2, 4, 8))
    row = apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

    mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=target)
    t1m, t2m, _, _, _ = mar_best
    mar_sel_test = np.where(test_margin > t2m, 1, np.where(test_margin > t1m, 4, 8))
    row = apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

# =========================
# Trade-off curves for all learned models
# =========================
matched_rows = []

for (util_mode, feat_name), clf in trained_classifiers.items():
    curve_df_tmp = search_classifier_operating_points(clf, X_val_features[feat_name], val_cache, CURVE_TARGETS)

    for _, rr in curve_df_tmp.iterrows():
        bias = np.array([rr["bias_1"], rr["bias_4"], rr["bias_8"]], dtype=np.float32)
        row = evaluate_classifier_with_bias(clf, X_test_features[feat_name], test_cache, bias, "POTTS-CLS")
        row["target_compute"] = rr["target_compute"]
        row["bias_1"] = rr["bias_1"]
        row["bias_4"] = rr["bias_4"]
        row["bias_8"] = rr["bias_8"]
        row["utility_mode"] = util_mode
        row["feature_set"] = feat_name
        curve_rows.append(row)

curve_df = pd.DataFrame(curve_rows)
curve_df.to_csv(os.path.join(RESULT_DIR, "tradeoff_curve.csv"), index=False)

print("\n=== TRADE-OFF CURVE ===")
display(curve_df.head(50))

# =========================
# Matched-compute summary
# =========================
def best_heuristic_at_target(target: float) -> Dict[str, float]:
    rows = []
    for method in ["Entropy-threshold", "Margin-threshold"]:
        sub = curve_df[(curve_df["method"] == method)]
        if len(sub) == 0:
            continue
        tmp = sub.copy()
        tmp["gap"] = (tmp["avg_compute"] - target).abs()
        rows.append(tmp.sort_values(["gap", "acc"], ascending=[True, False]).iloc[0])
    if not rows:
        return {}
    rows_df = pd.DataFrame(rows)
    best = rows_df.sort_values(["acc", "gap"], ascending=[False, True]).iloc[0]
    return best.to_dict()

def best_learned_at_target(target: float) -> Dict[str, float]:
    sub = curve_df[curve_df["method"] == "POTTS-CLS"].copy()
    if len(sub) == 0:
        return {}
    sub["gap"] = (sub["avg_compute"] - target).abs()
    best = sub.sort_values(["acc", "gap"], ascending=[False, True]).iloc[0]
    return best.to_dict()

matched_summary = []
for target in MATCH_TARGETS:
    h = best_heuristic_at_target(target)
    l = best_learned_at_target(target)
    if h and l:
        matched_summary.append({
            "target_compute": target,
            "heuristic_method": h["method"],
            "heuristic_compute": h["avg_compute"],
            "heuristic_acc": h["acc"],
            "heuristic_ece": h["ece"],
            "learned_compute": l["avg_compute"],
            "learned_acc": l["acc"],
            "learned_ece": l["ece"],
            "learned_utility_mode": l.get("utility_mode", ""),
            "learned_feature_set": l.get("feature_set", ""),
            "acc_delta_learned_minus_heuristic": l["acc"] - h["acc"],
        })

matched_df = pd.DataFrame(matched_summary)
matched_df.to_csv(os.path.join(RESULT_DIR, "matched_compute_summary.csv"), index=False)

print("\n=== MATCHED COMPUTE SUMMARY ===")
display(matched_df)

# =========================
# Choose main learned result automatically
# =========================
# Strategy:
# 1. Prefer learned rows that beat matched heuristic accuracy.
# 2. If none beat heuristic, choose smallest negative gap.
main_learned_row = None
if len(matched_df) > 0:
    winners = matched_df[matched_df["acc_delta_learned_minus_heuristic"] > 0]
    if len(winners) > 0:
        chosen = winners.sort_values(
            ["acc_delta_learned_minus_heuristic", "learned_compute"],
            ascending=[False, True]
        ).iloc[0]
    else:
        chosen = matched_df.sort_values(
            ["acc_delta_learned_minus_heuristic", "learned_compute"],
            ascending=[False, True]
        ).iloc[0]

    util_mode = chosen["learned_utility_mode"]
    feat_name = chosen["learned_feature_set"]
    target = chosen["target_compute"]

    sub = curve_df[
        (curve_df["method"] == "POTTS-CLS") &
        (curve_df["utility_mode"] == util_mode) &
        (curve_df["feature_set"] == feat_name)
    ].copy()
    sub["gap"] = (sub["avg_compute"] - target).abs()
    main_learned_row = sub.sort_values(["acc", "gap"], ascending=[False, True]).iloc[0].to_dict()

if main_learned_row is not None:
    row = {
        "method": "POTTS-CLS",
        "avg_compute": main_learned_row["avg_compute"],
        "acc": main_learned_row["acc"],
        "ece": main_learned_row["ece"],
        "main_key": f"({main_learned_row['utility_mode']}, {main_learned_row['feature_set']})"
    }
    main_rows.append(row)

main_df = pd.DataFrame(main_rows).sort_values(["avg_compute", "acc"], ascending=[True, False]).reset_index(drop=True)
main_df.to_csv(os.path.join(RESULT_DIR, "main_results.csv"), index=False)

print("\n=== MAIN RESULTS ===")
display(main_df)

# =========================
# Paper-friendly main table
# =========================
paper_rows = []
for _, r in main_df.iterrows():
    paper_rows.append({
        "Method": r["method"],
        "Compute": round(float(r["avg_compute"]), 3),
        "Acc (%)": round(float(r["acc"]), 3),
        "ECE": round(float(r["ece"]), 4),
    })

paper_df = pd.DataFrame(paper_rows)
paper_df.to_csv(os.path.join(RESULT_DIR, "paper_table_main.csv"), index=False)

print("\nSaved files:")
for fn in sorted(os.listdir(RESULT_DIR)):
    print(" -", os.path.join(RESULT_DIR, fn))
```

```python
# =========================================================
# ONE-CELL COLAB: "WIN HEURISTIC" VERSION
# =========================================================
# Goal:
# - maximize the chance of finding a learned point that beats heuristic
# - fixed TTA
# - label-based budget classifier on {1,4,8}
# - stronger middle/high-budget weighting
# - broader bias search
# - tighter matched-compute evaluation
#
# What is changed vs previous versions:
# 1) stronger class weighting toward K=4 / K=8
# 2) focus on utility lambdas around the useful range: 0.05 / 0.06 / 0.07
# 3) larger bias grid for operating-point search
# 4) matched-compute summary is strict and fair
# 5) main result chosen from true matched-compute comparisons only
#
# Outputs:
# /content/potts_results/main_results.csv
# /content/potts_results/ablation_results.csv
# /content/potts_results/tradeoff_curve.csv
# /content/potts_results/matched_compute_summary.csv
# /content/potts_results/paper_table_main.csv
# =========================================================

!pip -q install timm pandas scikit-learn tqdm

import os
import random
import warnings
from typing import Dict, List, Optional

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset

import torchvision
import torchvision.transforms as T
from torchvision.datasets import CIFAR100

import timm

warnings.filterwarnings("ignore")

# =========================
# Config
# =========================
SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
NUM_WORKERS = 0
BATCH_SIZE = 256

TRAIN_POLICY_SUBSET = 8000
VAL_SUBSET = 1000
TEST_SUBSET = 2000

IMAGE_SIZE = 224
NUM_CLASSES = 100

BACKBONE_NAME = "resnet50"
USE_PRETRAINED_HEAD = True
HEAD_EPOCHS = 1
HEAD_LR = 1e-3

POLICY_HIDDEN = 128           # slightly stronger classifier
POLICY_EPOCHS = 8             # a bit more training
POLICY_LR = 1e-3
POLICY_WEIGHT_DECAY = 1e-4

RESULT_DIR = "/content/potts_results"
os.makedirs(RESULT_DIR, exist_ok=True)

BUDGETS = [1, 4, 8]
COSTS = {1: 1.0, 4: 4.0, 8: 8.0}
BUDGET_TO_KEY = {1: "p1_probs", 4: "p4_probs", 8: "p8_probs"}
CLASS_TO_BUDGET = {0: 1, 1: 4, 2: 8}

FEATURE_ABLATIONS = {
    "entropy_only": ["entropy"],
    "full": ["entropy", "margin", "disagreement", "embedding"],
}

# Focus on the range that previously looked promising
UTILITY_MODES = ["util_0.05", "util_0.06", "util_0.07"]

CURVE_TARGETS = [1.05, 1.20, 1.50, 1.70, 2.00, 2.30, 2.50, 2.80, 3.00, 3.50, 4.00]
MATCH_TARGETS = [1.20, 1.50, 1.70, 2.00, 2.30, 2.50, 2.80, 3.00, 3.50, 4.00]
MATCH_TOL = 0.25

# stronger emphasis on middle / high budget classes
EXTRA_CLASS_MULTIPLIER = np.array([1.0, 3.0, 4.0], dtype=np.float32)

# larger bias grid to discover more operating points
BIAS_GRID = [-3.0, -2.0, -1.5, -1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0, 3.0]

# =========================
# Seed
# =========================
def seed_everything(seed: int = 42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)
if torch.cuda.is_available():
    torch.backends.cudnn.benchmark = True

print("DEVICE =", DEVICE)

# =========================
# Metrics / utils
# =========================
def compute_ece(probs: np.ndarray, labels: np.ndarray, n_bins: int = 15) -> float:
    confidences = probs.max(axis=1)
    predictions = probs.argmax(axis=1)
    accuracies = (predictions == labels).astype(np.float32)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        mask = (confidences > bins[i]) & (confidences <= bins[i + 1])
        if mask.any():
            bin_acc = accuracies[mask].mean()
            bin_conf = confidences[mask].mean()
            ece += np.abs(bin_acc - bin_conf) * mask.mean()
    return float(ece)

def top1_accuracy(probs: np.ndarray, labels: np.ndarray) -> float:
    return float((probs.argmax(axis=1) == labels).mean() * 100.0)

def predictive_entropy(probs: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return -(probs * (probs + eps).log()).sum(dim=1)

def prediction_margin(probs: torch.Tensor) -> torch.Tensor:
    top2 = torch.topk(probs, k=2, dim=1).values
    return top2[:, 0] - top2[:, 1]

def kl_divergence_batch(p: torch.Tensor, q: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return (p * ((p + eps).log() - (q + eps).log())).sum(dim=1)

def make_subset_indices(n: int, train_n: int, val_n: int, test_n: int, seed: int = 42):
    idx = np.arange(n)
    rng = np.random.default_rng(seed)
    rng.shuffle(idx)
    train_idx = idx[:train_n]
    val_idx = idx[train_n:train_n + val_n]
    test_idx = idx[train_n + val_n:train_n + val_n + test_n]
    return train_idx, val_idx, test_idx

def parse_util_lambda(mode: str) -> Optional[float]:
    if mode.startswith("util_"):
        return float(mode.split("_")[1])
    return None

def normalized_cost(budget: int) -> float:
    return (budget - 1.0) / 7.0

# =========================
# Data
# =========================
mean = (0.5071, 0.4867, 0.4408)
std = (0.2675, 0.2565, 0.2761)

base_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_transform = T.Compose([
    T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    T.RandomHorizontalFlip(),
    T.RandomCrop(IMAGE_SIZE, padding=16),
    T.ToTensor(),
    T.Normalize(mean, std),
])

train_ds_full = CIFAR100(root="/content/data", train=True, download=True, transform=train_transform)
eval_train_ds_full = CIFAR100(root="/content/data", train=True, download=False, transform=base_transform)
test_ds_full = CIFAR100(root="/content/data", train=False, download=True, transform=base_transform)

train_idx, val_idx, _ = make_subset_indices(
    n=len(train_ds_full),
    train_n=TRAIN_POLICY_SUBSET,
    val_n=VAL_SUBSET,
    test_n=TEST_SUBSET,
    seed=SEED
)

head_train_ds = Subset(train_ds_full, train_idx)
policy_train_ds = Subset(eval_train_ds_full, train_idx)
val_ds = Subset(eval_train_ds_full, val_idx)
test_ds = Subset(test_ds_full, np.arange(TEST_SUBSET))

head_train_loader = DataLoader(head_train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
policy_train_loader = DataLoader(policy_train_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

# =========================
# Backbone + head
# =========================
class BackboneWrapper(nn.Module):
    def __init__(self, backbone_name: str, num_classes: int = 100):
        super().__init__()
        if backbone_name == "resnet50":
            m = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
            feat_dim = m.fc.in_features
            m.fc = nn.Identity()
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        elif backbone_name == "deit_small_patch16_224":
            m = timm.create_model("deit_small_patch16_224", pretrained=True, num_classes=0)
            feat_dim = m.num_features
            self.backbone = m
            self.head = nn.Linear(feat_dim, num_classes)
            self.feat_dim = feat_dim
        else:
            raise ValueError(f"Unsupported backbone: {backbone_name}")

    def forward_features(self, x: torch.Tensor) -> torch.Tensor:
        return self.backbone(x)

    def forward(self, x: torch.Tensor):
        feat = self.forward_features(x)
        logits = self.head(feat)
        return logits, feat

model = BackboneWrapper(BACKBONE_NAME, num_classes=NUM_CLASSES).to(DEVICE)

for p in model.backbone.parameters():
    p.requires_grad = False

if USE_PRETRAINED_HEAD:
    print(f"[Info] Training classifier head for {HEAD_EPOCHS} epoch(s)...")
    optimizer = torch.optim.Adam(model.head.parameters(), lr=HEAD_LR)
    model.train()
    for epoch in range(HEAD_EPOCHS):
        pbar = tqdm(head_train_loader, desc=f"Head train {epoch+1}/{HEAD_EPOCHS}")
        for images, labels in pbar:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            optimizer.zero_grad()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
                logits, _ = model(images)
                loss = F.cross_entropy(logits, labels)
            loss.backward()
            optimizer.step()
            pbar.set_postfix(loss=float(loss.detach().cpu()))
else:
    raise RuntimeError("Set USE_PRETRAINED_HEAD=True or load your own checkpoint.")

model.eval()

# =========================
# Fixed TTA helper
# =========================
@torch.no_grad()
def batch_tta_predict(model: nn.Module, images: torch.Tensor, k: int = 1):
    model.eval()
    probs_list = []
    feats_list = []

    def apply_det_tta(x: torch.Tensor, aug_id: int):
        if aug_id == 0:
            return x
        elif aug_id == 1:
            return torch.flip(x, dims=[3])
        elif aug_id == 2:
            x2 = x[:, :, :, 4:]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 3:
            x2 = x[:, :, :, :-4]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 4:
            x2 = x[:, :, 4:, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 5:
            x2 = x[:, :, :-4, :]
            x2 = F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 6:
            H, W = x.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        elif aug_id == 7:
            x2 = torch.flip(x, dims=[3])
            H, W = x2.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x2[:, :, dh:H-dh, dw:W-dw]
            x2 = F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
            return x2
        else:
            return x

    with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
        for aug_id in range(k):
            x_aug = apply_det_tta(images, aug_id)
            logits, feats = model(x_aug)
            probs = F.softmax(logits, dim=1)
            probs_list.append(probs.float())
            feats_list.append(feats.float())

    probs_stack = torch.stack(probs_list, dim=0)
    feats_stack = torch.stack(feats_list, dim=0)
    avg_probs = probs_stack.mean(dim=0)
    avg_feats = feats_stack.mean(dim=0)

    if k == 1:
        disagreement = torch.zeros(images.size(0), device=images.device)
    else:
        kls = []
        for i in range(k):
            kls.append(kl_divergence_batch(probs_stack[i], avg_probs))
        disagreement = torch.stack(kls, dim=0).mean(dim=0)

    return avg_probs, avg_feats, disagreement

# =========================
# Cache candidate outputs
# =========================
@torch.no_grad()
def collect_policy_candidates(loader: DataLoader, name: str) -> Dict[str, np.ndarray]:
    model.eval()
    out = {
        "labels": [],
        "p1_probs": [], "p4_probs": [], "p8_probs": [],
        "p1_feats": [], "p4_feats": [], "p8_feats": [],
        "p1_dis": [], "p4_dis": [], "p8_dis": [],
    }

    for images, labels in tqdm(loader, desc=f"Collect [{name}]"):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        probs1, feat1, dis1 = batch_tta_predict(model, images, k=1)
        probs4, feat4, dis4 = batch_tta_predict(model, images, k=4)
        probs8, feat8, dis8 = batch_tta_predict(model, images, k=8)

        out["labels"].append(labels.cpu().numpy())
        out["p1_probs"].append(probs1.cpu().numpy())
        out["p4_probs"].append(probs4.cpu().numpy())
        out["p8_probs"].append(probs8.cpu().numpy())
        out["p1_feats"].append(feat1.cpu().numpy())
        out["p4_feats"].append(feat4.cpu().numpy())
        out["p8_feats"].append(feat8.cpu().numpy())
        out["p1_dis"].append(dis1.cpu().numpy())
        out["p4_dis"].append(dis4.cpu().numpy())
        out["p8_dis"].append(dis8.cpu().numpy())

    for k in list(out.keys()):
        out[k] = np.concatenate(out[k], axis=0)
    return out

print("[Info] Collecting cached candidate outputs...")
train_cache = collect_policy_candidates(policy_train_loader, "train")
val_cache = collect_policy_candidates(val_loader, "val")
test_cache = collect_policy_candidates(test_loader, "test")

print("\n=== QUICK TTA SANITY CHECK ===")
print(f"Single: {top1_accuracy(test_cache['p1_probs'], test_cache['labels']):.2f}")
print(f"TTA-4 : {top1_accuracy(test_cache['p4_probs'], test_cache['labels']):.2f}")
print(f"TTA-8 : {top1_accuracy(test_cache['p8_probs'], test_cache['labels']):.2f}")

# =========================
# Feature builder
# =========================
def build_feature_matrix(cache: Dict[str, np.ndarray], feature_set: List[str], source_budget: int = 1) -> np.ndarray:
    probs = cache[f"p{source_budget}_probs"]
    feats = cache[f"p{source_budget}_feats"]
    dis = cache["p4_dis"]

    probs_t = torch.from_numpy(probs)
    entropy = predictive_entropy(probs_t).numpy()[:, None]
    margin = prediction_margin(probs_t).numpy()[:, None]
    pmax = probs.max(axis=1, keepdims=True)
    disagreement = dis[:, None]

    xs = []
    for name in feature_set:
        if name == "entropy":
            xs.append(entropy)
        elif name == "margin":
            xs.append(margin)
            xs.append(pmax)
        elif name == "disagreement":
            xs.append(disagreement)
        elif name == "embedding":
            xs.append(feats)
        else:
            raise ValueError(f"Unknown feature: {name}")
    return np.concatenate(xs, axis=1).astype(np.float32)

# =========================
# Utility labels
# =========================
def sample_correct(cache: Dict[str, np.ndarray], budget: int) -> np.ndarray:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    preds = probs.argmax(axis=1)
    return (preds == labels).astype(np.float32)

def build_budget_targets(cache: Dict[str, np.ndarray], mode: str):
    n = len(cache["labels"])
    corr1 = sample_correct(cache, 1)
    corr4 = sample_correct(cache, 4)
    corr8 = sample_correct(cache, 8)

    util = np.zeros((n, 3), dtype=np.float32)

    lam = parse_util_lambda(mode)
    if lam is None:
        raise ValueError(f"Unknown utility mode: {mode}")

    util[:, 0] = corr1 - lam * normalized_cost(1)
    util[:, 1] = corr4 - lam * normalized_cost(4)
    util[:, 2] = corr8 - lam * normalized_cost(8)

    util[:, 0] += 1e-6
    util[:, 1] += 5e-7
    util[:, 2] += 0.0

    y = util.argmax(axis=1).astype(np.int64)
    return y, util

# =========================
# Budget classifier
# =========================
class BudgetClassifier(nn.Module):
    def __init__(self, input_dim: int, hidden_dim: int = 128, num_classes: int = 3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_classes),
        )

    def forward(self, x: torch.Tensor):
        return self.net(x)

def compute_class_weights(y: np.ndarray) -> torch.Tensor:
    counts = np.bincount(y, minlength=3).astype(np.float32)
    counts[counts == 0] = 1.0
    weights = counts.sum() / counts
    weights = weights / weights.mean()
    weights = weights * EXTRA_CLASS_MULTIPLIER
    weights = weights / weights.mean()
    return torch.tensor(weights, dtype=torch.float32)

def train_budget_classifier(X_train, y_train, X_val=None, y_val=None):
    clf = BudgetClassifier(input_dim=X_train.shape[1], hidden_dim=POLICY_HIDDEN).to(DEVICE)
    class_weights = compute_class_weights(y_train).to(DEVICE)
    opt = torch.optim.Adam(clf.parameters(), lr=POLICY_LR, weight_decay=POLICY_WEIGHT_DECAY)

    Xtr = torch.from_numpy(X_train)
    Ytr = torch.from_numpy(y_train)

    best_state = None
    best_val = 1e9

    for epoch in range(POLICY_EPOCHS):
        clf.train()
        perm = torch.randperm(len(Xtr))
        losses = []

        for start in range(0, len(Xtr), 1024):
            idx = perm[start:start + 1024]
            xb = Xtr[idx].to(DEVICE)
            yb = Ytr[idx].to(DEVICE)

            opt.zero_grad()
            logits = clf(xb)
            loss = F.cross_entropy(logits, yb, weight=class_weights)
            loss.backward()
            opt.step()
            losses.append(float(loss.detach().cpu()))

        val_loss = np.mean(losses)
        if X_val is not None and y_val is not None and len(X_val) > 0:
            clf.eval()
            with torch.no_grad():
                xv = torch.from_numpy(X_val).to(DEVICE)
                yv = torch.from_numpy(y_val).to(DEVICE)
                logits = clf(xv)
                val_loss = float(F.cross_entropy(logits, yv).detach().cpu())

        print(f"[Classifier] epoch {epoch+1}/{POLICY_EPOCHS} train={np.mean(losses):.4f} val={val_loss:.4f}")

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in clf.state_dict().items()}

    if best_state is not None:
        clf.load_state_dict(best_state)

    clf.eval()
    return clf

# =========================
# Evaluation helpers
# =========================
def evaluate_fixed(cache: Dict[str, np.ndarray], budget: int) -> Dict[str, float]:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    return {
        "method": f"TTA-{budget}" if budget > 1 else "Single",
        "avg_compute": float(COSTS[budget]),
        "acc": top1_accuracy(probs, labels),
        "ece": compute_ece(probs, labels),
    }

def apply_budget_selection(cache: Dict[str, np.ndarray], selected_budgets: np.ndarray, name: str) -> Dict[str, float]:
    labels = cache["labels"]
    final_probs = []
    costs = []

    for i, b in enumerate(selected_budgets.tolist()):
        final_probs.append(cache[BUDGET_TO_KEY[int(b)]][i])
        costs.append(COSTS[int(b)])

    final_probs = np.asarray(final_probs)
    return {
        "method": name,
        "avg_compute": float(np.mean(costs)),
        "acc": top1_accuracy(final_probs, labels),
        "ece": compute_ece(final_probs, labels),
    }

def tune_thresholds(values: np.ndarray, cache: Dict[str, np.ndarray], mode: str = "entropy", target_compute: Optional[float] = None):
    labels = cache["labels"]
    candidates = np.quantile(values, np.linspace(0.1, 0.9, 9))
    best = None
    best_score = -1e9

    for t1 in candidates:
        for t2 in candidates:
            if t2 < t1:
                continue

            if mode == "entropy":
                sel = np.where(values < t1, 1, np.where(values < t2, 4, 8))
            else:
                sel = np.where(values > t2, 1, np.where(values > t1, 4, 8))

            probs = np.stack([cache[BUDGET_TO_KEY[int(b)]][i] for i, b in enumerate(sel)], axis=0)
            acc = top1_accuracy(probs, labels)
            avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))

            score = acc
            if target_compute is not None:
                score -= 10.0 * abs(avg_compute - target_compute)

            if score > best_score:
                best_score = score
                best = (t1, t2, sel, acc, avg_compute)

    return best

def evaluate_random_policy(cache: Dict[str, np.ndarray], target_compute: float = 1.10, seed: int = 42):
    rng = np.random.default_rng(seed)
    best = None
    best_gap = 1e9

    probs_set = [
        np.array([0.98, 0.02, 0.00], dtype=np.float32),
        np.array([0.96, 0.04, 0.00], dtype=np.float32),
        np.array([0.94, 0.05, 0.01], dtype=np.float32),
        np.array([0.92, 0.06, 0.02], dtype=np.float32),
        np.array([0.90, 0.08, 0.02], dtype=np.float32),
        np.array([0.88, 0.10, 0.02], dtype=np.float32),
    ]

    for p in probs_set:
        sel_idx = rng.choice(3, size=len(cache["labels"]), p=p)
        sel = np.array([BUDGETS[i] for i in sel_idx])
        avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))
        gap = abs(avg_compute - target_compute)
        if gap < best_gap:
            best_gap = gap
            best = sel

    return apply_budget_selection(cache, best, f"Random(target={target_compute:.2f})")

@torch.no_grad()
def evaluate_classifier(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], name: str = "POTTS-CLS"):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, name)

@torch.no_grad()
def evaluate_classifier_with_bias(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], bias_vec: np.ndarray, name: str):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    logits = logits + bias_vec[None, :]
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, name)

def search_classifier_operating_points(clf: nn.Module, X_val: np.ndarray, cache_val: Dict[str, np.ndarray], targets: List[float]):
    rows = []
    bias_candidates = []
    for b4 in BIAS_GRID:
        for b8 in BIAS_GRID:
            bias_candidates.append(np.array([0.0, b4, b8], dtype=np.float32))

    for target in targets:
        best = None
        best_gap = 1e9
        for bias in bias_candidates:
            row = evaluate_classifier_with_bias(clf, X_val, cache_val, bias, "POTTS-CLS")
            gap = abs(row["avg_compute"] - target)
            if gap < best_gap:
                best_gap = gap
                best = (bias, row)
        rows.append({
            "target_compute": target,
            "bias_1": float(best[0][0]),
            "bias_4": float(best[0][1]),
            "bias_8": float(best[0][2]),
            **best[1]
        })
    return pd.DataFrame(rows)

# =========================
# Prepare features / baselines
# =========================
X_train_features = {name: build_feature_matrix(train_cache, feat_list, source_budget=1)
                    for name, feat_list in FEATURE_ABLATIONS.items()}
X_val_features = {name: build_feature_matrix(val_cache, feat_list, source_budget=1)
                  for name, feat_list in FEATURE_ABLATIONS.items()}
X_test_features = {name: build_feature_matrix(test_cache, feat_list, source_budget=1)
                   for name, feat_list in FEATURE_ABLATIONS.items()}

main_rows = []
main_rows.append(evaluate_fixed(test_cache, 1))
main_rows.append(evaluate_fixed(test_cache, 4))
main_rows.append(evaluate_fixed(test_cache, 8))
main_rows.append(evaluate_random_policy(test_cache, target_compute=1.10, seed=SEED))

val_entropy = predictive_entropy(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_entropy = predictive_entropy(torch.from_numpy(test_cache["p1_probs"])).numpy()

val_margin = prediction_margin(torch.from_numpy(val_cache["p1_probs"])).numpy()
test_margin = prediction_margin(torch.from_numpy(test_cache["p1_probs"])).numpy()

ent_best_main = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=1.10)
t1_ent, t2_ent, _, _, _ = ent_best_main
ent_sel_test = np.where(test_entropy < t1_ent, 1, np.where(test_entropy < t2_ent, 4, 8))
main_rows.append(apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold"))

mar_best_main = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=1.10)
t1_mar, t2_mar, _, _, _ = mar_best_main
mar_sel_test = np.where(test_margin > t2_mar, 1, np.where(test_margin > t1_mar, 4, 8))
main_rows.append(apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold"))

# =========================
# Train focused sweep
# =========================
ablation_rows = []
trained_classifiers = {}

for util_mode in UTILITY_MODES:
    y_train, _ = build_budget_targets(train_cache, util_mode)
    y_val, _ = build_budget_targets(val_cache, util_mode)

    print(f"\n========== UTILITY MODE: {util_mode} ==========")
    print("Train label counts:", np.bincount(y_train, minlength=3))
    print("Val   label counts:", np.bincount(y_val, minlength=3))

    for feat_name in FEATURE_ABLATIONS.keys():
        print(f"[Info] Training classifier: utility={util_mode}, feat={feat_name}")
        clf = train_budget_classifier(
            X_train_features[feat_name], y_train,
            X_val_features[feat_name], y_val
        )
        trained_classifiers[(util_mode, feat_name)] = clf

        row = evaluate_classifier(clf, X_test_features[feat_name], test_cache, name=f"POTTS-CLS[{util_mode}|{feat_name}]")
        row["utility_mode"] = util_mode
        row["feature_set"] = feat_name
        ablation_rows.append(row)

ablation_df = pd.DataFrame(ablation_rows)
ablation_df.to_csv(os.path.join(RESULT_DIR, "ablation_results.csv"), index=False)

print("\n=== ABLATION RESULTS ===")
display(ablation_df.sort_values(["feature_set", "acc", "avg_compute"], ascending=[True, False, True]).reset_index(drop=True))

# =========================
# Trade-off curves
# =========================
curve_rows = []

# heuristics
for target in CURVE_TARGETS:
    ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=target)
    t1, t2, _, _, _ = ent_best
    ent_sel_test = np.where(test_entropy < t1, 1, np.where(test_entropy < t2, 4, 8))
    row = apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

    mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=target)
    t1m, t2m, _, _, _ = mar_best
    mar_sel_test = np.where(test_margin > t2m, 1, np.where(test_margin > t1m, 4, 8))
    row = apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold")
    row["target_compute"] = target
    curve_rows.append(row)

# learned
for (util_mode, feat_name), clf in trained_classifiers.items():
    curve_df_tmp = search_classifier_operating_points(clf, X_val_features[feat_name], val_cache, CURVE_TARGETS)

    for _, rr in curve_df_tmp.iterrows():
        bias = np.array([rr["bias_1"], rr["bias_4"], rr["bias_8"]], dtype=np.float32)
        row = evaluate_classifier_with_bias(clf, X_test_features[feat_name], test_cache, bias, "POTTS-CLS")
        row["target_compute"] = rr["target_compute"]
        row["bias_1"] = rr["bias_1"]
        row["bias_4"] = rr["bias_4"]
        row["bias_8"] = rr["bias_8"]
        row["utility_mode"] = util_mode
        row["feature_set"] = feat_name
        curve_rows.append(row)

curve_df = pd.DataFrame(curve_rows)
curve_df.to_csv(os.path.join(RESULT_DIR, "tradeoff_curve.csv"), index=False)

print("\n=== TRADE-OFF CURVE ===")
display(curve_df.head(80))

# =========================
# Strict matched-compute summary
# =========================
def select_best_near_target(df: pd.DataFrame, target: float, tol: float):
    if len(df) == 0:
        return None
    tmp = df.copy()
    tmp["gap"] = (tmp["avg_compute"] - target).abs()
    tmp = tmp[tmp["gap"] <= tol].copy()
    if len(tmp) == 0:
        return None
    tmp = tmp.sort_values(["gap", "acc"], ascending=[True, False])
    best_gap = tmp.iloc[0]["gap"]
    near = tmp[tmp["gap"] <= best_gap + 1e-12].copy()
    near = near.sort_values(["acc", "ece"], ascending=[False, True])
    return near.iloc[0].to_dict()

matched_rows = []
heuristic_df = curve_df[curve_df["method"].isin(["Entropy-threshold", "Margin-threshold"])].copy()
learned_df = curve_df[curve_df["method"] == "POTTS-CLS"].copy()

for target in MATCH_TARGETS:
    h_best = select_best_near_target(heuristic_df, target, MATCH_TOL)
    l_best = select_best_near_target(learned_df, target, MATCH_TOL)

    matched_rows.append({
        "target_compute": target,
        "heuristic_found": h_best is not None,
        "learned_found": l_best is not None,
        "heuristic_method": h_best["method"] if h_best else None,
        "heuristic_compute": h_best["avg_compute"] if h_best else None,
        "heuristic_acc": h_best["acc"] if h_best else None,
        "heuristic_ece": h_best["ece"] if h_best else None,
        "heuristic_gap": h_best["gap"] if h_best else None,
        "learned_compute": l_best["avg_compute"] if l_best else None,
        "learned_acc": l_best["acc"] if l_best else None,
        "learned_ece": l_best["ece"] if l_best else None,
        "learned_gap": l_best["gap"] if l_best else None,
        "learned_utility_mode": l_best.get("utility_mode", None) if l_best else None,
        "learned_feature_set": l_best.get("feature_set", None) if l_best else None,
        "acc_delta_learned_minus_heuristic": (
            (l_best["acc"] - h_best["acc"]) if (h_best is not None and l_best is not None) else None
        ),
    })

matched_df = pd.DataFrame(matched_rows)
matched_df.to_csv(os.path.join(RESULT_DIR, "matched_compute_summary.csv"), index=False)

print("\n=== MATCHED COMPUTE SUMMARY ===")
display(matched_df)

# =========================
# Choose main learned result from fair matched-compute comparison
# =========================
main_learned_row = None
valid_comp = matched_df[
    (matched_df["heuristic_found"] == True) &
    (matched_df["learned_found"] == True)
].copy()

if len(valid_comp) > 0:
    winners = valid_comp[valid_comp["acc_delta_learned_minus_heuristic"] > 0].copy()
    if len(winners) > 0:
        chosen = winners.sort_values(
            ["acc_delta_learned_minus_heuristic", "learned_gap", "learned_compute"],
            ascending=[False, True, True]
        ).iloc[0]
    else:
        chosen = valid_comp.sort_values(
            ["acc_delta_learned_minus_heuristic", "learned_gap", "learned_compute"],
            ascending=[False, True, True]
        ).iloc[0]

    sub = learned_df[
        (learned_df["utility_mode"] == chosen["learned_utility_mode"]) &
        (learned_df["feature_set"] == chosen["learned_feature_set"])
    ].copy()
    sub["gap"] = (sub["avg_compute"] - chosen["target_compute"]).abs()
    sub = sub[sub["gap"] <= MATCH_TOL].copy()
    if len(sub) > 0:
        sub = sub.sort_values(["gap", "acc"], ascending=[True, False])
        best_gap = sub.iloc[0]["gap"]
        sub2 = sub[sub["gap"] <= best_gap + 1e-12].copy().sort_values(["acc", "ece"], ascending=[False, True])
        main_learned_row = sub2.iloc[0].to_dict()

if main_learned_row is not None:
    row = {
        "method": "POTTS-CLS",
        "avg_compute": main_learned_row["avg_compute"],
        "acc": main_learned_row["acc"],
        "ece": main_learned_row["ece"],
        "main_key": f"({main_learned_row['utility_mode']}, {main_learned_row['feature_set']})"
    }
    main_rows.append(row)

main_df = pd.DataFrame(main_rows).sort_values(["avg_compute", "acc"], ascending=[True, False]).reset_index(drop=True)
main_df.to_csv(os.path.join(RESULT_DIR, "main_results.csv"), index=False)

print("\n=== MAIN RESULTS ===")
display(main_df)

# =========================
# Paper-friendly main table
# =========================
paper_rows = []
for _, r in main_df.iterrows():
    paper_rows.append({
        "Method": r["method"],
        "Compute": round(float(r["avg_compute"]), 3),
        "Acc (%)": round(float(r["acc"]), 3),
        "ECE": round(float(r["ece"]), 4),
    })

paper_df = pd.DataFrame(paper_rows)
paper_df.to_csv(os.path.join(RESULT_DIR, "paper_table_main.csv"), index=False)

print("\nSaved files:")
for fn in sorted(os.listdir(RESULT_DIR)):
    print(" -", os.path.join(RESULT_DIR, fn))
```

```python
# =========================================================
# ONE-CELL COLAB: FINALIZATION RUN
# Only run 3 target compute regions + 2 seeds
# Targets: 1.7 / 2.5 / 3.5
# Seeds: 42 / 43
#
# Purpose:
# - finalize the paper with minimal extra experiments
# - verify the learned policy at the 3 budget regions you care about
# - compare fairly against entropy-threshold and margin-threshold
#
# Outputs:
# /content/potts_final_runs/final_results_all.csv
# /content/potts_final_runs/final_summary_by_seed.csv
# /content/potts_final_runs/final_best_points.csv
# =========================================================

!pip -q install timm pandas scikit-learn tqdm

import os
import random
import warnings
from typing import Dict, List, Optional

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset

import torchvision
import torchvision.transforms as T
from torchvision.datasets import CIFAR100

warnings.filterwarnings("ignore")

# =========================
# Global config
# =========================
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
NUM_WORKERS = 0
BATCH_SIZE = 256

TRAIN_POLICY_SUBSET = 8000
VAL_SUBSET = 1000
TEST_SUBSET = 2000

IMAGE_SIZE = 224
NUM_CLASSES = 100

HEAD_EPOCHS = 1
HEAD_LR = 1e-3

POLICY_HIDDEN = 128
POLICY_EPOCHS = 8
POLICY_LR = 1e-3
POLICY_WEIGHT_DECAY = 1e-4

BUDGETS = [1, 4, 8]
COSTS = {1: 1.0, 4: 4.0, 8: 8.0}
BUDGET_TO_KEY = {1: "p1_probs", 4: "p4_probs", 8: "p8_probs"}
CLASS_TO_BUDGET = {0: 1, 1: 4, 2: 8}

TARGETS = [1.7, 2.5, 3.5]
MATCH_TOL = 0.25
SEEDS = [42, 43]

# Use the setting that looked most promising
UTILITY_MODE = "util_0.07"
FEATURE_SET_NAME = "full"
FEATURE_SET = ["entropy", "margin", "disagreement", "embedding"]

EXTRA_CLASS_MULTIPLIER = np.array([1.0, 3.0, 4.0], dtype=np.float32)
BIAS_GRID = [-3.0, -2.0, -1.5, -1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0, 3.0]

OUTDIR = "/content/potts_final_runs"
os.makedirs(OUTDIR, exist_ok=True)

print("DEVICE =", DEVICE)

# =========================
# Utilities
# =========================
def seed_everything(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

def compute_ece(probs: np.ndarray, labels: np.ndarray, n_bins: int = 15) -> float:
    confidences = probs.max(axis=1)
    predictions = probs.argmax(axis=1)
    accuracies = (predictions == labels).astype(np.float32)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        mask = (confidences > bins[i]) & (confidences <= bins[i + 1])
        if mask.any():
            bin_acc = accuracies[mask].mean()
            bin_conf = confidences[mask].mean()
            ece += np.abs(bin_acc - bin_conf) * mask.mean()
    return float(ece)

def top1_accuracy(probs: np.ndarray, labels: np.ndarray) -> float:
    return float((probs.argmax(axis=1) == labels).mean() * 100.0)

def predictive_entropy(probs: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return -(probs * (probs + eps).log()).sum(dim=1)

def prediction_margin(probs: torch.Tensor) -> torch.Tensor:
    top2 = torch.topk(probs, k=2, dim=1).values
    return top2[:, 0] - top2[:, 1]

def kl_divergence_batch(p: torch.Tensor, q: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return (p * ((p + eps).log() - (q + eps).log())).sum(dim=1)

def make_subset_indices(n: int, train_n: int, val_n: int, test_n: int, seed: int):
    idx = np.arange(n)
    rng = np.random.default_rng(seed)
    rng.shuffle(idx)
    train_idx = idx[:train_n]
    val_idx = idx[train_n:train_n + val_n]
    test_idx = idx[train_n + val_n:train_n + val_n + test_n]
    return train_idx, val_idx, test_idx

def parse_util_lambda(mode: str) -> Optional[float]:
    if mode.startswith("util_"):
        return float(mode.split("_")[1])
    return None

def normalized_cost(budget: int) -> float:
    return (budget - 1.0) / 7.0

def select_best_near_target(df: pd.DataFrame, target: float, tol: float):
    if len(df) == 0:
        return None
    tmp = df.copy()
    tmp["gap"] = (tmp["avg_compute"] - target).abs()
    tmp = tmp[tmp["gap"] <= tol].copy()
    if len(tmp) == 0:
        return None
    tmp = tmp.sort_values(["gap", "acc"], ascending=[True, False])
    best_gap = tmp.iloc[0]["gap"]
    near = tmp[tmp["gap"] <= best_gap + 1e-12].copy()
    near = near.sort_values(["acc", "ece"], ascending=[False, True])
    return near.iloc[0].to_dict()

# =========================
# Model
# =========================
class BackboneWrapper(nn.Module):
    def __init__(self, num_classes: int = 100):
        super().__init__()
        m = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
        feat_dim = m.fc.in_features
        m.fc = nn.Identity()
        self.backbone = m
        self.head = nn.Linear(feat_dim, num_classes)
        self.feat_dim = feat_dim

    def forward_features(self, x: torch.Tensor) -> torch.Tensor:
        return self.backbone(x)

    def forward(self, x: torch.Tensor):
        feat = self.forward_features(x)
        logits = self.head(feat)
        return logits, feat

class BudgetClassifier(nn.Module):
    def __init__(self, input_dim: int, hidden_dim: int = 128, num_classes: int = 3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_classes),
        )

    def forward(self, x: torch.Tensor):
        return self.net(x)

# =========================
# TTA
# =========================
@torch.no_grad()
def batch_tta_predict(model: nn.Module, images: torch.Tensor, k: int = 1):
    model.eval()
    probs_list = []
    feats_list = []

    def apply_det_tta(x: torch.Tensor, aug_id: int):
        if aug_id == 0:
            return x
        elif aug_id == 1:
            return torch.flip(x, dims=[3])
        elif aug_id == 2:
            x2 = x[:, :, :, 4:]
            return F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
        elif aug_id == 3:
            x2 = x[:, :, :, :-4]
            return F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
        elif aug_id == 4:
            x2 = x[:, :, 4:, :]
            return F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
        elif aug_id == 5:
            x2 = x[:, :, :-4, :]
            return F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
        elif aug_id == 6:
            H, W = x.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x[:, :, dh:H-dh, dw:W-dw]
            return F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
        elif aug_id == 7:
            x2 = torch.flip(x, dims=[3])
            H, W = x2.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x2[:, :, dh:H-dh, dw:W-dw]
            return F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
        return x

    with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
        for aug_id in range(k):
            x_aug = apply_det_tta(images, aug_id)
            logits, feats = model(x_aug)
            probs = F.softmax(logits, dim=1)
            probs_list.append(probs.float())
            feats_list.append(feats.float())

    probs_stack = torch.stack(probs_list, dim=0)
    feats_stack = torch.stack(feats_list, dim=0)
    avg_probs = probs_stack.mean(dim=0)
    avg_feats = feats_stack.mean(dim=0)

    if k == 1:
        disagreement = torch.zeros(images.size(0), device=images.device)
    else:
        kls = [kl_divergence_batch(probs_stack[i], avg_probs) for i in range(k)]
        disagreement = torch.stack(kls, dim=0).mean(dim=0)

    return avg_probs, avg_feats, disagreement

# =========================
# Data features / labels
# =========================
def build_feature_matrix(cache: Dict[str, np.ndarray]) -> np.ndarray:
    probs = cache["p1_probs"]
    feats = cache["p1_feats"]
    dis = cache["p4_dis"]

    probs_t = torch.from_numpy(probs)
    entropy = predictive_entropy(probs_t).numpy()[:, None]
    margin = prediction_margin(probs_t).numpy()[:, None]
    pmax = probs.max(axis=1, keepdims=True)
    disagreement = dis[:, None]

    return np.concatenate([entropy, margin, pmax, disagreement, feats], axis=1).astype(np.float32)

def sample_correct(cache: Dict[str, np.ndarray], budget: int) -> np.ndarray:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    preds = probs.argmax(axis=1)
    return (preds == labels).astype(np.float32)

def build_budget_targets(cache: Dict[str, np.ndarray], mode: str):
    lam = parse_util_lambda(mode)
    if lam is None:
        raise ValueError("UTILITY_MODE must be util_xx")

    corr1 = sample_correct(cache, 1)
    corr4 = sample_correct(cache, 4)
    corr8 = sample_correct(cache, 8)

    util = np.zeros((len(cache["labels"]), 3), dtype=np.float32)
    util[:, 0] = corr1 - lam * normalized_cost(1)
    util[:, 1] = corr4 - lam * normalized_cost(4)
    util[:, 2] = corr8 - lam * normalized_cost(8)

    util[:, 0] += 1e-6
    util[:, 1] += 5e-7
    y = util.argmax(axis=1).astype(np.int64)
    return y

# =========================
# Training helpers
# =========================
def compute_class_weights(y: np.ndarray) -> torch.Tensor:
    counts = np.bincount(y, minlength=3).astype(np.float32)
    counts[counts == 0] = 1.0
    weights = counts.sum() / counts
    weights = weights / weights.mean()
    weights = weights * EXTRA_CLASS_MULTIPLIER
    weights = weights / weights.mean()
    return torch.tensor(weights, dtype=torch.float32)

def train_head(model, loader):
    opt = torch.optim.Adam(model.head.parameters(), lr=HEAD_LR)
    model.train()
    for epoch in range(HEAD_EPOCHS):
        pbar = tqdm(loader, desc=f"Head train {epoch+1}/{HEAD_EPOCHS}", leave=False)
        for images, labels in pbar:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            opt.zero_grad()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
                logits, _ = model(images)
                loss = F.cross_entropy(logits, labels)
            loss.backward()
            opt.step()

def train_budget_classifier(X_train, y_train, X_val, y_val):
    clf = BudgetClassifier(input_dim=X_train.shape[1], hidden_dim=POLICY_HIDDEN).to(DEVICE)
    class_weights = compute_class_weights(y_train).to(DEVICE)
    opt = torch.optim.Adam(clf.parameters(), lr=POLICY_LR, weight_decay=POLICY_WEIGHT_DECAY)

    Xtr = torch.from_numpy(X_train)
    Ytr = torch.from_numpy(y_train)
    Xv = torch.from_numpy(X_val)
    Yv = torch.from_numpy(y_val)

    best_state = None
    best_val = 1e9

    for epoch in range(POLICY_EPOCHS):
        clf.train()
        perm = torch.randperm(len(Xtr))
        losses = []
        for start in range(0, len(Xtr), 1024):
            idx = perm[start:start+1024]
            xb = Xtr[idx].to(DEVICE)
            yb = Ytr[idx].to(DEVICE)

            opt.zero_grad()
            logits = clf(xb)
            loss = F.cross_entropy(logits, yb, weight=class_weights)
            loss.backward()
            opt.step()
            losses.append(float(loss.detach().cpu()))

        clf.eval()
        with torch.no_grad():
            logits = clf(Xv.to(DEVICE))
            val_loss = float(F.cross_entropy(logits, Yv.to(DEVICE)).detach().cpu())

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in clf.state_dict().items()}

    if best_state is not None:
        clf.load_state_dict(best_state)
    clf.eval()
    return clf

# =========================
# Evaluation helpers
# =========================
def evaluate_fixed(cache: Dict[str, np.ndarray], budget: int, method_name: str):
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    return {
        "method": method_name,
        "avg_compute": float(COSTS[budget]),
        "acc": top1_accuracy(probs, labels),
        "ece": compute_ece(probs, labels),
    }

def apply_budget_selection(cache: Dict[str, np.ndarray], selected_budgets: np.ndarray, name: str):
    labels = cache["labels"]
    final_probs = []
    costs = []
    for i, b in enumerate(selected_budgets.tolist()):
        final_probs.append(cache[BUDGET_TO_KEY[int(b)]][i])
        costs.append(COSTS[int(b)])
    final_probs = np.asarray(final_probs)
    return {
        "method": name,
        "avg_compute": float(np.mean(costs)),
        "acc": top1_accuracy(final_probs, labels),
        "ece": compute_ece(final_probs, labels),
    }

def tune_thresholds(values: np.ndarray, cache: Dict[str, np.ndarray], mode: str, target_compute: float):
    labels = cache["labels"]
    candidates = np.quantile(values, np.linspace(0.1, 0.9, 9))
    best = None
    best_score = -1e9

    for t1 in candidates:
        for t2 in candidates:
            if t2 < t1:
                continue
            if mode == "entropy":
                sel = np.where(values < t1, 1, np.where(values < t2, 4, 8))
            else:
                sel = np.where(values > t2, 1, np.where(values > t1, 4, 8))

            probs = np.stack([cache[BUDGET_TO_KEY[int(b)]][i] for i, b in enumerate(sel)], axis=0)
            acc = top1_accuracy(probs, labels)
            avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))
            score = acc - 10.0 * abs(avg_compute - target_compute)

            if score > best_score:
                best_score = score
                best = (sel, acc, avg_compute)
    return best

@torch.no_grad()
def evaluate_classifier_with_bias(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], bias_vec: np.ndarray):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    logits = logits + bias_vec[None, :]
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, "POTTS-CLS")

def search_best_learned_at_target(clf, X_val, X_test, val_cache, test_cache, target):
    candidates = []
    for b4 in BIAS_GRID:
        for b8 in BIAS_GRID:
            bias = np.array([0.0, b4, b8], dtype=np.float32)
            val_row = evaluate_classifier_with_bias(clf, X_val, val_cache, bias)
            gap = abs(val_row["avg_compute"] - target)
            if gap <= MATCH_TOL:
                test_row = evaluate_classifier_with_bias(clf, X_test, test_cache, bias)
                candidates.append({
                    "bias_1": 0.0,
                    "bias_4": b4,
                    "bias_8": b8,
                    "val_compute": val_row["avg_compute"],
                    "val_gap": gap,
                    "avg_compute": test_row["avg_compute"],
                    "acc": test_row["acc"],
                    "ece": test_row["ece"],
                })

    if len(candidates) == 0:
        return None

    df = pd.DataFrame(candidates)
    df = df.sort_values(["val_gap", "acc"], ascending=[True, False])
    best_gap = df.iloc[0]["val_gap"]
    df = df[df["val_gap"] <= best_gap + 1e-12].copy()
    df = df.sort_values(["acc", "ece"], ascending=[False, True])
    return df.iloc[0].to_dict()

# =========================
# Main two-seed run
# =========================
all_rows = []

for seed in SEEDS:
    print(f"\n==================== SEED {seed} ====================")
    seed_everything(seed)

    train_idx, val_idx, _ = make_subset_indices(
        n=len(train_ds_full),
        train_n=TRAIN_POLICY_SUBSET,
        val_n=VAL_SUBSET,
        test_n=TEST_SUBSET,
        seed=seed
    )

    head_train_ds = Subset(train_ds_full, train_idx)
    policy_train_ds = Subset(eval_train_ds_full, train_idx)
    val_ds = Subset(eval_train_ds_full, val_idx)
    test_ds = Subset(test_ds_full, np.arange(TEST_SUBSET))

    head_train_loader = DataLoader(head_train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
    policy_train_loader = DataLoader(policy_train_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
    test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

    model = BackboneWrapper(num_classes=NUM_CLASSES).to(DEVICE)
    for p in model.backbone.parameters():
        p.requires_grad = False

    train_head(model, head_train_loader)
    model.eval()

    train_cache = collect_policy_candidates(policy_train_loader, f"train-s{seed}")
    val_cache = collect_policy_candidates(val_loader, f"val-s{seed}")
    test_cache = collect_policy_candidates(test_loader, f"test-s{seed}")

    X_train = build_feature_matrix(train_cache)
    X_val = build_feature_matrix(val_cache)
    X_test = build_feature_matrix(test_cache)

    y_train = build_budget_targets(train_cache, UTILITY_MODE)
    y_val = build_budget_targets(val_cache, UTILITY_MODE)

    clf = train_budget_classifier(X_train, y_train, X_val, y_val)

    val_entropy = predictive_entropy(torch.from_numpy(val_cache["p1_probs"])).numpy()
    test_entropy = predictive_entropy(torch.from_numpy(test_cache["p1_probs"])).numpy()

    val_margin = prediction_margin(torch.from_numpy(val_cache["p1_probs"])).numpy()
    test_margin = prediction_margin(torch.from_numpy(test_cache["p1_probs"])).numpy()

    for target in TARGETS:
        ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=target)
        mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=target)
        learned_best = search_best_learned_at_target(clf, X_val, X_test, val_cache, test_cache, target)

        if ent_best is not None:
            ent_sel, _, _ = ent_best
            ent_test = apply_budget_selection(
                test_cache,
                np.where(test_entropy < np.inf, ent_sel, ent_sel),  # preserve array
                "Entropy-threshold"
            )
            all_rows.append({
                "seed": seed,
                "target_compute": target,
                "method": "Entropy-threshold",
                **ent_test
            })

        if mar_best is not None:
            mar_sel, _, _ = mar_best
            mar_test = apply_budget_selection(
                test_cache,
                np.where(test_margin > -np.inf, mar_sel, mar_sel),
                "Margin-threshold"
            )
            all_rows.append({
                "seed": seed,
                "target_compute": target,
                "method": "Margin-threshold",
                **mar_test
            })

        if learned_best is not None:
            all_rows.append({
                "seed": seed,
                "target_compute": target,
                "method": "POTTS-CLS",
                "avg_compute": learned_best["avg_compute"],
                "acc": learned_best["acc"],
                "ece": learned_best["ece"],
                "bias_1": learned_best["bias_1"],
                "bias_4": learned_best["bias_4"],
                "bias_8": learned_best["bias_8"],
            })

    # also store fixed baselines once per seed
    all_rows.append({"seed": seed, "target_compute": "fixed", **evaluate_fixed(test_cache, 1, "Single")})
    all_rows.append({"seed": seed, "target_compute": "fixed", **evaluate_fixed(test_cache, 4, "TTA-4")})
    all_rows.append({"seed": seed, "target_compute": "fixed", **evaluate_fixed(test_cache, 8, "TTA-8")})

# =========================
# Summaries
# =========================
all_df = pd.DataFrame(all_rows)
all_df.to_csv(os.path.join(OUTDIR, "final_results_all.csv"), index=False)

print("\n=== FINAL RESULTS ALL ===")
display(all_df)

summary_df = (
    all_df[all_df["target_compute"] != "fixed"]
    .groupby(["target_compute", "method"], as_index=False)[["avg_compute", "acc", "ece"]]
    .mean()
)
summary_df.to_csv(os.path.join(OUTDIR, "final_summary_by_seed.csv"), index=False)

print("\n=== FINAL SUMMARY BY SEED ===")
display(summary_df)

best_rows = []
for target in TARGETS:
    sub = all_df[all_df["target_compute"] == target].copy()
    if len(sub) == 0:
        continue
    for seed in SEEDS:
        sub_seed = sub[sub["seed"] == seed].copy()
        if len(sub_seed) == 0:
            continue
        heur = sub_seed[sub_seed["method"].isin(["Entropy-threshold", "Margin-threshold"])].copy()
        learned = sub_seed[sub_seed["method"] == "POTTS-CLS"].copy()
        if len(heur) and len(learned):
            heur_best = heur.sort_values(["acc", "avg_compute"], ascending=[False, True]).iloc[0]
            learned_best = learned.sort_values(["acc", "avg_compute"], ascending=[False, True]).iloc[0]
            best_rows.append({
                "seed": seed,
                "target_compute": target,
                "heuristic_method": heur_best["method"],
                "heuristic_compute": heur_best["avg_compute"],
                "heuristic_acc": heur_best["acc"],
                "heuristic_ece": heur_best["ece"],
                "learned_compute": learned_best["avg_compute"],
                "learned_acc": learned_best["acc"],
                "learned_ece": learned_best["ece"],
                "acc_delta_learned_minus_heuristic": learned_best["acc"] - heur_best["acc"],
            })

best_df = pd.DataFrame(best_rows)
best_df.to_csv(os.path.join(OUTDIR, "final_best_points.csv"), index=False)

print("\n=== FINAL BEST POINTS ===")
display(best_df)

print("\nSaved files:")
for fn in sorted(os.listdir(OUTDIR)):
    print(" -", os.path.join(OUTDIR, fn))
```

```python
# =========================================================
# ONE-CELL COLAB: FINALIZATION RUN (FIXED)
# Only run 3 target compute regions + 2 seeds
# Targets: 1.7 / 2.5 / 3.5
# Seeds: 42 / 43
#
# Purpose:
# - finalize the paper with minimal extra experiments
# - verify the learned policy at the 3 budget regions you care about
# - compare fairly against entropy-threshold and margin-threshold
#
# This version fixes the broadcast bug by reusing thresholds (t1, t2)
# rather than reusing validation selection arrays on the test set.
#
# Outputs:
# /content/potts_final_runs/final_results_all.csv
# /content/potts_final_runs/final_summary_by_seed.csv
# /content/potts_final_runs/final_best_points.csv
# =========================================================

!pip -q install timm pandas scikit-learn tqdm

import os
import random
import warnings
from typing import Dict, Optional

import numpy as np
import pandas as pd
from tqdm.auto import tqdm

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, Subset

import torchvision
import torchvision.transforms as T
from torchvision.datasets import CIFAR100

warnings.filterwarnings("ignore")

# =========================
# Global config
# =========================
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
NUM_WORKERS = 0
BATCH_SIZE = 256

TRAIN_POLICY_SUBSET = 8000
VAL_SUBSET = 1000
TEST_SUBSET = 2000

IMAGE_SIZE = 224
NUM_CLASSES = 100

HEAD_EPOCHS = 1
HEAD_LR = 1e-3

POLICY_HIDDEN = 128
POLICY_EPOCHS = 8
POLICY_LR = 1e-3
POLICY_WEIGHT_DECAY = 1e-4

BUDGETS = [1, 4, 8]
COSTS = {1: 1.0, 4: 4.0, 8: 8.0}
BUDGET_TO_KEY = {1: "p1_probs", 4: "p4_probs", 8: "p8_probs"}
CLASS_TO_BUDGET = {0: 1, 1: 4, 2: 8}

TARGETS = [1.7, 2.5, 3.5]
MATCH_TOL = 0.25
SEEDS = [42, 43]

# Use the setting that looked most promising
UTILITY_MODE = "util_0.07"
FEATURE_SET_NAME = "full"
EXTRA_CLASS_MULTIPLIER = np.array([1.0, 3.0, 4.0], dtype=np.float32)
BIAS_GRID = [-3.0, -2.0, -1.5, -1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0, 3.0]

OUTDIR = "/content/potts_final_runs"
os.makedirs(OUTDIR, exist_ok=True)

print("DEVICE =", DEVICE)

# =========================
# Utilities
# =========================
def seed_everything(seed: int):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

def compute_ece(probs: np.ndarray, labels: np.ndarray, n_bins: int = 15) -> float:
    confidences = probs.max(axis=1)
    predictions = probs.argmax(axis=1)
    accuracies = (predictions == labels).astype(np.float32)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    ece = 0.0
    for i in range(n_bins):
        mask = (confidences > bins[i]) & (confidences <= bins[i + 1])
        if mask.any():
            bin_acc = accuracies[mask].mean()
            bin_conf = confidences[mask].mean()
            ece += np.abs(bin_acc - bin_conf) * mask.mean()
    return float(ece)

def top1_accuracy(probs: np.ndarray, labels: np.ndarray) -> float:
    return float((probs.argmax(axis=1) == labels).mean() * 100.0)

def predictive_entropy(probs: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return -(probs * (probs + eps).log()).sum(dim=1)

def prediction_margin(probs: torch.Tensor) -> torch.Tensor:
    top2 = torch.topk(probs, k=2, dim=1).values
    return top2[:, 0] - top2[:, 1]

def kl_divergence_batch(p: torch.Tensor, q: torch.Tensor, eps: float = 1e-12) -> torch.Tensor:
    return (p * ((p + eps).log() - (q + eps).log())).sum(dim=1)

def make_subset_indices(n: int, train_n: int, val_n: int, test_n: int, seed: int):
    idx = np.arange(n)
    rng = np.random.default_rng(seed)
    rng.shuffle(idx)
    train_idx = idx[:train_n]
    val_idx = idx[train_n:train_n + val_n]
    test_idx = idx[train_n + val_n:train_n + val_n + test_n]
    return train_idx, val_idx, test_idx

def parse_util_lambda(mode: str) -> Optional[float]:
    if mode.startswith("util_"):
        return float(mode.split("_")[1])
    return None

def normalized_cost(budget: int) -> float:
    return (budget - 1.0) / 7.0

# =========================
# Model
# =========================
class BackboneWrapper(nn.Module):
    def __init__(self, num_classes: int = 100):
        super().__init__()
        m = torchvision.models.resnet50(weights=torchvision.models.ResNet50_Weights.IMAGENET1K_V2)
        feat_dim = m.fc.in_features
        m.fc = nn.Identity()
        self.backbone = m
        self.head = nn.Linear(feat_dim, num_classes)
        self.feat_dim = feat_dim

    def forward_features(self, x: torch.Tensor) -> torch.Tensor:
        return self.backbone(x)

    def forward(self, x: torch.Tensor):
        feat = self.forward_features(x)
        logits = self.head(feat)
        return logits, feat

class BudgetClassifier(nn.Module):
    def __init__(self, input_dim: int, hidden_dim: int = 128, num_classes: int = 3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, num_classes),
        )

    def forward(self, x: torch.Tensor):
        return self.net(x)

# =========================
# TTA
# =========================
@torch.no_grad()
def batch_tta_predict(model: nn.Module, images: torch.Tensor, k: int = 1):
    model.eval()
    probs_list = []
    feats_list = []

    def apply_det_tta(x: torch.Tensor, aug_id: int):
        if aug_id == 0:
            return x
        elif aug_id == 1:
            return torch.flip(x, dims=[3])
        elif aug_id == 2:
            x2 = x[:, :, :, 4:]
            return F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
        elif aug_id == 3:
            x2 = x[:, :, :, :-4]
            return F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
        elif aug_id == 4:
            x2 = x[:, :, 4:, :]
            return F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
        elif aug_id == 5:
            x2 = x[:, :, :-4, :]
            return F.interpolate(x2, size=x.shape[-2:], mode="bilinear", align_corners=False)
        elif aug_id == 6:
            H, W = x.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x[:, :, dh:H-dh, dw:W-dw]
            return F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
        elif aug_id == 7:
            x2 = torch.flip(x, dims=[3])
            H, W = x2.shape[-2:]
            dh, dw = int(H * 0.05), int(W * 0.05)
            x2 = x2[:, :, dh:H-dh, dw:W-dw]
            return F.interpolate(x2, size=(H, W), mode="bilinear", align_corners=False)
        return x

    with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
        for aug_id in range(k):
            x_aug = apply_det_tta(images, aug_id)
            logits, feats = model(x_aug)
            probs = F.softmax(logits, dim=1)
            probs_list.append(probs.float())
            feats_list.append(feats.float())

    probs_stack = torch.stack(probs_list, dim=0)
    feats_stack = torch.stack(feats_list, dim=0)
    avg_probs = probs_stack.mean(dim=0)
    avg_feats = feats_stack.mean(dim=0)

    if k == 1:
        disagreement = torch.zeros(images.size(0), device=images.device)
    else:
        kls = [kl_divergence_batch(probs_stack[i], avg_probs) for i in range(k)]
        disagreement = torch.stack(kls, dim=0).mean(dim=0)

    return avg_probs, avg_feats, disagreement

# =========================
# Data features / labels
# =========================
def build_feature_matrix(cache: Dict[str, np.ndarray]) -> np.ndarray:
    probs = cache["p1_probs"]
    feats = cache["p1_feats"]
    dis = cache["p4_dis"]

    probs_t = torch.from_numpy(probs)
    entropy = predictive_entropy(probs_t).numpy()[:, None]
    margin = prediction_margin(probs_t).numpy()[:, None]
    pmax = probs.max(axis=1, keepdims=True)
    disagreement = dis[:, None]

    return np.concatenate([entropy, margin, pmax, disagreement, feats], axis=1).astype(np.float32)

def sample_correct(cache: Dict[str, np.ndarray], budget: int) -> np.ndarray:
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    preds = probs.argmax(axis=1)
    return (preds == labels).astype(np.float32)

def build_budget_targets(cache: Dict[str, np.ndarray], mode: str):
    lam = parse_util_lambda(mode)
    if lam is None:
        raise ValueError("UTILITY_MODE must be util_xx")

    corr1 = sample_correct(cache, 1)
    corr4 = sample_correct(cache, 4)
    corr8 = sample_correct(cache, 8)

    util = np.zeros((len(cache["labels"]), 3), dtype=np.float32)
    util[:, 0] = corr1 - lam * normalized_cost(1)
    util[:, 1] = corr4 - lam * normalized_cost(4)
    util[:, 2] = corr8 - lam * normalized_cost(8)

    util[:, 0] += 1e-6
    util[:, 1] += 5e-7
    y = util.argmax(axis=1).astype(np.int64)
    return y

# =========================
# Training helpers
# =========================
def compute_class_weights(y: np.ndarray) -> torch.Tensor:
    counts = np.bincount(y, minlength=3).astype(np.float32)
    counts[counts == 0] = 1.0
    weights = counts.sum() / counts
    weights = weights / weights.mean()
    weights = weights * EXTRA_CLASS_MULTIPLIER
    weights = weights / weights.mean()
    return torch.tensor(weights, dtype=torch.float32)

def train_head(model, loader):
    opt = torch.optim.Adam(model.head.parameters(), lr=HEAD_LR)
    model.train()
    for epoch in range(HEAD_EPOCHS):
        pbar = tqdm(loader, desc=f"Head train {epoch+1}/{HEAD_EPOCHS}", leave=False)
        for images, labels in pbar:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            opt.zero_grad()
            with torch.autocast(device_type="cuda", dtype=torch.float16, enabled=torch.cuda.is_available()):
                logits, _ = model(images)
                loss = F.cross_entropy(logits, labels)
            loss.backward()
            opt.step()

def train_budget_classifier(X_train, y_train, X_val, y_val):
    clf = BudgetClassifier(input_dim=X_train.shape[1], hidden_dim=POLICY_HIDDEN).to(DEVICE)
    class_weights = compute_class_weights(y_train).to(DEVICE)
    opt = torch.optim.Adam(clf.parameters(), lr=POLICY_LR, weight_decay=POLICY_WEIGHT_DECAY)

    Xtr = torch.from_numpy(X_train)
    Ytr = torch.from_numpy(y_train)
    Xv = torch.from_numpy(X_val)
    Yv = torch.from_numpy(y_val)

    best_state = None
    best_val = 1e9

    for epoch in range(POLICY_EPOCHS):
        clf.train()
        perm = torch.randperm(len(Xtr))
        for start in range(0, len(Xtr), 1024):
            idx = perm[start:start+1024]
            xb = Xtr[idx].to(DEVICE)
            yb = Ytr[idx].to(DEVICE)

            opt.zero_grad()
            logits = clf(xb)
            loss = F.cross_entropy(logits, yb, weight=class_weights)
            loss.backward()
            opt.step()

        clf.eval()
        with torch.no_grad():
            logits = clf(Xv.to(DEVICE))
            val_loss = float(F.cross_entropy(logits, Yv.to(DEVICE)).detach().cpu())

        if val_loss < best_val:
            best_val = val_loss
            best_state = {k: v.detach().cpu().clone() for k, v in clf.state_dict().items()}

    if best_state is not None:
        clf.load_state_dict(best_state)
    clf.eval()
    return clf

# =========================
# Evaluation helpers
# =========================
def evaluate_fixed(cache: Dict[str, np.ndarray], budget: int, method_name: str):
    probs = cache[BUDGET_TO_KEY[budget]]
    labels = cache["labels"]
    return {
        "method": method_name,
        "avg_compute": float(COSTS[budget]),
        "acc": top1_accuracy(probs, labels),
        "ece": compute_ece(probs, labels),
    }

def apply_budget_selection(cache: Dict[str, np.ndarray], selected_budgets: np.ndarray, name: str):
    labels = cache["labels"]
    final_probs = []
    costs = []
    for i, b in enumerate(selected_budgets.tolist()):
        final_probs.append(cache[BUDGET_TO_KEY[int(b)]][i])
        costs.append(COSTS[int(b)])
    final_probs = np.asarray(final_probs)
    return {
        "method": name,
        "avg_compute": float(np.mean(costs)),
        "acc": top1_accuracy(final_probs, labels),
        "ece": compute_ece(final_probs, labels),
    }

def tune_thresholds(values: np.ndarray, cache: Dict[str, np.ndarray], mode: str, target_compute: float):
    labels = cache["labels"]
    candidates = np.quantile(values, np.linspace(0.1, 0.9, 9))
    best = None
    best_score = -1e9

    for t1 in candidates:
        for t2 in candidates:
            if t2 < t1:
                continue
            if mode == "entropy":
                sel = np.where(values < t1, 1, np.where(values < t2, 4, 8))
            else:
                sel = np.where(values > t2, 1, np.where(values > t1, 4, 8))

            probs = np.stack([cache[BUDGET_TO_KEY[int(b)]][i] for i, b in enumerate(sel)], axis=0)
            acc = top1_accuracy(probs, labels)
            avg_compute = float(np.mean([COSTS[int(b)] for b in sel]))
            score = acc - 10.0 * abs(avg_compute - target_compute)

            if score > best_score:
                best_score = score
                best = (t1, t2, sel, acc, avg_compute)
    return best

@torch.no_grad()
def evaluate_classifier_with_bias(clf: nn.Module, X: np.ndarray, cache: Dict[str, np.ndarray], bias_vec: np.ndarray):
    logits = clf(torch.from_numpy(X).to(DEVICE)).cpu().numpy()
    logits = logits + bias_vec[None, :]
    pred_cls = logits.argmax(axis=1)
    sel = np.array([CLASS_TO_BUDGET[int(c)] for c in pred_cls])
    return apply_budget_selection(cache, sel, "POTTS-CLS")

def search_best_learned_at_target(clf, X_val, X_test, val_cache, test_cache, target):
    candidates = []
    for b4 in BIAS_GRID:
        for b8 in BIAS_GRID:
            bias = np.array([0.0, b4, b8], dtype=np.float32)
            val_row = evaluate_classifier_with_bias(clf, X_val, val_cache, bias)
            gap = abs(val_row["avg_compute"] - target)
            if gap <= MATCH_TOL:
                test_row = evaluate_classifier_with_bias(clf, X_test, test_cache, bias)
                candidates.append({
                    "bias_1": 0.0,
                    "bias_4": b4,
                    "bias_8": b8,
                    "val_compute": val_row["avg_compute"],
                    "val_gap": gap,
                    "avg_compute": test_row["avg_compute"],
                    "acc": test_row["acc"],
                    "ece": test_row["ece"],
                })

    if len(candidates) == 0:
        return None

    df = pd.DataFrame(candidates)
    df = df.sort_values(["val_gap", "acc"], ascending=[True, False])
    best_gap = df.iloc[0]["val_gap"]
    df = df[df["val_gap"] <= best_gap + 1e-12].copy()
    df = df.sort_values(["acc", "ece"], ascending=[False, True])
    return df.iloc[0].to_dict()

# =========================
# Cache collector
# =========================
@torch.no_grad()
def collect_policy_candidates(loader: DataLoader, name: str, model: nn.Module) -> Dict[str, np.ndarray]:
    model.eval()
    out = {
        "labels": [],
        "p1_probs": [], "p4_probs": [], "p8_probs": [],
        "p1_feats": [], "p4_feats": [], "p8_feats": [],
        "p1_dis": [], "p4_dis": [], "p8_dis": [],
    }

    for images, labels in tqdm(loader, desc=f"Collect [{name}]", leave=False):
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        probs1, feat1, dis1 = batch_tta_predict(model, images, k=1)
        probs4, feat4, dis4 = batch_tta_predict(model, images, k=4)
        probs8, feat8, dis8 = batch_tta_predict(model, images, k=8)

        out["labels"].append(labels.cpu().numpy())
        out["p1_probs"].append(probs1.cpu().numpy())
        out["p4_probs"].append(probs4.cpu().numpy())
        out["p8_probs"].append(probs8.cpu().numpy())
        out["p1_feats"].append(feat1.cpu().numpy())
        out["p4_feats"].append(feat4.cpu().numpy())
        out["p8_feats"].append(feat8.cpu().numpy())
        out["p1_dis"].append(dis1.cpu().numpy())
        out["p4_dis"].append(dis4.cpu().numpy())
        out["p8_dis"].append(dis8.cpu().numpy())

    for k in list(out.keys()):
        out[k] = np.concatenate(out[k], axis=0)
    return out

# =========================
# Main two-seed run
# =========================
all_rows = []

for seed in SEEDS:
    print(f"\n==================== SEED {seed} ====================")
    seed_everything(seed)

    if torch.cuda.is_available():
        torch.backends.cudnn.benchmark = True

    mean = (0.5071, 0.4867, 0.4408)
    std = (0.2675, 0.2565, 0.2761)

    base_transform = T.Compose([
        T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
        T.ToTensor(),
        T.Normalize(mean, std),
    ])

    train_transform = T.Compose([
        T.Resize((IMAGE_SIZE, IMAGE_SIZE)),
        T.RandomHorizontalFlip(),
        T.RandomCrop(IMAGE_SIZE, padding=16),
        T.ToTensor(),
        T.Normalize(mean, std),
    ])

    train_ds_full = CIFAR100(root="/content/data", train=True, download=True, transform=train_transform)
    eval_train_ds_full = CIFAR100(root="/content/data", train=True, download=False, transform=base_transform)
    test_ds_full = CIFAR100(root="/content/data", train=False, download=True, transform=base_transform)

    train_idx, val_idx, _ = make_subset_indices(
        n=len(train_ds_full),
        train_n=TRAIN_POLICY_SUBSET,
        val_n=VAL_SUBSET,
        test_n=TEST_SUBSET,
        seed=seed
    )

    head_train_ds = Subset(train_ds_full, train_idx)
    policy_train_ds = Subset(eval_train_ds_full, train_idx)
    val_ds = Subset(eval_train_ds_full, val_idx)
    test_ds = Subset(test_ds_full, np.arange(TEST_SUBSET))

    head_train_loader = DataLoader(head_train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
    policy_train_loader = DataLoader(policy_train_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
    val_loader = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
    test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

    model = BackboneWrapper(num_classes=NUM_CLASSES).to(DEVICE)
    for p in model.backbone.parameters():
        p.requires_grad = False

    train_head(model, head_train_loader)
    model.eval()

    train_cache = collect_policy_candidates(policy_train_loader, f"train-s{seed}", model)
    val_cache = collect_policy_candidates(val_loader, f"val-s{seed}", model)
    test_cache = collect_policy_candidates(test_loader, f"test-s{seed}", model)

    print(f"Seed {seed} sanity: Single={top1_accuracy(test_cache['p1_probs'], test_cache['labels']):.2f}, "
          f"TTA-4={top1_accuracy(test_cache['p4_probs'], test_cache['labels']):.2f}, "
          f"TTA-8={top1_accuracy(test_cache['p8_probs'], test_cache['labels']):.2f}")

    X_train = build_feature_matrix(train_cache)
    X_val = build_feature_matrix(val_cache)
    X_test = build_feature_matrix(test_cache)

    y_train = build_budget_targets(train_cache, UTILITY_MODE)
    y_val = build_budget_targets(val_cache, UTILITY_MODE)

    clf = train_budget_classifier(X_train, y_train, X_val, y_val)

    val_entropy = predictive_entropy(torch.from_numpy(val_cache["p1_probs"])).numpy()
    test_entropy = predictive_entropy(torch.from_numpy(test_cache["p1_probs"])).numpy()

    val_margin = prediction_margin(torch.from_numpy(val_cache["p1_probs"])).numpy()
    test_margin = prediction_margin(torch.from_numpy(test_cache["p1_probs"])).numpy()

    for target in TARGETS:
        ent_best = tune_thresholds(val_entropy, val_cache, mode="entropy", target_compute=target)
        mar_best = tune_thresholds(val_margin, val_cache, mode="margin", target_compute=target)
        learned_best = search_best_learned_at_target(clf, X_val, X_test, val_cache, test_cache, target)

        if ent_best is not None:
            t1, t2, _, _, _ = ent_best
            ent_sel_test = np.where(test_entropy < t1, 1, np.where(test_entropy < t2, 4, 8))
            ent_test = apply_budget_selection(test_cache, ent_sel_test, "Entropy-threshold")
            all_rows.append({
                "seed": seed,
                "target_compute": target,
                "method": "Entropy-threshold",
                **ent_test
            })

        if mar_best is not None:
            t1m, t2m, _, _, _ = mar_best
            mar_sel_test = np.where(test_margin > t2m, 1, np.where(test_margin > t1m, 4, 8))
            mar_test = apply_budget_selection(test_cache, mar_sel_test, "Margin-threshold")
            all_rows.append({
                "seed": seed,
                "target_compute": target,
                "method": "Margin-threshold",
                **mar_test
            })

        if learned_best is not None:
            all_rows.append({
                "seed": seed,
                "target_compute": target,
                "method": "POTTS-CLS",
                "avg_compute": learned_best["avg_compute"],
                "acc": learned_best["acc"],
                "ece": learned_best["ece"],
                "bias_1": learned_best["bias_1"],
                "bias_4": learned_best["bias_4"],
                "bias_8": learned_best["bias_8"],
            })

    all_rows.append({"seed": seed, "target_compute": "fixed", **evaluate_fixed(test_cache, 1, "Single")})
    all_rows.append({"seed": seed, "target_compute": "fixed", **evaluate_fixed(test_cache, 4, "TTA-4")})
    all_rows.append({"seed": seed, "target_compute": "fixed", **evaluate_fixed(test_cache, 8, "TTA-8")})

# =========================
# Summaries
# =========================
all_df = pd.DataFrame(all_rows)
all_df.to_csv(os.path.join(OUTDIR, "final_results_all.csv"), index=False)

print("\n=== FINAL RESULTS ALL ===")
display(all_df)

summary_df = (
    all_df[all_df["target_compute"] != "fixed"]
    .groupby(["target_compute", "method"], as_index=False)[["avg_compute", "acc", "ece"]]
    .mean()
)
summary_df.to_csv(os.path.join(OUTDIR, "final_summary_by_seed.csv"), index=False)

print("\n=== FINAL SUMMARY BY SEED ===")
display(summary_df)

best_rows = []
for target in TARGETS:
    sub = all_df[all_df["target_compute"] == target].copy()
    if len(sub) == 0:
        continue
    for seed in SEEDS:
        sub_seed = sub[sub["seed"] == seed].copy()
        if len(sub_seed) == 0:
            continue
        heur = sub_seed[sub_seed["method"].isin(["Entropy-threshold", "Margin-threshold"])].copy()
        learned = sub_seed[sub_seed["method"] == "POTTS-CLS"].copy()
        if len(heur) and len(learned):
            heur_best = heur.sort_values(["acc", "avg_compute"], ascending=[False, True]).iloc[0]
            learned_best = learned.sort_values(["acc", "avg_compute"], ascending=[False, True]).iloc[0]
            best_rows.append({
                "seed": seed,
                "target_compute": target,
                "heuristic_method": heur_best["method"],
                "heuristic_compute": heur_best["avg_compute"],
                "heuristic_acc": heur_best["acc"],
                "heuristic_ece": heur_best["ece"],
                "learned_compute": learned_best["avg_compute"],
                "learned_acc": learned_best["acc"],
                "learned_ece": learned_best["ece"],
                "acc_delta_learned_minus_heuristic": learned_best["acc"] - heur_best["acc"],
            })

best_df = pd.DataFrame(best_rows)
best_df.to_csv(os.path.join(OUTDIR, "final_best_points.csv"), index=False)

print("\n=== FINAL BEST POINTS ===")
display(best_df)

print("\nSaved files:")
for fn in sorted(os.listdir(OUTDIR)):
    print(" -", os.path.join(OUTDIR, fn))
```
